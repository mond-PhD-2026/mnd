**Мета**: спроєктувати робочу програму власного дисертаційного дослідження, перетворивши сформовані в ЛР1--ЛР4 дослідницькі питання, твердження про внесок і положення про наукову новизну на систему взаємопов'язаних досліджень із визначеними методами, даними, базовими рівнями порівняння, контрольними умовами та експериментальним/аналітичним планом.


::: callout-note
## Примітка

Попередньо опрацювати матеріали **Лекції 5 «Дизайн дослідження та система методів»**.

ЛР5 продовжує наскрізну роботу над *власною дисертацією*. Основними вхідними артефактами є результати ЛР4:

- `Gap Claim v1`;
- `Contribution Map v1`;
- `Novelty Claims v1`;
- матриця логічної узгодженості;
- `Methodological Architecture v2`.

У цій роботі ми не виконуємо повний дисертаційний експеримент. Завдання полягає в іншому -- *спроєктувати дослідження так, щоб майбутні результати могли стати релевантними доказами для конкретних наукових тверджень*.

Магістральний перехід ЛР5:

$$
\text{Claim}
\rightarrow
\text{Research Design}
\rightarrow
\text{Study}
\rightarrow
\text{Method}
\rightarrow
\text{Data}
\rightarrow
\text{Analysis}
\rightarrow
\text{Evidence}.
$$
:::


## Що ви будете вміти?

Після виконання лабораторної роботи ви зможете:

* переводити дослідницьке питання та наукове твердження у відповідний дизайн дослідження (research design);
* розрізняти опис, асоціацію, прогнозування, пояснення та причинне виведення;
* проєктувати дисертаційну програму як систему окремих досліджень (Study Units);
* обґрунтовувати вибір теоретичних, експериментальних, модельних та обчислювальних методів;
* операціоналізувати ключові поняття власного дослідження;
* визначати популяцію, вибірку, одиницю аналізу та стратегію поділу даних;
* добирати базові рівні порівняння (baselines) і контрольні умови;
* формувати план роботи з даними (data plan);
* будувати експериментальний протокол (experimental protocol) та аналітичний план (analysis plan);
* перевіряти простежуваність ланцюга `Gap → RQ → Claim → Study → Evidence`.


## Короткі теоретичні відомості


### 1. Від наукового твердження до дизайну дослідження

Дизайн дослідження не є переліком алгоритмів або програмних засобів. Він визначає логіку, за якою конкретне дослідницьке питання пов'язується з даними, методами, аналізом і майбутнім доказом.

$$
\text{Research Question}
\rightarrow
\text{Scientific Claim}
\rightarrow
\text{Research Design}
\rightarrow
\text{Evidence}.
$$

Тому спочатку потрібно визначити, *що саме ми плануємо встановити*, і лише після цього -- *який метод дозволяє це встановити*.



### 2. Тип твердження визначає тип доказу

Перед вибором методу потрібно класифікувати основну дослідницьку задачу.

| Тип задачі | Основне питання | Типовий результат |
|---|---|---|
| *Опис* (description) | Що спостерігається? | структура, розподіл, патерн |
| *Асоціація* (association) | Що з чим пов'язано? | статистичний або структурний зв'язок |
| *Прогнозування* (prediction) | Наскільки добре можна передбачити $Y$? | прогнозна якість |
| *Пояснення* (explanation) | Який механізм або структура пояснює явище? | пояснювальна модель |
| *Причинне виведення* (causal inference) | Що зміниться в $Y$, якщо змінити $X$? | оцінка причинного ефекту |

Сила майбутнього Claim не повинна перевищувати силу дизайну, який використано для його перевірки.


### 3. Study Unit як одиниця дослідницької програми

Для дисертації доцільно проєктувати не один великий експеримент, а систему взаємопов'язаних досліджень.

Кожний *Study Unit* повинен мати:

1. Research Question;
2. Claim;
3. тип дослідження;
4. дані;
5. метод;
6. baseline / control;
7. аналітичну процедуру;
8. очікуваний Evidence;
9. межі допустимого висновку.

Один Claim може потребувати кількох Study Units, а один Study Unit може давати докази для кількох пов'язаних тверджень -- але ці зв'язки мають бути явними.


### 4. Дані та порівняння є частиною дизайну

Стратегія вибірки та поділу даних повинна відповідати майбутньому твердженню про узагальнення.

Наприклад:

| Майбутній Claim | Релевантний принцип перевірки |
|---|---|
| узагальнення на нові об'єкти | незалежний test set |
| узагальнення на нових користувачів | group split |
| прогноз майбутнього | temporal split |
| переносимість між доменами | cross-domain split |
| переносимість між джерелами | leave-one-source-out |

Так само baseline потрібний не для формального «порівняння з чимось», а для визначення *додаткового ефекту саме запропонованого внеску*.


### 5. Експериментальний та аналітичний план

Експериментальний протокол фіксує *що і за яких умов робимо*.

Аналітичний план визначає *як результати перетворюються на доказ*.

Мінімальний ланцюг:

$$
\text{Data}
\rightarrow
\text{Procedure}
\rightarrow
\text{Measurement}
\rightarrow
\text{Analysis}
\rightarrow
\text{Evidence}
\rightarrow
\text{Claim}.
$$

У наступній ЛР6 ця конструкція буде розширена через валідацію, робастність, аналіз чутливості та загрози валідності.


## Хід роботи


### Завдання 1. Зафіксуйте вхідну методологічну архітектуру

Перенесіть актуальні результати ЛР4.

| Артефакт | Актуальна версія | Короткий зміст |
|---|---|---|
| Research Problem | ... | |
| Gap Claim | v1 | |
| Research Questions | ... | |
| Research Tasks | ... | |
| Contribution Claims | ... | |
| Novelty Claims | v1 | |
| Methodological Architecture | v2 | |

Окремо вкажіть, які елементи після `Checkpoint Review 1` були змінені та чому.


### Завдання 2. Класифікуйте дослідницькі питання за типом задачі

Для кожного основного RQ визначте, яку логіку доказу воно передбачає.

| RQ ID | Research Question | Тип: опис / асоціація / прогнозування / пояснення / причинність | Обґрунтування |
|---|---|---|---|
| RQ1 | | | |
| RQ2 | | | |
| RQ3 | | | |

Якщо одне питання поєднує кілька принципово різних типів, перевірте, чи не доцільно розділити його на підпитання.


### Завдання 3. Побудуйте `Claim -- Evidence Requirements Matrix`

Для кожного ключового Claim визначте, *який тип доказу був би достатнім для його підтримки*.

| Claim ID | Claim | Тип твердження | Який Evidence потрібний? | Що не буде достатнім доказом? |
|---|---|---|---|---|
| C1 | | | | |
| C2 | | | | |
| C3 | | | | |

Особливо важливо заповнити останню колонку. Вона обмежує можливість зробити висновок сильнішим, ніж дозволяє дизайн.


### Завдання 4. Спроєктуйте систему `Study Units`

Розбийте дисертаційну програму на окремі дослідження.

| Study ID | RQ | Claim | Мета Study | Тип дослідження | Очікуваний Evidence |
|---|---|---|---|---|---|
| S1 | | | | | |
| S2 | | | | | |
| S3 | | | | | |

Тип дослідження може бути:

- теоретичним;
- експериментальним;
- модельним;
- обчислювальним;
- комбінованим.

Не намагайтеся штучно створити велику кількість Study Units. Їх має бути стільки, скільки потрібно для логічного покриття основних Claims.


### Завдання 5. Виконайте операціоналізацію ключових конструктів

Виберіть 3--6 центральних понять дисертації, які не спостерігаються безпосередньо.

| Construct | Робоче означення | Operational Definition | Змінна / індикатор | Процедура вимірювання | Обмеження |
|---|---|---|---|---|---|
| | | | | | |
| | | | | | |
| | | | | | |

Приклади конструктів: *стійкість, пояснюваність, ботоподібність, ситуаційна обізнаність, якість взаємодії, семантична подібність*.

Перевірте, чи справді обрана метрика вимірює потрібний конструкт, а не лише зручну його проксі-ознаку.


### Завдання 6. Сформуйте `Data Plan v1`

Для кожного Study Unit опишіть дані.

| Study | Джерело | Популяція | Вибірка | Одиниця аналізу | Основні змінні | Мітки / ground truth | Потенційний bias |
|---|---|---|---|---|---|---|---|
| S1 | | | | | | | |
| S2 | | | | | | | |

Далі зафіксуйте план руху даних:

$$
\text{Source}
\rightarrow
\text{Acquisition}
\rightarrow
\text{Cleaning}
\rightarrow
\text{Transformation}
\rightarrow
\text{Split}
\rightarrow
\text{Analysis}.
$$

Для кожного переходу коротко вкажіть, що саме змінюється та який ризик зміщення або витоку інформації може виникнути.


### Завдання 7. Обґрунтуйте стратегію вибірки та поділу даних

Для кожного Study Unit заповніть таблицю.

| Study | Generalization Claim | Стратегія вибірки | Train / Validation / Test або інший поділ | Чому цей поділ відповідає Claim? | Ризик leakage |
|---|---|---|---|---|---|
| S1 | | | | | |
| S2 | | | | | |

Якщо машинне навчання у вашому дослідженні не використовується, адаптуйте колонку поділу даних до відповідної процедури формування порівнюваних груп, сценаріїв, симуляцій або наборів спостережень.


### Завдання 8. Визначте методи, baselines та контрольні умови

Не починайте з переліку бібліотек або алгоритмів. Для кожного Study Unit покажіть функцію методу в логіці доказу.

| Study | Метод / підхід | Чому він відповідає RQ? | Baseline | Control condition | Що саме порівнюємо? |
|---|---|---|---|---|---|
| S1 | | | | | |
| S2 | | | | | |

Для кожного baseline вкажіть його роль:

- наївна точка відліку;
- класичний метод;
- сильний сучасний метод;
- попередня версія власного підходу;
- контроль без ключового компонента;
- інша змістовно обґрунтована точка порівняння.


### Завдання 9. Побудуйте `Experimental Protocol v1`

Для одного центрального Study Unit сформуйте повний протокол.

| Компонент | Ваше рішення | Обґрунтування |
|---|---|---|
| Study ID | | |
| Research Question | | |
| Claim / Hypothesis | | |
| Дані | | |
| Фактори | | |
| Залежні змінні | | |
| Контрольовані змінні | | |
| Потенційні confounders | | |
| Baselines | | |
| Control conditions | | |
| Процедура split / sampling | | |
| Preprocessing | | |
| Метод / модель | | |
| Налаштування / tuning | | |
| Кількість повторень | | |
| Основні метрики | | |
| Додаткові метрики | | |
| Умова підтримки Claim | | |
| Умова, за якої Claim не підтримується | | |

Якщо ваше дослідження не є експериментальним у вузькому сенсі, адаптуйте протокол до теоретичного, модельного або обчислювального Study, зберігаючи логіку `Claim → Procedure → Evidence`.


### Завдання 10. Сформуйте `Analysis Plan v1`

Для того самого центрального Study Unit визначте, як результати будуть перетворені на Evidence.

| Питання | Рішення |
|---|---|
| Які результати є первинними? | |
| Які порівняння є основними? | |
| Як агрегуються повторення? | |
| Як оцінюється невизначеність? | |
| Як обробляються пропуски? | |
| Як обробляються викиди? | |
| Які припущення потрібно перевірити? | |
| Який результат підтримує Claim? | |
| Який результат суперечить Claim? | |
| Як інтерпретується нульовий результат? | |
| Які аналізи є exploratory? | |
| Які аналізи є confirmatory? | |

На цьому етапі не потрібно детально розробляти систему валідації -- це предмет ЛР6. Але логіка основного аналізу має бути визначена вже зараз.


### Завдання 11. Побудуйте матрицю `Claim -- Study -- Data -- Method -- Evidence`

Інтегруйте результати попередніх завдань.

| Claim | RQ | Study | Data | Method | Baseline / Control | Analysis | Expected Evidence |
|---|---|---|---|---|---|---|---|
| C1 | | | | | | | |
| C2 | | | | | | | |
| C3 | | | | | | | |

Перевірте:

1. чи кожний основний Claim має хоча б один Study;
2. чи кожний Study підтримує конкретний Claim;
3. чи відповідають дані силі Claim;
4. чи є порівняння, яке дозволяє інтерпретувати ефект;
5. чи можна з результату запланованого аналізу отримати саме той Evidence, який потрібний.


### Завдання 12. Проведіть аудит альтернативних пояснень

Для кожного центрального Claim вкажіть щонайменше 2--3 альтернативні пояснення майбутнього позитивного результату.

| Claim | Альтернативне пояснення | Як дизайн його контролює? | Залишковий ризик |
|---|---|---|---|
| C1 | | | |
| C1 | | | |
| C2 | | | |

Наприклад, якщо нова модель показує кращий результат, можливими альтернативними поясненнями можуть бути:

- більша кількість параметрів;
- інший preprocessing;
- більший обсяг навчальних даних;
- інтенсивніший tuning;
- витік інформації;
- особливість конкретного dataset.

Це завдання готує перехід до систематичного аналізу загроз валідності в ЛР6.


### Завдання 13. Сформуйте `Research Design v1`

Зберіть результати роботи у компактну архітектуру.

```{mermaid}
%%| fig-cap: "Шаблон Research Design v1"
%%{init: {'theme': 'base', 'themeVariables': {
  'fontSize': '17px',
  'fontFamily': 'Trebuchet MS, Verdana, Arial, sans-serif',
  'primaryColor': '#E8F3FF',
  'primaryBorderColor': '#3478A8',
  'secondaryColor': '#FFF0C9',
  'tertiaryColor': '#E7F6EC',
  'lineColor': '#536878'
}}}%%
flowchart LR
    G("🧩 Gap") --> RQ("❓ RQ")
    RQ --> C("💬 Claim")
    C --> S1("🧪 Study S1")
    C --> S2("💻 Study S2")
    S1 --> D1("🗃️ Data")
    S2 --> D2("🗃️ Data")
    S1 --> M1("🛠️ Method")
    S2 --> M2("🛠️ Method")
    D1 --> E("🔎 Evidence")
    M1 --> E
    D2 --> E
    M2 --> E
    E --> V("✅ Validation<br/>у ЛР6")
```



## Контрольні питання

1. Чим дизайн дослідження відрізняється від переліку методів?
2. Чому Research Question має передувати вибору алгоритму?
3. Чим опис відрізняється від асоціації?
4. Чому прогнозування не є автоматично поясненням?
5. Чому асоціативний дизайн не дає достатніх підстав для причинного Claim?
6. Що таке Study Unit?
7. Навіщо операціоналізувати теоретичні конструкти?
8. Чим популяція відрізняється від вибірки?
9. Що таке одиниця аналізу?
10. Чому стратегія split залежить від Generalization Claim?
11. Для чого потрібний baseline?
12. Чим baseline відрізняється від control condition?
13. Що таке confounder?
14. Які компоненти повинен містити експериментальний протокол?
15. Чим experimental protocol відрізняється від analysis plan?
16. Чому exploratory та confirmatory analyses потрібно розрізняти?
17. Що означає `Claim -- Evidence Requirements`?
18. Навіщо аналізувати альтернативні пояснення ще до проведення експерименту?
19. Як перевірити простежуваність `Gap → RQ → Claim → Study → Evidence`?
20. Які елементи Research Design v1 потребуватимуть валідації у ЛР6?


## Вимоги до звіту

Звіт має містити:

1. актуальну методологічну архітектуру після ЛР4;
2. класифікацію RQ за типом дослідницької задачі;
3. `Claim -- Evidence Requirements Matrix`;
4. систему `Study Units v1`;
5. операціоналізацію ключових конструктів;
6. `Data Plan v1`;
7. обґрунтування вибірки та стратегії поділу даних;
8. таблицю методів, baselines і контрольних умов;
9. `Experimental Protocol v1`;
10. `Analysis Plan v1`;
11. матрицю `Claim -- Study -- Data -- Method -- Evidence`;
12. аудит альтернативних пояснень;
13. схему `Research Design v1`;
14. короткий висновок про те, *які саме докази повинна породити запроєктована система досліджень*.

Таблиці можуть бути розширені відповідно до специфіки дисертації, але ключові зв'язки між Claim, Study, Data, Method та Evidence мають залишатися явними.


## Критерії оцінювання

Максимальна оцінка за ЛР5 -- **7 балів**.

| Компонент | Максимум |
|---|---:|
| Узгодженість `RQ → Claim → Study` | 1.5 |
| Обґрунтованість даних, вибірки, методів, baselines і контрольних умов | 1.5 |
| Якість `Experimental Protocol v1` та `Analysis Plan v1` | 1.5 |
| Простежуваність матриці `Claim -- Study -- Data -- Method -- Evidence` | 1.5 |
| Якість `Research Design v1`, аргументація на захисті та оформлення | 1.0 |
| **Разом** | **7.0** |


## Що потрібно показати на захисті?

Аспірант повинен уміти:

* показати зв'язок між `Gap Claim v1`, Research Questions і запланованими Study Units;
* пояснити, чому кожний Study Unit має саме такий дизайн;
* обґрунтувати вибір даних, одиниці аналізу та стратегії вибірки;
* пояснити роль кожного baseline і контрольної умови;
* показати, як конкретний результат експерименту буде перетворено на Evidence;
* назвати щонайменше два альтернативні пояснення очікуваного ефекту та спосіб їх контролю;
* пояснити, де проходить межа між допустимим і надмірно сильним висновком;
* показати власну схему `Research Design v1`;
* пояснити, які компоненти дизайну ще потребують валідації та перевірок робастності в ЛР6.


## Підсумок

Після ЛР5 дисертаційна програма повинна перейти від системи *запланованих наукових тверджень* до системи *запланованих досліджень, здатних породити релевантні докази*.

Ключовий результат:

$$
\text{Gap}
\rightarrow
\text{RQ}
\rightarrow
\text{Claim}
\rightarrow
\text{Study}
\rightarrow
\text{Data + Method + Comparison}
\rightarrow
\text{Analysis}
\rightarrow
\text{Expected Evidence}.
$$

`Research Design v1` ще не гарантує достовірності майбутніх висновків. Наступний крок -- перевірити, *наскільки надійною буде вся система доказів*: які метрики потрібні, як оцінювати невизначеність, які перевірки робастності та чутливості провести, які загрози внутрішній, конструктній та зовнішній валідності потрібно контролювати.

Саме це стане предметом ЛР6.
