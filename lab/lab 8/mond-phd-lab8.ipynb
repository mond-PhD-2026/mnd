**Мета**: інтегрувати результати ЛР1--ЛР7 у фінальний методологічний каркас власного дисертаційного дослідження, побудувати карту доказів дисертації та карту відповідності публікацій науковим твердженням, виконати другу контрольну ревізію дослідницької програми і підготувати її до аргументованого методологічного захисту.


::: callout-note
## Примітка

ЛР8 -- *фінальна інтеграційна робота курсу*. Вона не передбачає механічного копіювання результатів ЛР1--ЛР7.

Ваше завдання -- показати, що окремі артефакти утворюють *одну логічно узгоджену дослідницьку програму*:

$$
\text{Problem}
\rightarrow
\text{Gap}
\rightarrow
\text{RQ/H}
\rightarrow
\text{Design}
\rightarrow
\text{Evidence}
\rightarrow
\text{Validation}
\rightarrow
\text{Claim}
\rightarrow
\text{Contribution}.
$$

Оскільки дисципліна вивчається у першому семестрі першого року PhD, від вас не очікується завершена дисертація або вже отримані всі результати. Фінальний каркас має бути *доказово обґрунтованим планом дослідження*, достатньо сильним для подальшої систематичної роботи і водночас відкритим до перегляду в міру появи нових доказів.

Якщо певний елемент ще не може бути заповнений фактичним результатом, чітко позначте його як `PLANNED` і вкажіть, *який саме доказ має бути отриманий*.
:::

## Короткі теоретичні відомості

### 1. Фінальний методологічний каркас -- це система зв'язків

Наприкінці курсу важливо не просто мати набір правильних таблиць. Потрібно показати *зв'язки між ними*.

Мінімальна логіка:

$$
Problem
\rightarrow Gap
\rightarrow RQ/H
\rightarrow Task
\rightarrow Design/Method
\rightarrow Evidence
\rightarrow Validation
\rightarrow Claim
\rightarrow Contribution.
$$

До неї додаються:

$$
Limitations + Reproducibility + Traceability + Integrity + Publications.
$$

Якщо один із ключових переходів неможливо аргументувати, це сигнал до перегляду відповідної частини дослідницької програми.


### 2. Двобічна простежуваність

Каркас має дозволяти пройти логіку у двох напрямах.

*Вперед*:

$$
Problem \rightarrow \cdots \rightarrow Contribution.
$$

*Назад*:

$$
Contribution \rightarrow Claim \rightarrow Evidence \rightarrow Research \rightarrow RQ \rightarrow Gap \rightarrow Problem.
$$

Другий маршрут особливо важливий для захисту: кожне твердження про внесок повинно мати простежувану доказову основу.


### 3. Карта доказів дисертації

*Dissertation Evidence Map* показує не лише, які докази плануються або вже отримані, а й:

* який Claim підтримує кожний Evidence;
* яким способом Evidence буде валідовано;
* які обмеження залишаються;
* де зберігається або буде зберігатися дослідницький артефакт;
* який статус має ланка: `PLANNED`, `PARTIAL`, `SUPPORTED`, `GAP`.

У фінальній роботі ця карта стає центральним інструментом перевірки цілісності.


### 4. Контрольна точка перегляду 2

Перша контрольна точка в ЛР4 перевіряла, як доказова база змінила початкову концепцію.

Тепер запитання інше:

> *Чи утворюють усі спроєктовані компоненти одну логічно та доказово узгоджену програму дисертаційного дослідження?*

Використовуйте чотири рішення:

* `KEEP` -- залишити;
* `MODIFY` -- уточнити;
* `REMOVE` -- вилучити;
* `ADD` -- додати відсутній компонент.

Обґрунтоване `MODIFY`, `REMOVE` або `ADD` є ознакою методологічної рефлексії, а не помилкою.


Замініть шаблон конкретними вузлами `G1`, `RQ1`, `CL1`, `N1`, `E1` тощо. Не намагайтеся показати абсолютно все: схема повинна демонструвати *структуру аргументації*, а не бути нечитабельним каталогом.


## Що потрібно подати на захист

Підсумковий пакет ЛР8 має містити:

1. `Final Methodological Framework v1`;
2. аудит `Task ↔ Result ↔ Claim`;
3. `Dissertation Evidence Map v1`;
4. `Contribution Map vFinal`;
5. `Scope & Generalizability Map v1`;
6. `Publication--Claim Map v1`;
7. два повні ланцюги відтворюваності та простежуваності;
8. `Checkpoint Review 2`;
9. `Red Flag Audit`;
10. `Methodological Defense Map v1`;
11. відповіді на критичні запитання;
12. `Final Research Program Snapshot v1`;
13. фінальну Mermaid-карту дисертаційного дослідження.

Артефакти можуть бути включені безпосередньо до зошита або подані як окремі файли з чіткими посиланнями та версіями.


## Критерії оцінювання

Максимальна оцінка за ЛР8 -- **21 бал**.

| Критерій | Максимум |
|---|---:|
| Повнота та логічна узгодженість `Final Methodological Framework` | 4.0 |
| `Dissertation Evidence Map` та зв'язок Evidence--Validation--Claim | 3.0 |
| `Contribution Map`, науковий зміст внесків і межі Claims | 2.5 |
| `Publication--Claim Map` і простежуваність результатів | 1.5 |
| `Checkpoint Review 2` та обґрунтованість ревізій | 2.0 |
| `Red Flag Audit` і виявлення методологічних ризиків | 1.5 |
| `Final Research Program Snapshot` і фінальна карта | 1.5 |
| Методологічний захист: аргументація, відповіді на критичні запитання, володіння власним каркасом | 5.0 |
| **Разом** | **21.0** |

Висока оцінка передбачає не «ідеальність» початкової теми, а здатність *побачити слабкі місця, доказово їх пояснити та запропонувати методологічно коректне рішення*.


## Що потрібно показати на методологічному захисті?

Аспірант повинен уміти без читання великого тексту:

* за 1--2 хвилини пояснити Problem → Gap → RQ/H;
* показати, чому обраний Design відповідає типу ключових Claims;
* простежити щонайменше один Claim до Evidence, Validation та дослідницьких артефактів;
* пояснити, у чому полягає очікуване *нове знання*;
* назвати межі застосовності ключового Claim;
* показати, як публікації пов'язані з Claims;
* пояснити найважливіше рішення `MODIFY / REMOVE / ADD` у Checkpoint 2;
* назвати найбільший методологічний ризик, що залишається;
* пояснити, який наступний Evidence найбільше змінить рівень упевненості в дослідницькій програмі.


::: callout-important
## Важливо

Методологічний захист -- це не конкурс презентацій і не перевірка того, наскільки «переконливо» аспірант може рекламувати тему.

Якщо під час захисту виявлено слабке місце, коректна відповідь може мати форму:

> «Цей Claim поки має статус `PLANNED`; наявні Evidence недостатні для сильнішого формулювання. Для його перевірки передбачено ...».

Така відповідь методологічно сильніша, ніж спроба захистити твердження, для якого немає достатньої доказової основи.
:::


## Контрольні запитання

1. Що означає інтегрувати результати в дисертацію?
2. Чому набір коректних окремих досліджень ще не гарантує цілісності дисертації?
3. Чим Result відрізняється від Claim?
4. Що таке двобічна простежуваність?
5. Для чого потрібна Dissertation Evidence Map?
6. Що таке evidence gap?
7. Що таке orphan result?
8. Як виявити Claim, сильніший за Evidence?
9. Чому Contribution не можна ототожнювати з програмним артефактом?
10. Як перевірити узгодженість задач і результатів?
11. Що визначає Scope of Claim?
12. Чим generalizability відрізняється від універсальності?
13. Яку роль відіграє Publication--Claim Map?
14. Чому публікація не є заміною Evidence?
15. Що перевіряє Checkpoint Review 2?
16. Чим Checkpoint Review 2 відрізняється від Checkpoint Review 1?
17. Для чого потрібен Red Flag Audit?
18. Що є предметом методологічного захисту?
19. Чому визнання невизначеності може бути ознакою сильного захисту?
20. Який статус має фінальний каркас наприкінці першого семестру першого року PhD?


## Підсумок

ЛР8 завершує не дисертацію, а *першу системну методологічну ітерацію* роботи над нею.

Результатом курсу має бути не набір незалежних документів, а простежувана система:

$$
\text{Problem}
\rightarrow
\text{Gap}
\rightarrow
\text{RQ/H}
\rightarrow
\text{Design}
\rightarrow
\text{Evidence}
\rightarrow
\text{Validation}
\rightarrow
\text{Claim}
\rightarrow
\text{Contribution}.
$$

Фінальний каркас стає робочою картою подальшого PhD-дослідження. Він має змінюватися, коли нові дані, результати або література цього вимагають.

Тому головний результат ЛР8 -- здатність аспіранта не лише *показати власну дослідницьку програму*, а й аргументовано пояснити, чому кожний її ключовий компонент потрібний, якими доказами він підтримується і за яких умов має бути переглянутий.


## Рекомендовані джерела

1. Booth, W. C., Colomb, G. G., Williams, J. M., Bizup, J., & FitzGerald, W. T. *The Craft of Research*. 4th ed. Chicago: University of Chicago Press, 2016.

2. National Academies of Sciences, Engineering, and Medicine. *Reproducibility and Replicability in Science*. Washington, DC: The National Academies Press, 2019. DOI: 10.17226/25303.

3. Wohlin, C., Runeson, P., Höst, M., Ohlsson, M. C., Regnell, B., & Wesslén, A. *Experimentation in Software Engineering*. Berlin, Heidelberg: Springer, 2012.

4. Kitchenham, B., & Charters, S. *Guidelines for Performing Systematic Literature Reviews in Software Engineering*. EBSE Technical Report EBSE-2007-01, 2007.

5. Shadish, W. R., Cook, T. D., & Campbell, D. T. *Experimental and Quasi-Experimental Designs for Generalized Causal Inference*. Boston: Houghton Mifflin, 2002.

6. Nosek, B. A. et al. Promoting an open research culture. *Science*. 2015. Vol. 348, No. 6242. P. 1422--1425. DOI: 10.1126/science.aab2374.

7. Munafò, M. R. et al. A manifesto for reproducible science. *Nature Human Behaviour*. 2017. Vol. 1. Article 0021. DOI: 10.1038/s41562-016-0021.

8. Pineau, J. et al. Improving Reproducibility in Machine Learning Research: A Report from the NeurIPS 2019 Reproducibility Program. *Journal of Machine Learning Research*. 2021. Vol. 22, No. 164. P. 1--20.

9. Wilkinson, M. D. et al. The FAIR Guiding Principles for scientific data management and stewardship. *Scientific Data*. 2016. Vol. 3. Article 160018. DOI: 10.1038/sdata.2016.18.

10. ACM. *Artifact Review and Badging*. Association for Computing Machinery.
