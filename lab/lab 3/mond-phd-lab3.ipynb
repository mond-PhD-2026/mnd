**Мета**: побудувати відтворюваний протокол пошуку наукових джерел, сформувати первинну матрицю доказів та карту сучасного стану досліджень для власної дисертаційної проблематики, а також перевірити, наскільки сформульовані в ЛР1--ЛР2 наукова проблема, дослідницькі питання та `Research Gap v1` підтримуються наявними науковими доказами.

::: callout-note
## Примітка

Попередньо опрацювати матеріали **Лекції 3 «Формування доказової бази та сучасного стану досліджень»**.

ЛР3 є прямим продовженням ЛР1--ЛР2. Вихідними матеріалами є:

- `Research Problem v1`;
- `Conceptual Framework v1`;
- `Research Gap v1`;
- дослідницькі питання (*research questions*, RQ);
- попередня карта наукових тверджень (*Claim Map*);
- результати методологічного аудиту та архітектури дисертації.

На цьому етапі `Research Gap v1` розглядається як *гіпотеза про стан наукового поля*, яку необхідно перевірити за допомогою систематично організованого пошуку та синтезу доказів.

Результат ЛР3 не є остаточним літературним оглядом дисертації. Це перша відтворювана версія її доказової основи.
:::

## Що ви будете вміти?

Після виконання лабораторної роботи ви зможете:

* перетворювати дослідницьке питання на систему пошукових концептів (*search concepts*);
* будувати відтворювану пошукову стратегію та фіксувати її у протоколі пошуку (*search protocol*);
* визначати критерії включення та виключення джерел (*inclusion/exclusion criteria*);
* застосовувати прямий пошук і ланцюговий пошук (*snowballing*);
* відокремлювати бібліографічне джерело від доказу (*evidence*);
* формувати матрицю доказів (*evidence matrix*);
* зіставляти твердження різних досліджень з урахуванням умов, даних, методів і способів валідації;
* виявляти узгоджені, суперечливі та недостатньо підтверджені результати;
* будувати карту сучасного стану досліджень (*state-of-the-art map*);
* перевіряти першоджерела, знайдені за підтримки ШІ (*AI-assisted search*);
* формулювати аргументований висновок щодо того, які елементи `Research Gap v1` підтверджуються, потребують уточнення або мають бути відхилені.

## Короткі теоретичні відомості

### 1. Від списку публікацій до доказової бази

У ЛР3 ми не ставимо завдання «зібрати якомога більше статей». Нас цікавить інша логіка:

$$
RQ \rightarrow
\text{пошукові концепти} \rightarrow
\text{пошук} \rightarrow
\text{відбір} \rightarrow
\text{докази} \rightarrow
\text{синтез}.
$$

Окрема публікація ще не є доказовою базою. Вона стає її елементом лише тоді, коли з першоджерела вилучено конкретний результат або аргумент і встановлено, *яке твердження він підтримує або обмежує та за яких умов*.

Тому одиницею аналітичної роботи в цій лабораторній є не стаття сама по собі, а зв'язок:

$$
\text{Твердження} \leftrightarrow \text{Доказ} \leftrightarrow \text{Умови отримання}.
$$


### 2. Протокол пошуку

Протокол пошуку (*search protocol*) фіксує правила, за якими формується корпус джерел. Мінімально він повинен містити:

| Компонент | Що потрібно зафіксувати |
|---|---|
| Дослідницьке питання | RQ, для якого збираються докази |
| Пошукові концепти | основні поняття та їх синоніми |
| Джерела пошуку | бази даних, цифрові бібліотеки, пошукові системи |
| Пошуковий запит | точний рядок або його версії |
| Поля пошуку | title, abstract, keywords тощо |
| Часові межі | період публікацій |
| Дата пошуку | коли виконано пошук |
| Критерії включення | які роботи залишаємо |
| Критерії виключення | які роботи відкидаємо |
| Ланцюговий пошук | як перевіряємо посилання назад і вперед |
| Результат | кількість знайдених, переглянутих і включених робіт |

Важлива вимога -- інший дослідник повинен мати змогу зрозуміти, *як саме сформовано корпус джерел*.

### 3. Матриця доказів

Матриця доказів (*evidence matrix*) перетворює набір публікацій на структурований аналітичний матеріал.

Рекомендоване ядро матриці:

| ID | Джерело | Дослідницька задача | Дані / контекст | Метод | Основний результат | Доказ для нашого RQ | Обмеження | Пов'язане твердження |
|---|---|---|---|---|---|---|---|---|

За потреби аспірант може додати поля: метрики, обсяг вибірки, спосіб поділу даних, тип валідації, код/репозиторій, DOI, якість джерела, тип дослідження тощо.

Не слід заповнювати матрицю переказом анотації. Поле «Доказ для нашого RQ» має відповідати на питання:

> Що саме з цієї роботи ми можемо використати для підтримки, уточнення або спростування конкретного твердження у власному дослідженні?

### 4. Карта сучасного стану досліджень

Карта сучасного стану досліджень (*state-of-the-art map*) -- це вже результат синтезу, а не каталог джерел.

Вона має показати принаймні:

* які основні напрями сформувалися у предметному полі;
* які методи та типи даних переважають;
* які результати відтворюються у різних роботах;
* де результати суперечать один одному;
* за яких умов отримано позитивні та негативні результати;
* які аспекти досліджено недостатньо;
* які твердження про прогалину поки не мають достатньої доказової підтримки.

Саме ця карта стане основою для ЛР4, де `Research Gap v1` буде перетворено на аргументовану дослідницьку прогалину та пов'язано з майбутнім науковим внеском.

## Хід роботи

### Завдання 1. Зафіксуйте вихідну дослідницьку конструкцію

Перенесіть із ЛР2:

1. `Research Problem v1`;
2. `Research Gap v1`;
3. 2--4 основні дослідницькі питання;
4. ключові попередні твердження з `Claim Map`.

Для кожного RQ коротко запишіть, *які зовнішні наукові докази потрібні*, щоб оцінити його актуальність і стан дослідженості.

| RQ | Яке знання потрібно знайти в літературі? | Які твердження потрібно перевірити? |
|---|---|---|
| RQ1 | ... | ... |
| RQ2 | ... | ... |
| RQ3 | ... | ... |

### Завдання 2. Побудуйте систему пошукових концептів

Для одного основного RQ виділіть 2--4 концептуальні блоки. Для кожного визначте:

* основний термін;
* англомовний відповідник;
* синоніми;
* близькі терміни;
* небажані значення, які можуть створювати пошуковий шум.

Приклад структури:

| Концепт | Основний термін | Синоніми / варіанти | Небажані значення |
|---|---|---|---|
| $C_1$ | bot detection | social bot detection, automated account detection | chatbot |
| $C_2$ | robustness | generalization, stability | hardware robustness |
| $C_3$ | domain shift | topic shift, temporal shift | image domain adaptation |

Після цього сформуйте логічну конструкцію:

$$
Q=(C_{1,1}\lor C_{1,2}\lor\ldots)
\land
(C_{2,1}\lor C_{2,2}\lor\ldots)
\land \ldots
$$

Зафіксуйте не лише фінальний запит, а й 2--3 його попередні версії та коротко поясніть, чому вони були змінені.

### Завдання 3. Сформуйте протокол пошуку

Створіть таблицю протоколу.

| Поле | Ваше рішення |
|---|---|
| RQ | ... |
| Джерела пошуку | ... |
| Пошуковий запит | ... |
| Поля пошуку | ... |
| Часові межі | ... |
| Типи публікацій | ... |
| Мова | ... |
| Дата виконання пошуку | ... |
| Критерії включення | ... |
| Критерії виключення | ... |
| Процедура ланцюгового пошуку | ... |

Для дослідження рівня PhD не обмежуйте пошук одним сервісом або лише звичайним веб-пошуком. Обґрунтуйте вибір джерел відповідно до предметної області.

### Завдання 4. Проведіть первинний відбір джерел

Виконайте пошук за протоколом та зафіксуйте послідовність відбору:

1. результати пошуку;
2. усунення очевидних дублікатів;
3. перегляд назв та анотацій;
4. повнотекстова перевірка кандидатів;
5. остаточне включення до первинного корпусу;
6. ланцюговий пошук для ключових робіт.

Для лабораторної роботи достатньо сформувати *керований первинний корпус*, який дає змогу продемонструвати методологію. Не потрібно штучно збільшувати кількість джерел заради самого числа.

Зафіксуйте кількісний маршрут:

$$
N_{found}\rightarrow N_{screened}\rightarrow N_{fulltext}\rightarrow N_{included}.
$$

Коротко поясніть основні причини виключення робіт.

### Завдання 5. Перевірте першоджерела

Для кожної включеної роботи перевірте щонайменше:

* назву;
* авторів;
* рік;
* видання;
* DOI або інший стійкий ідентифікатор, якщо він існує;
* відповідність анотації реальному змісту;
* наявність у тексті результату, який ви плануєте використати як доказ.

::: callout-warning
## Використання ШІ

Пошук за підтримки ШІ (*AI-assisted search*) можна використовувати для генерації синонімів, пошукових концептів, пошуку кандидатів у джерела та первинного структурування матеріалу.

Проте відповідь LLM, згенерований бібліографічний запис або переказ статті не є першоджерелом.

Ланцюг перевірки:

$$
\text{AI suggestion}
\rightarrow
\text{candidate source}
\rightarrow
\text{primary source}
\rightarrow
\text{verified evidence}.
$$
:::

### Завдання 6. Побудуйте матрицю доказів

Для відібраних робіт створіть матрицю доказів. Мінімальна рекомендована структура:

| ID | Джерело | Дані / контекст | Метод | Результат | Доказ | Обмеження | Claim / RQ |
|---|---|---|---|---|---|---|---|
| S01 | ... | ... | ... | ... | ... | ... | ... |
| S02 | ... | ... | ... | ... | ... | ... | ... |

Важливо: не записуйте «автори показали високу ефективність». Вкажіть, *що саме отримано, за яких умов і чому цей результат релевантний вашому твердженню*.

Якщо дві роботи повідомляють різні результати, не усереднюйте їх механічно. Зафіксуйте можливі причини відмінностей: дані, вибірка, протокол, метрика, часовий період, предметна область, спосіб валідації тощо.

### Завдання 7. Побудуйте карту сучасного стану досліджень

Згрупуйте докази не за авторами, а за змістовими або методологічними напрямами.

Побудуйте власну схему. Наприклад:

```{mermaid}
%%| fig-cap: "Шаблон карти сучасного стану досліджень"
%%{init: {'theme': 'base', 'themeVariables': {
  'fontSize': '17px',
  'fontFamily': 'Trebuchet MS, Verdana, Arial, sans-serif',
  'primaryColor': '#E8F3FF',
  'primaryBorderColor': '#3478A8',
  'secondaryColor': '#FFF0C9',
  'tertiaryColor': '#E7F6EC',
  'lineColor': '#536878'
}}}%%
flowchart TB
    P("Наукова проблема") --> A("Напрям A")
    P --> B("Напрям B")
    P --> C("Напрям C")
    A --> A1("Що встановлено")
    A --> A2("Що обмежено")
    B --> B1("Узгоджені результати")
    B --> B2("Суперечливі результати")
    C --> C1("Недостатньо досліджено")
    A2 --> G("Кандидат у прогалину")
    B2 --> G
    C1 --> G

    classDef problem fill:#e0e7ff,stroke:#6366f1,stroke-width:2px,color:#111827;
    classDef process fill:#dbeafe,stroke:#2563eb,stroke-width:2px,color:#111827;
    classDef evidence fill:#dcfce7,stroke:#16a34a,stroke-width:2px,color:#111827;
    classDef result fill:#fef3c7,stroke:#d97706,stroke-width:2px,color:#111827;

    class P problem;
    class A,B,C process;
    class A1,B1 evidence;
    class A2,B2,C1,G result;
```

### Завдання 8. Проведіть аудит `Research Gap v1`

Поверніться до формулювання з ЛР2 та розкладіть його на окремі твердження.

| Твердження про прогалину | Докази «за» | Докази, що обмежують / суперечать | Рішення |
|---|---|---|---|
| G1 | S01, S04, ... | S07, ... | підтверджується / уточнити / відхилити |
| G2 | ... | ... | ... |

Не намагайтеся будь-якою ціною «довести», що початкова прогалина існує. Методологічно сильним результатом є також ситуація, коли літературний аналіз показує, що `Research Gap v1` був сформульований занадто широко або вже закритий попередніми роботами.

Сформулюйте `Research Gap v2` лише настільки конкретно, наскільки це дозволяє зібрана доказова база.

### Завдання 9. Сформулюйте синтез сучасного стану досліджень

Напишіть короткий аналітичний синтез, організований навколо проблеми та доказів, а не авторів.

Рекомендована логіка:

1. що в предметному полі вже встановлено достатньо переконливо;
2. які підходи домінують;
3. за яких умов отримано основні результати;
4. де результати суперечливі;
5. які обмеження повторюються;
6. які аспекти залишаються недостатньо дослідженими;
7. як це змінює вашу початкову постановку проблеми.

Кожне суттєве фактичне твердження має бути простежуваним до конкретних елементів матриці доказів.


## Контрольні питання

1. Чим відрізняється список літератури від доказової бази?
2. Для чого потрібен протокол пошуку?
3. Як дослідницьке питання перетворюється на пошукові концепти?
4. Чому пошуковий запит потрібно зберігати у точному вигляді?
5. Яку роль виконують критерії включення та виключення?
6. Що таке ланцюговий пошук (*snowballing*)?
7. Чому одиницею синтезу доцільно вважати твердження, а не публікацію?
8. Які поля повинна містити матриця доказів?
9. Як працювати із суперечливими результатами різних досліджень?
10. Чому відповідь LLM не можна використовувати як першоджерело?
11. Які перевірки потрібні для джерела, знайденого за підтримки ШІ?
12. Чим карта сучасного стану досліджень відрізняється від тематичного списку публікацій?
13. За яких умов `Research Gap v1` потрібно відхилити або суттєво переформулювати?
14. Як матриця доказів забезпечує простежуваність майбутніх тверджень дисертації?

## Вимоги до звіту

Звіт подається у вигляді відтворюваного документа та повинен містити:

1. вихідні `Research Problem v1`, `Research Gap v1` та RQ;
2. систему пошукових концептів;
3. повний протокол пошуку;
4. кількісний маршрут відбору джерел;
5. критерії включення та виключення;
6. матрицю доказів;
7. карту сучасного стану досліджень;
8. аудит `Research Gap v1`;
9. `Research Gap v2` -- якщо його уточнення вже достатньо обґрунтоване;
10. короткий аналітичний синтез сучасного стану досліджень;
11. список використаних джерел.

У звіті повинна зберігатися простежуваність:

$$
RQ \rightarrow
\text{Search Protocol} \rightarrow
\text{Source} \rightarrow
\text{Evidence} \rightarrow
\text{Claim} \rightarrow
\text{Research Gap}.
$$

## Критерії оцінювання

Максимальна оцінка за ЛР3 -- **7 балів**.

| Компонент | Максимум |
|---|---:|
| Коректність і відтворюваність протоколу пошуку | 1.5 |
| Обґрунтованість критеріїв відбору та процедура пошуку | 1.0 |
| Якість і простежуваність матриці доказів | 2.0 |
| Аналітичність карти сучасного стану досліджень | 1.0 |
| Доказовий аудит `Research Gap v1` та обґрунтованість висновків | 1.0 |
| Якість оформлення, відтворюваність і усний захист | 0.5 |
| **Разом** | **7.0** |

## Що потрібно показати на захисті?

Аспірант повинен уміти:

* відтворити логіку побудови пошукового запиту від RQ;
* пояснити вибір джерел пошуку та критеріїв відбору;
* показати приклад включеної та виключеної роботи;
* для 2--3 ключових джерел перейти від бібліографічного запису до конкретного доказу;
* пояснити хоча б один випадок узгоджених або суперечливих результатів;
* показати, як матриця доказів перетворюється на карту сучасного стану досліджень;
* аргументувати, що сталося з `Research Gap v1`: підтвердження, звуження, переформулювання або відхилення;
* пояснити, які питання залишаються відкритими перед переходом до ЛР4.

## Підсумок

Після ЛР3 початкова методологічна архітектура дисертації вперше проходить *зовнішню доказову перевірку*.

Результат лабораторної роботи -- не «огляд літератури за темою», а простежуваний ланцюг:

$$
\text{Research Problem}
\rightarrow
RQ
\rightarrow
\text{Search Protocol}
\rightarrow
\text{Evidence Matrix}
\rightarrow
\text{State-of-the-Art Map}
\rightarrow
\text{Research Gap v2}.
$$

У ЛР4 цей результат буде використано для обґрунтування дослідницької прогалини, карти наукових внесків та положень про наукову новизну.
