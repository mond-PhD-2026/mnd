::: callout-note
## Примітка

ЛР4 продовжує наскрізну роботу над *власною дисертацією*. Вхідними артефактами є результати ЛР1--ЛР3, передусім `Research Problem v1`, `Methodological Architecture v1`, `Evidence Matrix v1`, `State-of-the-Art Map v1` та уточнений `Research Gap v2`.

На цьому етапі не потрібно «доводити новизну дисертації назавжди». Потрібно побудувати **доказово обґрунтовану робочу версію**, яку надалі можна буде перевіряти, уточнювати або переглядати.
:::

## Що ви будете вміти?

Після виконання лабораторної роботи ви зможете:

* відрізняти *research gap* (дослідницьку прогалину) від загальної актуальності теми та простого факту відсутності певної реалізації;
* формулювати *Gap Claim* (твердження про прогалину) і простежувати його до доказової бази;
* відрізняти технічну новизну від нового наукового знання;
* будувати *Contribution Map* (карту наукових внесків);
* формулювати *Contribution Claim* (твердження про внесок) та *Novelty Claim* (твердження про новизну);
* перевіряти узгодженість ланцюга «проблема -- прогалина -- внесок -- новизна -- майбутній доказ»;
* виконувати контрольну ревізію методологічної архітектури дисертації.

## Короткі теоретичні відомості

### 1. Дослідницька прогалина -- це не «ніхто ще не робив»

Для PhD-дослідження недостатньо показати, що певний алгоритм, модель або технологія ще не застосовувалися до конкретного набору даних.

Методологічно сильна прогалина вказує, **якого знання бракує**, чому ця нестача є суттєвою для розв'язання наукової проблеми та якими доказами це підтверджується.

Зручно мислити так:

$$
\text{Research Gap}
=
\text{Known State}
+
\text{Limitation}
+
\text{Evidence}
+
\text{Scientific Consequence}.
$$

Отже, прогалина повинна бути не лише сформульована, а й *обґрунтована*.

### 2. Три твердження, які не можна змішувати

| Твердження | Основне питання | Що його підтримує |
|---|---|---|
| *Gap Claim* (твердження про прогалину) | Чого істотно бракує у сучасному знанні? | література, матриця доказів, суперечності, обмеження |
| *Contribution Claim* (твердження про внесок) | Що саме додає дослідження? | майбутній результат + його перевірка |
| *Novelty Claim* (твердження про новизну) | У чому нове знання порівняно з відомим? | порівняння з наявним доробком + докази відмінності |

Ці три твердження пов'язані, але не є взаємозамінними.


## Контрольні питання

1. Чим дослідницька прогалина відрізняється від актуальності теми?
2. Чому твердження «метод X ще не застосовували до задачі Y» не є достатнім Gap Claim?
3. Якими доказами має підтримуватися твердження про прогалину?
4. Чим `Gap Claim` відрізняється від `Contribution Claim`?
5. Чим науковий внесок відрізняється від технічного артефакту?
6. Що має містити `Novelty Claim`?
7. Чому сила твердження про новизну не може перевищувати силу доказів?
8. Як пов'язані дослідницьке питання, задача та науковий внесок?
9. Для чого потрібна `Contribution Map`?
10. Коли технічна новизна може стати носієм нового наукового знання?
11. Для чого використовується `Checkpoint Review 1`?
12. Чому перегляд початкового формулювання проблеми не є методологічною помилкою?
13. Які результати ЛР3 можуть змусити змінити `Research Gap`?
14. Які артефакти ЛР4 стануть входом для проєктування дизайну дослідження у ЛР5?

## Вимоги до звіту

Звіт має містити:

1. актуальні вхідні артефакти ЛР1--ЛР3;
2. доказовий аудит `Research Gap v2`;
3. `Gap Evidence Map`;
4. `Gap Claim v1`;
5. аудит `Expected Contributions v1`;
6. `Contribution Map v1`;
7. `Novelty Claims v1`;
8. тест «технічна новизна чи нове наукове знання?»;
9. матрицю логічної узгодженості;
10. `Checkpoint Review 1`;
11. `Methodological Architecture v2`;
12. короткий висновок про те, що саме змінилося у дослідницькій конструкції та чому.

Усі твердження про сучасний стан і прогалину мають бути простежуваними до матриці доказів ЛР3.

## Критерії оцінювання

Максимальна оцінка за ЛР4 -- **7 балів**.

| Компонент | Максимум |
|---|---:|
| Доказовість і коректність `Gap Claim v1` | 1.5 |
| Якість та простежуваність `Contribution Map v1` | 1.5 |
| Обґрунтованість `Novelty Claims v1` | 1.5 |
| Логічна узгодженість Problem -- Gap -- RQ -- Contribution -- Novelty | 1.0 |
| Якість `Checkpoint Review 1` та аргументація змін | 1.0 |
| Якість оформлення, версіонування та усний захист | 0.5 |
| **Разом** | **7.0** |

## Що потрібно показати на захисті?

Аспірант повинен уміти:

* коротко сформулювати свій `Gap Claim v1`;
* показати, **які саме докази ЛР3** його підтримують;
* пояснити структуру `Contribution Map v1`;
* для кожного основного внеску пояснити, у чому полягає очікуване нове знання;
* захистити щонайменше одне `Novelty Claim v1`;
* показати логічний зв'язок між проблемою, прогалиною, RQ, задачами, внесками та новизною;
* пояснити рішення `KEEP / MODIFY / REMOVE` у `Checkpoint Review 1`;
* назвати, які елементи ще залишаються недостатньо доказовими та що потрібно перевірити далі.

## Підсумок

Після ЛР4 аспірант має перейти від загальної заяви про «новизну теми» до **простежуваної архітектури наукового внеску**.

Результат роботи -- не остаточні формулювання для дисертації, а доказово обґрунтована версія:

$$
\text{Evidence}
\rightarrow
\text{Research Gap}
\rightarrow
\text{Contribution}
\rightarrow
\text{Novelty Claim}
\rightarrow
\text{Evidence Needed}.
$$

`Checkpoint Review 1` фіксує, як доказова база змінила початкову методологічну конструкцію. Саме `Methodological Architecture v2` стає входом для ЛР5 -- проєктування дизайну дослідження та системи методів.
