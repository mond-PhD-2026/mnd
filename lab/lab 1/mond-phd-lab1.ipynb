**Мета**: виконати первинний методологічний аудит вихідної дисертаційної проблематики, відокремити практичну або інженерну проблему від власне наукової проблеми, сформулювати *Research Problem v1* (наукова проблема -- версія 1) та побудувати *Conceptual Framework v1* (первинний концептуальний каркас) майбутнього дисертаційного дослідження.

::: callout-note
## Примітка

Попередньо опрацювати матеріали **Лекції 1 «Наукове дослідження рівня PhD у комп'ютерних науках»**.

Ця лабораторна робота виконується **на матеріалі власної дисертаційної проблематики аспіранта**. На цьому етапі не вимагається остаточне формулювання теми дисертації, об'єкта, предмета, мети, задач або гіпотез -- їх методологічне узгодження буде предметом наступної лабораторної роботи.
:::

## Що ви будете вміти?

Після виконання лабораторної роботи ви зможете:

* проводити *problem audit* (аудит проблематики) майбутнього дисертаційного дослідження;
* розрізняти *practical problem* (практичну проблему), *engineering problem* (інженерну проблему) та *research problem* (наукову проблему);
* відокремлювати артефакт, який планується створити, від наукового знання, яке необхідно отримати;
* формулювати попередню версію наукової проблеми без передчасної декларації «готової» наукової новизни;
* визначати ключові концепти, сутності та передбачувані зв'язки між ними;
* розрізняти встановлені факти, робочі припущення та твердження, які ще потребують доказів;
* будувати *Conceptual Framework v1* (первинний концептуальний каркас);
* формувати перелік питань, які повинні бути перевірені під час подальшого систематичного пошуку літератури.

## Короткі теоретичні відомості

### 1. З чого реально починається дисертаційне дослідження?

На старті PhD-дослідження аспірант зазвичай має тему або напрям, певний професійний досвід, набір технологій, попередні публікації, доступні дані та уявлення про те, «що хотілося б зробити».

Проте цей набір ще не є науковою проблемою.

Для комп'ютерних наук особливо характерна ситуація, коли вихідне формулювання фактично описує **артефакт**:

> розробити систему, модель, метод, алгоритм, програмний комплекс, інформаційну технологію або інтелектуального агента.

Створення артефакту може бути необхідною частиною дисертації, але саме по собі воно ще не визначає, **яке нове знання буде отримано**.

Тому перша методологічна операція -- не «затвердити красиву назву теми», а виконати аудит того, що ми вже знаємо про проблему і що поки лише припускаємо.

### 2. Практична, інженерна та наукова проблема

Розглянемо три рівні постановки проблеми.

| Рівень | Основне питання | Типовий результат |
|---|---|---|
| *Practical problem* (практична проблема) | Що в реальному світі працює незадовільно або створює потребу в змінах? | опис потреби, обмеження або небажаного стану |
| *Engineering problem* (інженерна проблема) | Що потрібно створити або вдосконалити, щоб змінити цей стан? | система, алгоритм, модель, програмний засіб, архітектура |
| *Research problem* (наукова проблема) | Якого знання бракує, щоб обґрунтовано пояснити, передбачити або розв'язати проблему? | нове перевірюване знання |

Наприклад, твердження

> «потрібно створити точніший метод виявлення ботоподібної активності»

ще є переважно інженерною постановкою.

Наукове питання виникає тоді, коли ми уточнюємо, **які закономірності, ознаки, механізми, залежності або межі застосовності існуючих підходів недостатньо встановлені**.


### 3. Artifact ≠ Claim

Для дисертаційного дослідження важливо від самого початку розділяти:

* **Artifact** (артефакт) -- те, що дослідник створює;
* **Claim** (наукове твердження) -- те, що дослідник стверджує про досліджуваний об'єкт, метод, закономірність, механізм або ефект;
* **Evidence** (доказова база) -- результати, на яких це твердження ґрунтується.

Наприклад:

| Елемент | Приклад |
|---|---|
| Artifact | новий алгоритм класифікації |
| Claim | запропонований спосіб представлення поведінкових ознак підвищує стійкість класифікації за певних умов |
| Evidence | результати експериментів, порівняння з baseline, аналіз чутливості, статистична оцінка |

У ЛР1 ми ще **не формулюємо остаточні claims**. Наше завдання -- знайти ті місця у вихідній проблематиці, де потенційно може виникнути перевірюване нове знання.

### 4. Не все, що здається фактом, уже є фактом

На початку дослідження корисно маркувати твердження за їхнім епістемічним статусом:

* **F -- Fact / established evidence** (встановлений факт / наявне підтвердження): твердження вже має надійне джерело або власні перевірені результати;
* **A -- Assumption** (робоче припущення): твердження видається правдоподібним, але ще не перевірене;
* **E -- Evidence needed** (потребує доказів): твердження принципово важливе для майбутньої аргументації та має бути перевірене;
* **U -- Unknown** (невідомо): на цьому етапі інформації недостатньо навіть для обґрунтованого припущення.

Таке маркування запобігає поширеній помилці -- перетворенню власних очікувань дослідника на «відомі факти».

### 5. Що таке первинний концептуальний каркас?

*Conceptual Framework* (концептуальний каркас) -- це не декоративна схема майбутньої дисертації. Він фіксує, **які ключові концепти та сутності ми вважаємо суттєвими і які зв'язки між ними припускаємо**.

У найпростішому вигляді:

$$
CF_1 = (C, R, A),
$$

де

* $C$ -- множина ключових *concepts* (концептів);
* $R$ -- множина передбачуваних *relationships* (зв'язків);
* $A$ -- множина *assumptions* (припущень), на яких ці зв'язки поки ґрунтуються.

Позначення $CF_1$ підкреслює, що це лише **версія 1**. Після аналізу літератури, уточнення прогалини та проєктування досліджень каркас має змінюватися.

## Хід роботи

### Крок 1. Зафіксуйте вихідний стан дисертаційної проблематики

Не намагайтеся одразу «покращувати» формулювання. Спочатку зафіксуйте той стан, з якого реально починається робота.

Заповніть таблицю.

| Елемент | Вихідне формулювання |
|---|---|
| Робоча назва теми / напрям дослідження | |
| Предметна область | |
| Яку реальну проблему ви бачите? | |
| Який об'єкт або процес вас цікавить? | |
| Що ви попередньо плануєте створювати / досліджувати? | |
| Які дані або інші емпіричні матеріали потенційно доступні? | |
| Які методи / технології вже розглядаються? | |
| Який результат ви зараз інтуїтивно вважаєте «новим»? | |

Останній рядок є особливо важливим: на цьому етапі це **не декларація наукової новизни**, а лише вихідна гіпотеза дослідника про можливий внесок.

### Крок 2. Розділіть проблему на три рівні

Сформулюйте кожен рівень одним або кількома реченнями.

| Рівень | Ваше формулювання |
|---|---|
| *Practical problem* (практична проблема) | |
| *Engineering problem* (інженерна проблема) | |
| *Research problem* (наукова проблема) | |

Після заповнення перевірте себе:

1. Чи не зводиться *research problem* до фрази «необхідно розробити...»?
2. Чи містить вона вказівку на дефіцит знання?
3. Чи можна у принципі отримати докази, які підтвердять або спростують майбутні твердження?
4. Чи не закладено у формулювання наперед очікуваний «позитивний» результат?

### Крок 3. Виконайте аудит ключових тверджень

Випишіть **не менше 8** тверджень, на яких зараз фактично тримається ваша проблематика.

Приклади типів тверджень:

* «існуючі методи недостатньо ...»;
* «ознака $X$ пов'язана з ...»;
* «метод $Y$ має обмеження за умов ...»;
* «наявні датасети не відображають ...»;
* «запропонований клас моделей потенційно дозволяє ...».

| № | Твердження | Статус F / A / E / U | На чому воно зараз ґрунтується? | Що потрібно перевірити? |
|---:|---|:---:|---|---|
| 1 | | | | |
| 2 | | | | |
| 3 | | | | |
| 4 | | | | |
| 5 | | | | |
| 6 | | | | |
| 7 | | | | |
| 8 | | | | |

::: callout-warning
## Важливо

Відсутність доказів на цьому етапі не є помилкою. Помилкою було б **не помітити**, що важливе твердження поки є лише припущенням.
:::

### Крок 4. Сформулюйте *Research Problem v1*

Спробуйте побудувати формулювання за логікою:

> У контексті **[предметна область / клас систем / процесів]** існує проблема **[що саме недостатньо пояснено, встановлено або обґрунтовано]**, унаслідок чого **[яке науково або практично значуще обмеження виникає]**. Наявні підходи **[попередньо відоме обмеження]**, однак характер / механізм / умови / межі **[чого саме]** потребують дослідження.

Це **шаблон мислення, а не шаблон речення**. Не потрібно механічно зберігати його синтаксис.

Запишіть власну версію:

> **Research Problem v1:** ...

### Крок 5. Виділіть ключові концепти

Побудуйте словник первинного концептуального каркаса.

| Концепт / сутність | Робоче визначення | Роль у проблемі | Статус визначення |
|---|---|---|---|
| $C_1$ | | | усталене / потребує уточнення |
| $C_2$ | | | |
| $C_3$ | | | |
| $C_4$ | | | |
| $C_5$ | | | |

Уникайте ситуації, коли в одному дослідженні ключовий термін використовується у кількох різних значеннях без явного розмежування.

### Крок 6. Побудуйте *Conceptual Framework v1*

Покажіть:

1. основні концепти або сутності;
2. передбачувані зв'язки між ними;
3. контекст або умови, які можуть змінювати ці зв'язки;
4. очікуваний результат / явище, яке потрібно пояснити або оцінити.

Використайте Mermaid або інший засіб візуалізації. Нижче наведено **каркас**, який потрібно перебудувати відповідно до власної дисертаційної проблематики.

```{mermaid}
%%| fig-cap: "Шаблон первинного концептуального каркаса"
flowchart LR
    C1("Ключовий концепт C1") --> R("Досліджуваний зв'язок / механізм")
    C2("Ключовий концепт C2") --> R
    K("Контекст / умови") --> R
    R --> Y("Результат / явище Y")
    U("Невизначеність / обмеження") -.-> R

    classDef concept fill:#e0e7ff,stroke:#6366f1,stroke-width:2px,color:#111827;
    classDef process fill:#dbeafe,stroke:#2563eb,stroke-width:2px,color:#111827;
    classDef result fill:#dcfce7,stroke:#16a34a,stroke-width:2px,color:#111827;
    classDef audit fill:#fee2e2,stroke:#dc2626,stroke-width:2px,color:#111827;

    class C1,C2,K concept;
    class R process;
    class Y result;
    class U audit;
```

## Підсумковий артефакт лабораторної роботи

За результатами ЛР1 аспірант формує єдиний артефакт:

> **Problem Audit v1 + Research Problem v1 + Conceptual Framework v1**

Він повинен містити:

1. вихідний опис дисертаційної проблематики;
2. розмежування практичної, інженерної та наукової проблем;
3. аудит не менше 8 ключових тверджень із маркуванням F / A / E / U;
4. формулювання *Research Problem v1*;
5. словник ключових концептів;
6. *Conceptual Framework v1* та таблицю передбачуваних зв'язків;
7. результати *opponent audit*;
8. перелік питань, які потребують подальшої перевірки.

::: callout-tip
## Критерій якісного результату

Після ЛР1 аспірант повинен не «довести, що тема хороша», а **чіткіше бачити межу між тим, що вже відомо, тим, що він лише припускає, і тим, що реально потребує наукового дослідження**.
:::
