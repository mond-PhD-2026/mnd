## Що ви будете вміти?

Після виконання лабораторної роботи ви зможете:

* розрізняти Result, Evidence та Validated Evidence;
* будувати матрицю `Claim -- Evidence -- Validation`;
* узгоджувати метрики з науковим конструктом і типом Claim;
* планувати оцінювання невизначеності (uncertainty) та розміру ефекту (effect size);
* відрізняти статистичну значущість від практичної значущості;
* проєктувати robustness checks, ablation studies та sensitivity analysis;
* ідентифікувати загрози внутрішній, конструктній, зовнішній валідності та валідності висновків;
* визначати межі допустимого наукового висновку;
* коректно планувати інтерпретацію негативних і нульових результатів;
* виявляти `evidence debt` у власній дисертаційній програмі.


## Короткі теоретичні відомості


### 1. Result, Evidence та Validation

У дослідницькій програмі важливо не ототожнювати числовий результат із доказом.

$$
\text{Result} \neq \text{Evidence} \neq \text{Validated Evidence}.
$$

*Result* -- безпосередньо отриманий результат вимірювання, експерименту, моделювання або аналізу.

*Evidence* -- результат, який є релевантним конкретному Claim.

*Validation* -- система перевірок, що визначає надійність Evidence, альтернативні пояснення та межі допустимого висновку.

Отже, магістральна логіка ЛР6:

$$
\text{Claim}
\rightarrow
\text{Evidence}
\rightarrow
\text{Validation}
\rightarrow
\text{Bounded Conclusion}.
$$



### 2. Метрика є частиною операціоналізації

Метрика повинна відповідати тому конструкту, про який сформульовано Claim.

Наприклад, якщо Claim стосується *стійкості*, одного значення accuracy недостатньо. Потрібно операціоналізувати стійкість як зміну релевантної метрики за контрольованої зміни умов.

Для порівняльного Claim можна визначити:

$$
\Delta=M_{\text{proposed}}-M_{\text{baseline}},
$$

але $\Delta$ потрібно інтерпретувати разом із невизначеністю, контекстом і практичною значущістю.


### 3. Сукупна доказова база

Одне ключове твердження може підтримуватися кількома взаємодоповнювальними доказами:

$$
\mathcal{E}(C)=\{E_1,E_2,\ldots,E_k\}.
$$

При цьому п'ять повторень одного експерименту не еквівалентні п'ятьом незалежним типам перевірки.

До сукупної доказової бази можуть входити:

- primary experiment;
- repeated evaluation;
- robustness study;
- ablation study;
- sensitivity analysis;
- external validation;
- перевірка альтернативних пояснень.

Завдання аспіранта -- визначити, *які саме типи Evidence потрібні для сили конкретного Claim*.


### 4. Чотири виміри валідності

| Вид валідності | Основне питання |
|---|---|
| Внутрішня (internal validity) | Чи не пояснюється ефект стороннім фактором або особливістю дизайну? |
| Конструктна (construct validity) | Чи вимірюємо ми саме той конструкт, про який робимо Claim? |
| Зовнішня (external validity) | На які дані, домени, популяції та умови можна переносити висновок? |
| Валідність висновків (conclusion validity) | Чи достатньо коректні дані й аналіз для зробленого висновку? |

Кожна процедура Validation повинна бути пов'язана з конкретною загрозою валідності.


### 5. Сила Claim не повинна перевищувати силу Evidence

Робоче правило:

$$
\boxed{\text{Strength(Claim)} \leq \text{Strength(Evidence)}}.
$$

Наприклад:

| Evidence | Надто сильне твердження | Методологічно обережніше твердження |
|---|---|---|
| один dataset | метод універсально кращий | метод показав перевагу в перевіреному контексті |
| один seed | метод стабільно кращий | у цьому запуску отримано вищу метрику |
| кореляція | X спричиняє Y | X асоційований з Y |
| ablation | компонент є причиною явища | компонент робить внесок у результат системи |
| $p>0.05$ | методи еквівалентні | не отримано достатніх підстав виявити відмінність |


## Хід роботи


### Завдання 1. Зафіксуйте вхідну доказову архітектуру

Перенесіть актуальні результати ЛР5.

| Артефакт | Актуальна версія | Що використовується у ЛР6 |
|---|---|---|
| Research Question(s) |  |  |
| Contribution / Novelty Claims |  |  |
| Study Units |  |  |
| Data Plan |  |  |
| Experimental Protocol |  |  |
| Analysis Plan |  |  |
| Claim -- Study -- Data -- Method -- Evidence |  |  |

Після заповнення коротко зафіксуйте, які Claims є *центральними* для дисертації, а які -- допоміжними.


### Завдання 2. Побудуйте `Claim -- Evidence -- Validation Matrix v1`

Для кожного ключового Claim сформуйте основну матрицю ЛР6.

| Claim ID | Claim | Primary Evidence | Additional Evidence | Validation procedure | Допустимий висновок |
|---|---|---|---|---|---|
| C1 |  |  |  |  |  |
| C2 |  |  |  |  |  |
| C3 |  |  |  |  |  |

Для кожного рядка перевірте:

1. Evidence справді відповідає Claim?
2. Validation перевіряє конкретну слабкість Evidence?
3. Допустимий висновок не ширший за перевірений контекст?


### Завдання 3. Проведіть аудит метрик

Для кожного центрального Claim визначте, *що саме вимірює метрика*.

| Claim | Конструкт | Операціоналізація | Primary metric | Secondary metric(s) | Чому метрика валідна для Claim |
|---|---|---|---|---|---|
|  |  |  |  |  |  |

Окремо вкажіть метрики, які *не слід* використовувати як основні, та поясніть чому.

Наприклад, для незбалансованої класифікації accuracy може бути допоміжною, але не primary metric.


### Завдання 4. Спроєктуйте оцінювання невизначеності

Для кожного ключового числового результату визначте джерела варіативності та спосіб їх врахування.

| Result / Metric | Джерело невизначеності | Що повторюється / варіюється | Кількість повторень або схема | Показник uncertainty |
|---|---|---|---|---|
|  |  |  |  |  |

Можливі джерела:

- sampling;
- train/test split;
- random initialization;
- stochastic optimization;
- sampling у генеративних моделях;
- measurement noise;
- temporal / domain variation.

Необхідно пояснити, чому обрана схема відповідає саме вашому дослідженню.


### Завдання 5. Визначте розмір ефекту та практичну значущість

Для порівняльних Claims заповніть таблицю.

| Claim | Порівнювані умови | Effect size / різниця | Що вважається practically meaningful | Обґрунтування порогу |
|---|---|---|---|---|
|  |  |  |  |  |

Не обмежуйтеся формулюванням «різниця статистично значуща».

Потрібно відповісти на два окремі питання:

1. *Чи достатньо даних, щоб виявити відмінність?*
2. *Чи є величина відмінності науково або практично суттєвою?*


### Завдання 6. Сформуйте `Robustness Plan v1`

Для кожного центрального Claim визначте обґрунтовані perturbations або альтернативні налаштування.

| Claim | Що варіюємо | Діапазон / альтернативи | Що має залишатися стабільним | Критерій втрати робастності |
|---|---|---|---|---|
|  |  |  |  |  |

Розгляньте за потреби:

- random seed;
- data split;
- preprocessing;
- subset / sample;
- noise level;
- missingness;
- hyperparameters;
- time period;
- domain;
- implementation choice.

Не потрібно механічно перевіряти всі пункти. Оберіть ті, що відповідають реальним загрозам вашого Claim.


### Завдання 7. Спроєктуйте `Ablation Study v1`

Виконується для Claims, у яких науковий внесок пов'язаний зі структурою методу, моделі, архітектури або інформаційної технології.

| Компонент | Повна система | Варіант без компонента | Очікувана зміна | Метрика | Який Claim перевіряється |
|---|---|---|---|---|---|
| A |  |  |  |  |  |
| B |  |  |  |  |  |
| C |  |  |  |  |  |

Якщо компоненти можуть взаємодіяти, додайте факторіальний план.

Якщо ablation study для вашого дослідження методологічно не застосовується, *не вигадуйте його*. Обґрунтуйте це та запропонуйте релевантніший спосіб декомпозиції Evidence.


### Завдання 8. Сформуйте `Sensitivity Analysis Plan v1`

Визначте параметри або припущення, від яких потенційно залежить висновок.

| Параметр / припущення | Базове значення | Діапазон / альтернативи | Результат, що відстежується | Критична межа |
|---|---:|---|---|---|
|  |  |  |  |  |

Мета -- знайти не «найкраще» значення параметра, а *область стабільності наукового висновку*.


### Завдання 9. Побудуйте `Validity Threats Matrix`

Для кожного центрального Study Unit проведіть аудит чотирьох видів валідності.

| Study / Claim | Вид валідності | Конкретна загроза | Наслідок для Claim | Запобігання / перевірка | Residual risk |
|---|---|---|---|---|---|
|  | Internal |  |  |  |  |
|  | Construct |  |  |  |  |
|  | External |  |  |  |  |
|  | Conclusion |  |  |  |  |

`Residual risk` -- ризик, який залишається навіть після запланованої процедури контролю.

Цей стовпець потрібний, щоб не створювати ілюзію «повністю валідованого» дослідження.


### Завдання 10. Перевірте зовнішню валідність і межі узагальнення

Для кожного Claim явно визначте область, на яку він може поширюватися.

| Claim | Перевірені дані / популяція | Перевірений домен | Час / середовище | Що НЕ перевірено | Максимально допустиме узагальнення |
|---|---|---|---|---|---|
|  |  |  |  |  |  |

Після таблиці сформулюйте один приклад:

- *надто широкого* висновку;
- *коректного bounded conclusion* для тих самих результатів.


### Завдання 11. Спроєктуйте правила для null / negative results

Для 2--4 ключових гіпотез або очікуваних ефектів заздалегідь визначте сценарії.

| Hypothesis / Expected effect | Позитивний результат | Null result | Negative result | Як зміниться Claim |
|---|---|---|---|---|
|  |  |  |  |  |

Важливо: нульовий результат не повинен автоматично перетворюватися на твердження про еквівалентність, а негативний результат -- приховуватися як «невдалий експеримент».


### Завдання 12. Проведіть аудит `Evidence Debt`

Позначте твердження, для яких доказова архітектура поки що неповна.

| Claim | Відсутній / слабкий Evidence | Чого бракує у Validation | Ризик для дисертації | Пріоритет | Запланована дія |
|---|---|---|---|---|---|
|  |  |  |  | High / Medium / Low |  |

Приклади `evidence debt`:

- відсутня external validation;
- Claim про robustness перевіряється лише в одній умові;
- uncertainty не оцінюється;
- novelty claim спирається лише на benchmark;
- внесок компонента заявлено без ablation;
- межі узагальнення не визначені.


### Завдання 13. Побудуйте інтегровану `Evidence Architecture v1`

Об'єднайте результати ЛР5 і ЛР6 для 2--4 центральних Claims.

| Claim | Study | Primary Evidence | Metric | Uncertainty | Effect size | Robustness | Ablation / Sensitivity | Validity threats | Bounded conclusion |
|---|---|---|---|---|---|---|---|---|---|
|  |  |  |  |  |  |  |  |  |  |

Це одна з ключових таблиць поточного етапу курсу.

Вона повинна дозволити сторонньому досліднику зрозуміти:

> *який саме шлях веде від наукового твердження до обґрунтованого висновку*.


### Завдання 14. Візуалізуйте доказову архітектуру одного Claim

Для одного центрального Claim адаптуйте схему під власне дослідження.

```{mermaid}
%%| fig-cap: "Шаблон Evidence Architecture для одного Claim"
%%{init: {'theme': 'base', 'themeVariables': {
  'fontSize': '17px',
  'fontFamily': 'Trebuchet MS, Verdana, Arial, sans-serif',
  'primaryColor': '#E8F3FF',
  'primaryBorderColor': '#3478A8',
  'secondaryColor': '#FFF0C9',
  'tertiaryColor': '#EAF7EA',
  'lineColor': '#315B7D'
}}}%%
flowchart LR
    C([Claim C1])
    S([Study Unit])
    E([Primary<br/>Evidence])
    U([Uncertainty<br/>+ Effect size])
    R([Robustness])
    A([Ablation /<br/>Sensitivity])
    V([Validity<br/>Threats])
    B([Bounded<br/>Conclusion])

    C --> S --> E --> U --> R --> A --> V --> B
```

### Завдання 15. Проведіть фінальний тест узгодженості

Для кожного центрального Claim перевірте ланцюг:

$$
\text{Claim}
\rightarrow
\text{Construct}
\rightarrow
\text{Metric}
\rightarrow
\text{Study}
\rightarrow
\text{Result}
\rightarrow
\text{Uncertainty}
\rightarrow
\text{Validation}
\rightarrow
\text{Conclusion}.
$$

Заповніть короткий аудит.

| Перехід | Узгоджено? | Потенційний розрив | Що потрібно змінити |
|---|---|---|---|
| Claim → Construct |  |  |  |
| Construct → Metric |  |  |  |
| Metric → Study |  |  |  |
| Study → Result |  |  |  |
| Result → Evidence |  |  |  |
| Evidence → Validation |  |  |  |
| Validation → Conclusion |  |  |  |

Особливу увагу приділіть трьом небезпечним переходам:

$$
\text{Metric} \not\Rightarrow \text{Construct},
$$

$$
\text{Result} \not\Rightarrow \text{Claim},
$$

$$
\text{Local Evidence} \not\Rightarrow \text{Universal Conclusion}.
$$



## Контрольні питання

1. Чим Result відрізняється від Evidence?
2. Що перетворює Evidence на Validated Evidence?
3. Чому одна висока метрика не є достатнім доказом загальної переваги?
4. Що означає сукупна доказова база?
5. Як метрика пов'язана з construct validity?
6. Чому один `random_state` не доводить стабільність результату?
7. Чим effect size відрізняється від statistical significance?
8. Чим statistical significance відрізняється від practical significance?
9. Яку функцію виконує robustness study?
10. Для яких Claims потрібна ablation study?
11. Чим sensitivity analysis відрізняється від ablation study?
12. Які основні загрози internal validity у ML-експерименті?
13. Що означає construct validity?
14. Що обмежує external validity?
15. Що таке conclusion validity?
16. Чому null result не означає автоматичної еквівалентності?
17. Що означає `Strength(Claim) ≤ Strength(Evidence)`?
18. Що таке bounded conclusion?
19. Що таке evidence debt?
20. Який зв'язок між Evidence Architecture і Research Design?


## Вимоги до звіту

Звіт має містити:

1. актуальні вхідні артефакти ЛР5;
2. `Claim -- Evidence -- Validation Matrix v1`;
3. аудит метрик та операціоналізації;
4. план оцінювання uncertainty;
5. effect size / practical significance plan;
6. `Robustness Plan v1`;
7. `Ablation Study v1` або обґрунтування його незастосовності;
8. `Sensitivity Analysis Plan v1`;
9. `Validity Threats Matrix`;
10. аналіз external validity та меж узагальнення;
11. правила для null / negative results;
12. `Evidence Debt Register`;
13. `Evidence Architecture v1`;
14. Mermaid-схему доказової архітектури одного Claim;
15. фінальний тест узгодженості;
16. короткий висновок: *які Claims уже мають достатньо спроєктовану доказову підтримку, а які потребують посилення*.


## Критерії оцінювання

Максимальна оцінка за ЛР6 -- **7 балів**.

| Компонент | Максимум |
|---|---:|
| Узгодженість `Claim → Evidence → Validation → Conclusion` | 1.5 |
| Обґрунтованість метрик, uncertainty, effect size та practical significance | 1.5 |
| Якість robustness / ablation / sensitivity plans | 1.5 |
| Повнота `Validity Threats Matrix`, меж узагальнення та `Evidence Debt Register` | 1.5 |
| Якість `Evidence Architecture v1`, аргументація на захисті та оформлення | 1.0 |
| **Разом** | **7.0** |


## Що потрібно показати на захисті?

Аспірант повинен уміти:

* показати зв'язок між центральним Claim та конкретними Evidence;
* пояснити, чому обрані метрики справді операціоналізують потрібний конструкт;
* обґрунтувати спосіб оцінювання uncertainty;
* пояснити, який effect size має практичний або науковий сенс;
* назвати щонайменше одну реальну загрозу кожному релевантному виду валідності;
* пояснити призначення robustness, ablation та sensitivity checks у власному дослідженні;
* сформулювати межу узагальнення хоча б одного центрального Claim;
* пояснити, як буде інтерпретовано null або negative result;
* показати найбільш критичний `evidence debt`;
* захистити логіку `Evidence Architecture v1` як частину цілісної програми дисертаційного дослідження.


## Підсумок

Після ЛР6 дисертаційна програма повинна перейти від системи *запланованих досліджень* до системи *запланованих і перевірюваних доказів*.

Ключовий результат роботи:

$$
\text{Research Design}
\rightarrow
\text{Evidence Architecture}
\rightarrow
\text{Validated Claims}.
$$

При цьому «Validated Claim» не означає абсолютної істини. Це означає, що аспірант:

- явно визначив Evidence;
- спроєктував перевірки його надійності;
- ідентифікував загрози валідності;
- визначив межі узагальнення;
- не формулює висновок сильніше, ніж дозволяє доказова база.

Наступний крок курсу -- зробити цей шлях *відтворюваним і простежуваним*: від даних, коду, версій і дослідницького журналу до конкретного результату та наукового твердження.
