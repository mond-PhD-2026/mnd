**Мета**: сформувати для власного дисертаційного дослідження практичний план відтворюваності, простежуваності та дослідницької доброчесності; визначити правила керування даними, кодом, середовищем, версіями, експериментами та науковими результатами; побудувати матрицю простежуваності ключових наукових тверджень і підготувати декларацію використання штучного інтелекту.


::: callout-note
## Примітка

ЛР7 продовжує наскрізну роботу над *власною дисертацією*. Вхідними артефактами є актуальні результати ЛР1--ЛР6, передусім `Methodological Architecture`, `Research Design`, `Claim--Evidence--Validation Matrix` та `Evidence Architecture`.

Тут не потрібно створювати «ідеальний репозиторій готової дисертації». Потрібно спроєктувати правила, за якими подальше дослідження виконуватиметься так, щоб походження ключових результатів можна було відновити, а суттєві процедури -- повторити й перевірити.

Якщо певний елемент не застосовується до вашого типу дослідження, позначте `N/A` і коротко обґрунтуйте.
:::


## Що ви будете вміти?

Після виконання лабораторної роботи ви зможете:

* розрізняти *відтворюваність* (reproducibility), *відтворення незалежною реалізацією* (replication) і *простежуваність* (traceability);
* проєктувати структуру дослідницького репозиторію;
* документувати походження даних (data provenance), коду, конфігурацій і результатів;
* формувати `Reproducibility Protocol`;
* будувати `Claim Traceability Matrix`;
* організовувати `Research Log`;
* виявляти ризики fabrication, falsification, selective reporting, p-hacking і HARKing;
* формувати `AI Usage Statement`;
* оцінювати власний *борг відтворюваності* (reproducibility debt).


## Короткі теоретичні відомості

### 1. Відтворюваність і простежуваність

Для обчислювального дослідження:

$$
R=f(D,C,P,E),
$$

де $D$ -- дані, $C$ -- код, $P$ -- параметри та конфігурація, $E$ -- обчислювальне середовище.

Отже, одного файлу з кодом недостатньо. Потрібно встановити версію даних, стан коду, параметри, середовище та зв'язок результату з конкретним науковим твердженням.

*Простежуваність* (traceability) означає можливість пройти від Claim назад до його доказової основи:

$$
\text{Claim}\leftarrow\text{Evidence}\leftarrow\text{Result}\leftarrow
\text{Experiment}\leftarrow\text{Data + Code + Configuration}.
$$

Саме тому результат ЛР6 -- `Claim--Evidence--Validation Matrix` -- є природним входом до ЛР7.


Замініть шаблон власною картою. На ній мають бути видимі входи, версії коду/процедур, конфігурація, експеримент, результат, валідація, Claim, журналювання рішень та AI-assisted компонент -- якщо він є.


### Завдання 14. Сформуйте підсумковий `Reproducibility, Traceability & Integrity Plan v1`

| Компонент | Рішення | Артефакт / механізм | Коли оновлюється? | Критерій готовності |
|---|---|---|---|---|
| Repository structure | ... | ... | ... | ... |
| Data provenance | ... | ... | ... | ... |
| Code/version control | ... | ... | ... | ... |
| Environment | ... | ... | ... | ... |
| Experiment tracking | ... | ... | ... | ... |
| Claim traceability | ... | ... | ... | ... |
| Research log | ... | ... | ... | ... |
| Integrity safeguards | ... | ... | ... | ... |
| AI usage | ... | ... | ... | ... |
| Openness/access | ... | ... | ... | ... |

Завершіть роботу відповіддю:

> *Які конкретні властивості мого дослідницького процесу дозволять іншому досліднику перевірити походження та обґрунтованість ключових наукових тверджень?*


## Що потрібно подати на захист

У зошиті мають бути представлені:

1. актуалізований перелік вхідних артефактів ЛР1--ЛР6;
2. структура дослідницького репозиторію;
3. `Data & Artifact Provenance Map v1`;
4. `Reproducibility Protocol v1`;
5. політика версіонування та ідентифікації;
6. `Claim Traceability Matrix v1`;
7. `Research Log Template v1`;
8. `Research Integrity Risk Register v1`;
9. правила confirmatory/exploratory analysis;
10. `AI Usage Statement v1`;
11. політика відкритості та доступу;
12. аудит `Reproducibility Debt`;
13. `Reproducibility & Traceability Map v1`;
14. `Reproducibility, Traceability & Integrity Plan v1`;
15. короткий підсумковий висновок.

Усі артефакти мають стосуватися *власного дисертаційного дослідження*.


## Критерії оцінювання

Максимальна оцінка за лабораторну роботу -- **7 балів**.

| Компонент | Бали |
|---|---:|
| Структура репозиторію, provenance та політика версіонування | 1.25 |
| `Reproducibility Protocol v1` | 1.25 |
| `Claim Traceability Matrix v1` та інтегрована карта | 1.50 |
| Research log і аудит ризиків доброчесності | 1.00 |
| `AI Usage Statement v1` | 1.00 |
| Політика відкритості, reproducibility debt і цілісність плану | 1.00 |
| **Разом** | **7.00** |

::: callout-note
## Захист роботи

Під час захисту потрібно показати *один повний ланцюг простежуваності* для ключового майбутнього Claim: від джерела даних/іншого входу та версії коду до результату, валідації й наукового твердження.

Також потрібно пояснити один найбільш суттєвий ризик дослідницької доброчесності для власної дисертації та спосіб його контролю.
:::


## Контрольні запитання

1. Чим відтворюваність відрізняється від відтворення незалежною реалізацією?
2. Чому відкритий код не гарантує відтворюваність?
3. Що таке data provenance?
4. Які компоненти потрібно зафіксувати для відновлення стану експерименту?
5. Яку роль відіграють Git commit та experiment ID?
6. Чому один `random seed` не гарантує стійкість результату?
7. Що таке Claim Traceability Matrix?
8. Що означає traceability gap?
9. Навіщо потрібен research log за наявності Git?
10. Чим selective reporting відрізняється від коректного model selection?
11. У чому полягає p-hacking?
12. Що таке HARKing?
13. Як коректно працювати з гіпотезою, що виникла після аналізу?
14. Чому негативні та нульові результати важливі для доброчесності?
15. Які ризики виникають при використанні LLM у дослідницькому pipeline?
16. Що потрібно включити до AI Usage Statement?
17. Чому потрібно зберігати AI output, якщо він впливає на Evidence?
18. Чим FAIR відрізняється від повністю відкритих даних?
19. Що таке reproducibility debt?
20. Як пов'язати Claim з конкретними даними, кодом і валідацією?


## Підсумок

Результатом ЛР7 є *план керування доказовим походженням майбутніх результатів дисертації*:

$$
\text{Source}
\rightarrow
\text{Data / Artifact}
\rightarrow
\text{Versioned Procedure}
\rightarrow
\text{Experiment}
\rightarrow
\text{Result}
\rightarrow
\text{Validation}
\rightarrow
\text{Claim}
\rightarrow
\text{Contribution}.
$$

На наступному етапі ця система разом з артефактами ЛР1--ЛР6 стане входом до *фінального методологічного каркаса дисертаційного дослідження* та другої контрольної точки перегляду.


## Рекомендовані джерела

1. National Academies of Sciences, Engineering, and Medicine. *Reproducibility and Replicability in Science*. Washington, DC: The National Academies Press, 2019. DOI: 10.17226/25303.
2. Peng R. D. Reproducible Research in Computational Science. *Science*. 2011. Vol. 334, No. 6060. P. 1226--1227. DOI: 10.1126/science.1213847.
3. Sandve G. K., Nekrutenko A., Taylor J., Hovig E. Ten Simple Rules for Reproducible Computational Research. *PLoS Computational Biology*. 2013. Vol. 9, No. 10. e1003285. DOI: 10.1371/journal.pcbi.1003285.
4. Wilkinson M. D. et al. The FAIR Guiding Principles for scientific data management and stewardship. *Scientific Data*. 2016. Vol. 3. Article 160018. DOI: 10.1038/sdata.2016.18.
5. Nosek B. A. et al. Promoting an open research culture. *Science*. 2015. Vol. 348, No. 6242. P. 1422--1425. DOI: 10.1126/science.aab2374.
6. Munafò M. R. et al. A manifesto for reproducible science. *Nature Human Behaviour*. 2017. Vol. 1. Article 0021. DOI: 10.1038/s41562-016-0021.
7. Wilson G. et al. Good enough practices in scientific computing. *PLoS Computational Biology*. 2017. Vol. 13, No. 6. e1005510. DOI: 10.1371/journal.pcbi.1005510.
8. Kerr N. L. HARKing: Hypothesizing After the Results are Known. *Personality and Social Psychology Review*. 1998. Vol. 2, No. 3. P. 196--217. DOI: 10.1207/s15327957pspr0203_4.
9. Simmons J. P., Nelson L. D., Simonsohn U. False-Positive Psychology: Undisclosed Flexibility in Data Collection and Analysis Allows Presenting Anything as Significant. *Psychological Science*. 2011. Vol. 22, No. 11. P. 1359--1366. DOI: 10.1177/0956797611417632.
10. Pineau J. et al. Improving Reproducibility in Machine Learning Research: A Report from the NeurIPS 2019 Reproducibility Program. *Journal of Machine Learning Research*. 2021. Vol. 22, No. 164. P. 1--20.
