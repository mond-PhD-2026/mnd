## Примітка

Попередньо опрацювати матеріали **Лекції 2 «Від наукової проблеми до архітектури дисертації»**.

ЛР2 є **прямим продовженням ЛР1**. Не потрібно формулювати нову тему або починати дослідницький задум заново. Вихідними матеріалами є ваші:

- `Research Problem v1`;
- `Conceptual Framework v1`;
- результати аудиту проблематики з ЛР1.

На цьому етапі ми перетворюємо їх на *методологічну архітектуру* (*methodological architecture*) дисертації.

Усі формулювання мають статус **робочих версій**. У ЛР3--ЛР4 вони будуть перевірятися та уточнюватися на основі систематичного аналізу наукової літератури.

## Що ви будете вміти?

Після виконання лабораторної роботи ви зможете:

* відрізняти проблемну ситуацію, наукову проблему та дослідницьку прогалину (*research gap*);
* уточнювати наукову проблему без передчасної декларації наукової новизни;
* формулювати об'єкт, предмет і мету як взаємопов'язані компоненти дослідження;
* декомпозувати проблему в систему дослідницьких питань (*research questions*, RQ);
* визначати, де потрібні перевірювані гіпотези (*hypotheses*), а де достатньо дослідницьких питань;
* формулювати задачі як кроки отримання нового знання, а не перелік технічних операцій;
* будувати попередню карту наукових тверджень (*Claim Map*);
* визначати, які докази (*evidence*) потенційно потрібні для кожного твердження;
* виконувати пряму та зворотну перевірку логічної узгодженості архітектури;
* сформувати `Methodological Architecture v1` як артефакт другого етапу власної PhD-програми.

## Короткі теоретичні відомості

### 1. Архітектура дисертації -- це не зміст розділів

На цьому етапі нас цікавить не майбутня нумерація розділів дисертації, а **логіка дослідження**.

Мінімальний ланцюг має вигляд:

$$
\text{Проблема}
\rightarrow
\text{Питання}
\rightarrow
\text{Дослідження}
\rightarrow
\text{Докази}
\rightarrow
\text{Твердження}
\rightarrow
\text{Внесок}.
$$

Кожна стрілка означає змістовий зв'язок, який аспірант повинен уміти пояснити.

Якщо певна задача не відповідає жодному дослідницькому питанню, або заявлений внесок не можна простежити до конкретних доказів, архітектура потребує перегляду.

### 2. Від проблеми до дослідницької прогалини

**Наукова проблема** (*research problem*) описує науково значуще невідоме.

**Дослідницька прогалина** (*research gap*) конкретизує, чого саме бракує в наявному науковому знанні для усунення цього невідомого.

На ЛР2 ми формуємо лише `Research Gap v1`.

Це важливе обмеження:

> `Research Gap v1` -- **робоча гіпотеза про стан наукового поля**, а не вже доведений факт.

Систематичну доказову перевірку цієї конструкції ми почнемо в ЛР3.

### 3. Дослідницьке питання як міст між проблемою та дослідженням

Дослідницьке питання (*research question*, RQ) переводить загальну проблему в конкретне невідоме, для якого можна визначити спосіб отримання доказів.

$$
RQ_i
\rightarrow
\text{потрібні докази}
\rightarrow
\text{потенційне твердження}.
$$

Хороше RQ:

* безпосередньо випливає з проблеми;
* не має тривіальної відповіді;
* допускає дослідницьку процедуру;
* не містить наперед заданого «правильного» результату;
* може привести до обґрунтованого наукового твердження.

### 4. Гіпотеза потрібна не завжди

Гіпотеза (*hypothesis*) доцільна тоді, коли можна сформулювати конкретне перевірюване та спростовуване припущення.

Наприклад:

> За однакового раннього вікна спостереження поєднання поведінкових і часових ознак забезпечує вищу якість класифікації, ніж використання лише поведінкових ознак.

Таке твердження можна перевірити.

Натомість питання:

> Які групи ознак залишаються інформативними за скорочення історії спостережень?

може залишатися RQ без попереднього припущення про відповідь.

Отже, **не потрібно штучно створювати гіпотезу для кожного RQ**.

### 5. Задача дослідження -- не технічна операція

Порівняйте:

| Формулювання | Тип |
|---|---|
| Зібрати набір даних | технічна операція |
| Реалізувати модель | технічна операція |
| Побудувати графіки | технічна операція |
| Встановити, які групи ознак зберігають інформативність за скорочення вікна спостереження | дослідницька задача |
| Експериментально визначити межі застосовності методу | дослідницька задача |

Технічні операції необхідні, але вони є **засобами виконання дослідницьких задач**.

### 6. Карта тверджень

Карта тверджень (*Claim Map*) -- це попередня система наукових тверджень, які дисертація потенційно повинна мати право захищати після отримання достатніх доказів.

На ЛР2 ми будуємо лише `Claim Map v1`.

Тому формулюємо не:

> «Доведено, що метод X є кращим...»

а:

> «Попереднє твердження C1: метод X потенційно забезпечує ...; для перевірки потрібні ...»

Це захищає дослідницький дизайн від підміни дослідження бажаним результатом.


За потреби змініть кількість вузлів. Схема повинна відображати **вашу реальну логіку**, а не структуру прикладу.

### Крок 12. Сформулюйте `Expected Contributions v1`

Сформулюйте 1--3 очікувані внески.

| ID | Очікуваний внесок | Які твердження його підтримують? | У чому потенційна новизна? |
|---|---|---|---|
| EC1 | | | |
| EC2 | | | |
| EC3 | | | |

На цьому етапі використовуйте обережну модальність:

* «очікується встановити...»;
* «передбачається обґрунтувати...»;
* «потенційний внесок полягає у...»;
* «потребує подальшої перевірки...».

Не використовуйте «вперше», якщо це ще не підтверджено систематичним аналізом літератури.

### Крок 13. Побудуйте матрицю логічної узгодженості

Це центральний артефакт ЛР2.

| RQ | Гіпотеза | Дослідницька задача | Очікувані докази | Попереднє твердження | Очікуваний внесок |
|---|---|---|---|---|---|
| RQ1 | | | | | |
| RQ2 | | | | | |
| RQ3 | | | | | |
| RQ4 | | | | | |
| RQ5 | | | | | |

Після заповнення проведіть два аудити.

**Пряма простежуваність** (*forward traceability*):

$$
\text{Problem}
\rightarrow
RQ
\rightarrow
Task
\rightarrow
Evidence
\rightarrow
Claim
\rightarrow
Contribution.
$$

**Зворотна простежуваність** (*backward traceability*):

$$
\text{Contribution}
\rightarrow
Claim
\rightarrow
Evidence
\rightarrow
Task
\rightarrow
RQ
\rightarrow
Problem.
$$

Позначте всі місця, де зв'язок є слабким або потребує додаткового обґрунтування.

### Крок 14. Проведіть аудит невідповідностей

Перевірте власну архітектуру.

| Потенційний дефект | Виявлено? | Що потрібно змінити? |
|---|---|---|
| Невідповідність між проблемою та метою (*Problem--Aim mismatch*) | | |
| Невідповідність між метою та задачами (*Aim--Task mismatch*) | | |
| Невідповідність між задачею та твердженням (*Task--Claim mismatch*) | | |
| Невідповідність між доказами та твердженням (*Evidence--Claim mismatch*) | | |
| Необґрунтоване завищення внеску (*Contribution inflation*) | | |
| Ізольована задача (*orphan task*) | | |
| Ізольований внесок (*orphan contribution*) | | |

Для кожного виявленого дефекту внесіть зміну до відповідного артефакту.

### Крок 15. Зберіть `Methodological Architecture v1`

Після коригувань подайте фінальну версію компонентів ЛР2 в одній таблиці.

| Компонент | Версія після ЛР2 |
|---|---|
| `Research Problem v2` | |
| `Research Gap v1` | |
| `Object v1` | |
| `Subject v1` | |
| `Aim v1` | |
| `RQ System v1` | |
| `Hypotheses v1` -- за потреби | |
| `Research Tasks v1` | |
| `Claim Map v1` | |
| `Expected Contributions v1` | |

Ця таблиця -- **не резюме для звіту**, а контрольна збірка вашої поточної архітектури.

## 16. Архітектура як система версій

Після ЛР2 компоненти не стають «остаточними».

Зафіксуйте журнал змін:

| Компонент | Було | Стало | Причина зміни | Що ще потрібно перевірити |
|---|---|---|---|---|
| Research Problem | v1 | v2 | | |
| Research Gap | -- | v1 | | |
| Object / Subject | -- | v1 | | |
| Aim | -- | v1 | | |
| RQ System | -- | v1 | | |
| Claim Map | -- | v1 | | |

У ЛР3 до цього журналу повернемося після побудови протоколу пошуку, матриці доказів і карти сучасного стану досліджень.

## 17. Артефакти як система версій

На другому етапі PhD компоненти повинні мати явну версійність:

- `Research Problem v2`;
- `Research Gap v1`;
- `Object / Subject v1`;
- `Aim v1`;
- `RQ System v1`;
- `Hypotheses v1` -- за потреби;
- `Research Tasks v1`;
- `Claim Map v1`;
- `Expected Contributions v1`;
- `Methodological Architecture v1`.

Після ЛР3--ЛР4 частина цих артефактів буде переглянута на основі систематичної доказової бази.

::: callout-important
Зміна формулювань після отримання нових доказів є нормальною частиною розвитку дослідницької програми. Методологічна помилка -- не зміна версії, а приховування того, **чому** вона змінилася.
:::

# 18. Мінімальна методологічна архітектура дисертації v1

На цьому етапі всі компоненти повинні утворювати одну систему.

```{mermaid}
%%{init: {'theme': 'base', 'themeVariables': {
  'fontSize': '16px',
  'fontFamily': 'Trebuchet MS, Verdana, Arial, sans-serif',
  'lineColor': '#6366f1',
  'edgeLabelBackground': '#ffffff'
}}}%%
flowchart TD
    PS("🌍 Проблемна ситуація")
    P("❓ Наукова проблема v2")
    G("🕳️ Дослідницька прогалина v1")
    O("🎯 Об'єкт / предмет v1")
    A("🧭 Мета v1")
    RQ("❓ Система RQ v1")
    H("💡 Гіпотези v1")
    T("🔬 Дослідницькі задачі v1")
    ST("🧪 Окремі дослідження")
    E("🔎 Очікувані докази")
    C("💬 Карта тверджень v1")
    CONTR("🌱 Очікувані внески v1")

    PS --> P
    P --> G
    P --> O
    G --> A
    O --> A
    A --> RQ
    RQ --> H
    RQ --> T
    H --> T
    T --> ST --> E --> C --> CONTR
    C -. "перевірка узгодженості" .-> RQ

    classDef early fill:#fef3c7,stroke:#d97706,stroke-width:2px,color:#78350f
    classDef design fill:#e0e7ff,stroke:#6366f1,stroke-width:2px,color:#312e81
    classDef evidence fill:#dbeafe,stroke:#2563eb,stroke-width:2px,color:#1e3a8a
    classDef result fill:#dcfce7,stroke:#16a34a,stroke-width:2px,color:#14532d

    class PS,P,G,O,A early
    class RQ,H,T,ST design
    class E,C evidence
    class CONTR result
```

# 19. Самоперевірка архітектури

Перед завершенням ЛР2 дайте короткі відповіді.

1. Чи сформульована проблема як дефіцит знання, а не як потреба щось створити?
2. Чи відрізняю я наукову проблему від дослідницької прогалини?
3. Чи предмет справді звужує об'єкт до того, що буде досліджуватися?
4. Чи мета спрямована на науковий результат, а не лише на артефакт?
5. Чи кожне RQ випливає з проблеми?
6. Чи кожна гіпотеза є спростовуваною?
7. Чи кожна задача підтримує хоча б одне RQ?
8. Чи можна для кожного попереднього твердження назвати необхідні докази?
9. Чи не перевищує очікуваний внесок те, що реально можна буде довести?
10. Чи можна простежити логіку в обидва боки -- від проблеми до внеску і назад?

Для кожної відповіді «ні» або «частково» внесіть зміни до відповідного артефакту.

# 20. Перехід до ЛР3

Після ЛР2 ви маєте не остаточну дисертаційну тему, а:

$$
\boxed{
\text{Methodological Architecture}_{v1}
}
$$

Наступний крок -- **не розширювати архітектуру новими красивими формулюваннями**, а почати її систематичну перевірку науковою літературою.

У ЛР3 потрібно буде побудувати:

1. протокол пошуку;
2. критерії включення та виключення джерел;
3. матрицю доказів;
4. карту сучасного стану досліджень;
5. основу для доказового перегляду `Research Gap v1`, RQ та попередніх тверджень.

Отже:

$$
\text{Architecture}_{v1}
\rightarrow
\text{Systematic Evidence}
\rightarrow
\text{Architecture}_{v2}.
$$
