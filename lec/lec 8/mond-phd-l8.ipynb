Ця лекція завершує курс і *змістовий модуль 4 «Відтворюваність, простежуваність і цілісність дисертаційного дослідження»*.

# 1. Від набору досліджень до дисертації

Дисертація не повинна бути просто хронологічним звітом:

> «спочатку я зробив A, потім B, потім C».

На рівні PhD потрібна інша логіка:

$$
\text{Problem}
\rightarrow
\text{Gap}
\rightarrow
\text{RQ/H}
\rightarrow
\text{Research}
\rightarrow
\text{Evidence}
\rightarrow
\text{Claims}
\rightarrow
\text{Contributions}.
$$

Отже, *цілісність дисертації* визначається не кількістю експериментів, моделей, статей чи програмних артефактів, а тим, наскільки вони підтримують узгоджену систему наукових тверджень.

## 3.1. Claim не обов'язково дорівнює розділу дисертації

Структура аргументації та структура рукопису -- різні речі.

Один Claim може підтримуватися:

* теоретичним результатом із одного розділу;
* експериментом з іншого;
* зовнішньою валідацією з третього.

І навпаки, один розділ може містити докази для кількох Claims.

Тому спочатку будуємо *логіку знання*, а вже потім -- композицію тексту.


# 4. Карта доказів дисертації

У ЛР6 ми будували `Claim--Evidence--Validation Matrix`. Тепер її потрібно розглядати на рівні всієї дисертації.

Позначимо множину Claims:

$$
C=\{C_1,C_2,\ldots,C_n\},
$$

а множину доказів:

$$
E=\{E_1,E_2,\ldots,E_m\}.
$$

Відношення підтримки можна подати матрицею:

$$
A_{ij}=
\begin{cases}
1, & E_j \text{ підтримує } C_i,\\
0, & \text{інакше}.
\end{cases}
$$

На практиці бінарного зв'язку часто недостатньо. Доцільно додавати:

* тип Evidence;
* силу підтримки;
* спосіб Validation;
* Threats to Validity;
* статус доказу;
* provenance.


## 4.1. Evidence Map як діагностичний інструмент

Карта доказів потрібна не для красивої схеми. Вона дозволяє виявити:

### Claim без достатнього Evidence

$$
C_i \not\Leftarrow E
$$

Це *evidence gap*.

### Evidence без Claim

Є експеримент або результат, але незрозуміло, яке наукове твердження він підтримує.

Це ознака можливого *research drift* -- дослідження починає накопичувати результати, які не працюють на його основну аргументацію.

### Один Evidence використовується для надто багатьох Claims

Це не завжди помилка, але потребує перевірки, чи не перевищує сила Claims силу доказу.



## 5.1. Внесок -- це не список створених речей

Недостатньо написати:

* розроблено алгоритм;
* створено програмну систему;
* сформовано dataset;
* побудовано модель.

Це можуть бути *артефакти* або *результати*.

Науковий внесок потребує відповіді:

> Яке нове знання отримано завдяки цьому артефакту або результату?

Наприклад, не:

> «Розроблено нову архітектуру класифікатора».

А методологічно сильніше:

> «Встановлено, що включення X до архітектури Y за умов Z забезпечує стійке покращення властивості Q порівняно з baseline B, що підтверджено ...».

Конкретне формулювання залежить від фактичних результатів, але структура показує перехід від *artifact claim* до *knowledge claim*.


# 6. Узгодження задач дисертації з результатами

Одна з найважливіших фінальних перевірок:

$$
\text{Research Task}
\rightarrow
\text{Method}
\rightarrow
\text{Result}
\rightarrow
\text{Claim}
\rightarrow
\text{Contribution}.
$$

Для кожної задачі потрібно перевірити:

| Елемент | Контрольне питання |
|---|---|
| Task | Яке дослідницьке питання вона закриває? |
| Method | Чому цей метод придатний? |
| Result | Який результат реально отримано? |
| Evidence | Чим результат підтверджено? |
| Claim | Що саме дозволено стверджувати? |
| Contribution | Який внесок формує це твердження? |

Якщо ланцюг розривається, формальна наявність задачі у вступі не робить її методологічно виправданою.


## 6.1. Перевірка повноти

Нехай є $k$ задач:

$$
T=\{T_1,\ldots,T_k\}.
$$

Для кожної $T_i$ має існувати хоча б один результат $R_j$:

$$
\forall T_i \; \exists R_j: T_i \rightarrow R_j.
$$

Але потрібна й зворотна перевірка:

$$
\forall R_j \; \exists T_i: T_i \rightarrow R_j,
$$

якщо результат претендує на місце в основній аргументації дисертації.

Інакше маємо або *незакриту задачу*, або *результат-сироту*.


# 7. Межі застосовності та узагальнюваність

Сильна дисертація не лише пояснює, *де результат працює*, а й визначає, *де твердження перестає бути обґрунтованим*.

Для Claim доцільно фіксувати область застосовності:

$$
\Omega_C =
\{\text{data},\text{population},\text{domain},\text{conditions},\text{assumptions}\}.
$$

Тоді Claim має неявну форму:

$$
C \mid \Omega_C.
$$

Тобто твердження справедливе *за визначених умов*.


## 7.1. Узагальнюваність не означає універсальність

Наприклад, модель може бути перевірена:

* на одному dataset;
* у межах однієї платформи;
* для однієї мови;
* у конкретному часовому періоді.

У такому випадку формулювання:

> «метод ефективний для виявлення ...»

може бути надто сильним.

Коректніше сформулювати межі:

> «у досліджених умовах ...».

А потім окремо визначити, які Evidence потрібні для ширшого узагальнення.


# 8. Негативні та нульові результати в інтегрованій аргументації

Негативний результат не обов'язково є «невдалим експериментом».

Якщо дослідження коректно спроєктоване, результат:

$$
H_1 \text{ не отримала достатньої підтримки}
$$

може:

* звузити область застосовності методу;
* спростувати очікуване пояснення;
* показати нестійкість ефекту;
* виявити важливу умову;
* змінити Claim;
* сформувати нове дослідницьке питання.

Тому інтеграція результатів не повинна перетворюватися на *selection of success stories*.


# 9. Публікації та дисертація

Публікація і дисертація мають різні рівні композиції.

Окрема стаття може містити:

$$
\text{RQ}_i \rightarrow \text{Result}_i \rightarrow \text{Claim}_i.
$$

Дисертація повинна інтегрувати:

$$
\{C_1,C_2,\ldots,C_n\}
\rightarrow
\{\text{Contribution}_1,\ldots,\text{Contribution}_p\}.
$$

Тому набір статей автоматично не утворює цілісної дисертації.


## 9.1. Карта відповідності публікацій науковим твердженням

Корисний інструмент -- *Publication--Claim Map*.

| Publication | RQ / Task | Result | Claim | Contribution | Role |
|---|---|---|---|---|---|
| P1 | RQ1 | R1 | C1 | N1 | primary |
| P2 | RQ2 | R2 | C2 | N1 | primary |
| P3 | RQ1/RQ3 | R3 | C1/C3 | N1/N2 | supporting |

Карта дозволяє побачити:

* які Claims уже апробовані;
* які ще не представлені в публікаціях;
* чи не дублюють статті одна одну;
* як публікаційна траєкторія підтримує дисертаційну аргументацію.

При цьому *публікація не замінює Evidence*. Вона є формою наукової комунікації та апробації результату.


# 10. Простежуваність від дисертації до первинних артефактів

Після ЛР7 можемо побудувати повний ланцюг:

$$
\text{Contribution}
\leftarrow
\text{Claim}
\leftarrow
\text{Evidence}
\leftarrow
\text{Result}
\leftarrow
\text{Experiment}
\leftarrow
\text{Data + Code + Config + Environment}.
$$

Це означає, що сильне положення про наукову новизну повинно мати не лише красиве формулювання, а й *простежуване походження*.


```{mermaid}
%%{init: {'theme': 'base', 'themeVariables': {
  'fontSize': '17px',
  'fontFamily': 'Trebuchet MS, Verdana, Arial, sans-serif',
  'primaryColor': '#E8F3FF',
  'primaryBorderColor': '#3478A8',
  'secondaryColor': '#FFF0C9',
  'tertiaryColor': '#E7F6EC',
  'lineColor': '#536878'
}}}%%
flowchart RL
    N("🏆 Contribution") --> C("💡 Claim")
    C --> E("📊 Evidence")
    E --> R("📈 Result")
    R --> X("🧪 Experiment")
    X --> D("📦 Data")
    X --> K("💻 Code")
    X --> P("⚙️ Config")
    X --> V("🖥️ Environment")
```


# 11. Фінальний методологічний каркас

Наприкінці курсу методологічний каркас можна подати як систему:

$$
\mathcal{M}=
(P,G,Q,H,T,D,M,E,V,C,N,L,R),
$$

де:

* $P$ -- research problem;
* $G$ -- research gap;
* $Q$ -- research questions;
* $H$ -- hypotheses;
* $T$ -- research tasks;
* $D$ -- research design;
* $M$ -- methods;
* $E$ -- evidence architecture;
* $V$ -- validation;
* $C$ -- claims;
* $N$ -- contributions;
* $L$ -- limitations;
* $R$ -- reproducibility and traceability.

Цінність такої моделі не в символах. Вона показує, що жодний ключовий компонент не повинен існувати ізольовано.


## 11.1. Логічна замкненість

Методологічна архітектура є логічно замкненою, якщо можна пройти два маршрути.

### Вперед

$$
Problem \rightarrow Gap \rightarrow RQ \rightarrow Design \rightarrow Evidence \rightarrow Claim \rightarrow Contribution.
$$

### Назад

$$
Contribution \rightarrow Claim \rightarrow Evidence \rightarrow Experiment \rightarrow RQ \rightarrow Gap \rightarrow Problem.
$$

Якщо зворотний маршрут неможливий, внесок може бути декларативним.


# 12. Контрольна точка перегляду 2

Перша контрольна точка після ЛР4 відповідала на питання:

> Чи витримує початкова концепція перевірку сучасним станом досліджень?

Друга контрольна точка має інше питання:

> Чи утворюють усі спроєктовані компоненти одну доказово й логічно узгоджену програму дисертаційного дослідження?

Для кожного елемента знову корисні рішення:

* `KEEP` -- залишити;
* `MODIFY` -- уточнити;
* `REMOVE` -- вилучити;
* `ADD` -- додати відсутній компонент.

На цьому етапі `MODIFY` -- нормальний результат методологічного аудиту, а не ознака невдалої роботи.


## 12.1. Що перевіряємо

1. Problem ↔ Gap.
2. Gap ↔ RQ/H.
3. RQ/H ↔ Tasks.
4. Tasks ↔ Design/Methods.
5. Methods ↔ Data/Evidence.
6. Evidence ↔ Validation.
7. Evidence ↔ Claims.
8. Claims ↔ Contributions/Novelty.
9. Claims ↔ Limitations.
10. Claims ↔ Publications.
11. Claims ↔ Reproducibility/Traceability.

Тобто друга контрольна точка -- це *інтеграційний аудит*, а не повторення ЛР1--ЛР7.


# 13. Методологічний захист

Методологічний захист -- це не презентація теми дисертації.

Його предметом є *обґрунтованість дослідницької програми*.

Аспірант повинен уміти відповісти:

* чому проблема є науковою;
* чим доведена прогалина;
* чому RQ/H випливають із неї;
* чому обраний дизайн здатний дати потрібний Evidence;
* чому метрики та validation відповідають Claim;
* де проходять межі узагальнення;
* як забезпечено reproducibility і traceability;
* чому очікуваний Contribution є науковим, а не лише технічним.


## 13.1. Сила захисту -- у зв'язках

Слабкий захист складається з окремих відповідей:

> «ось моя мета»  
> «ось мої методи»  
> «ось моя модель».

Сильний захист показує залежності:

> «RQ1 випливає з Gap G1; для його перевірки потрібен Claim C1; C1 потребує Evidence E1 і E2; тому обрано Design D1; його валідність контролюється V1--V3».

Саме *причинно-логічний зв'язок між методологічними рішеннями* є предметом захисту.


# 14. Типові запитання на методологічному захисті

### «У чому тут наукова проблема, а не інженерна задача?»

Потрібно показати невизначеність знання, а не лише необхідність створити систему.

### «Звідки ви знаєте, що це research gap?»

Потрібно перейти до систематизованої Evidence Base.

### «Чому саме цей метод?»

Відповідь повинна виходити з типу Claim та дизайну, а не з популярності технології.

### «Чому ця метрика доводить ваш Claim?»

Потрібно показати зв'язок construct → operationalization → metric.

### «Що спростувало б вашу гіпотезу?»

Якщо відповідь неможлива, гіпотеза може бути нефальсифікованою або сформульованою надто розмито.

### «Де межі вашого твердження?»

Потрібно назвати область застосовності та threats to validity.

### «Як можна перевірити цей результат?»

Потрібно перейти до reproducibility/replication/traceability plan.


# 15. Приклад: ботоподібна активність у дискусіях

Розглянемо спрощений наскрізний приклад.

### Problem

Поведінкові сигнали ботоподібної активності можуть змінюватися між тематиками та часовими періодами, що ускладнює стійке виявлення.

### Gap

Недостатньо встановлено, які групи поведінкових ознак зберігають дискримінативність за domain shift.

### RQ

Які поведінкові ознаки забезпечують найбільш стійке розділення класів за зміни тематичного домену?

### Design

Кілька тематичних підмножин, temporal/domain holdout, baseline models, ablation study.

### Evidence

Розподіли метрик на повторних запусках, effect sizes, confidence intervals, ablation results, external/domain validation.

### Claim

Певна група ознак демонструє стійкішу дискримінативність у визначених умовах.

### Contribution

Нове знання стосується не факту створення класифікатора, а встановленої залежності між групою поведінкових ознак, зміною домену та стійкістю виявлення.


## 15.1. Що станеться, якщо результат інший?

Припустімо, очікувана група ознак *не* виявилася стійкою.

Тоді можливий Claim:

> стійкість, очікувана з попередніх досліджень, не відтворюється за умов $Z$.

Це може бути науково значущим результатом, якщо:

* дизайн коректний;
* Evidence достатній;
* альтернативні пояснення перевірені;
* межі Claim сформульовані обережно.

Отже, дисертаційна аргументація не повинна бути побудована так, щоб «успіх» був єдиним допустимим результатом.


Розділи повинні підтримувати аргументацію, а не просто віддзеркалювати послідовність виконаних технічних робіт.


# 17. «Червоні прапорці» фінальної архітектури

Перевірте, чи немає таких ознак:

* тема ширша за фактичний предмет;
* мета не досягається сукупністю задач;
* задача не має результату;
* результат не відповідає жодній задачі;
* Claim не має достатнього Evidence;
* Evidence не проходить Validation;
* Contribution сформульовано як назву артефакту;
* Novelty Claim сильніше за Evidence;
* межі застосовності не визначені;
* публікації не покривають ключові Claims;
* негативні результати приховано;
* неможливо простежити походження ключового результату;
* у фінальній аргументації з'явилися поняття, яких не було в Problem/RQ/Design.


# 18. Мінімальна інтегрована матриця дисертації

Для практичної роботи корисно звести ключову логіку в одну таблицю.

| Gap | RQ/H | Task | Method/Design | Evidence | Validation | Claim | Contribution | Limitation | Publication |
|---|---|---|---|---|---|---|---|---|---|
| G1 | RQ1 | T1 | D1 | E1,E2 | V1 | C1 | N1 | L1 | P1 |
| G2 | RQ2 | T2 | D2 | E3 | V2,V3 | C2 | N1 | L2 | P2 |
| ... | ... | ... | ... | ... | ... | ... | ... | ... | ... |

Це не заміна тексту дисертації. Це *діагностична модель її методологічної цілісності*.


# 19. Що означає «дисертація готова методологічно»?

На першому семестрі першого року PhD мова, звичайно, не йде про готову дисертацію.

Коректніше говорити про *готовність дослідницької програми до систематичного виконання*.

Це означає, що аспірант:

* знає, яку проблему досліджує;
* має доказово обґрунтовану робочу Gap;
* сформував RQ/H;
* розуміє, які Claims планує перевірити;
* знає, який Evidence для цього потрібний;
* спроєктував Design і Validation;
* визначив очікувані Contributions;
* усвідомлює Limitations;
* має план reproducibility, traceability та integrity;
* готовий змінювати початкову архітектуру, якщо майбутні Evidence цього вимагатимуть.


# 20. Фінальна модель курсу

Увесь курс можна звести до одного циклу:

```{mermaid}
%%{init: {'theme': 'base', 'themeVariables': {
  'fontSize': '18px',
  'fontFamily': 'Trebuchet MS, Verdana, Arial, sans-serif',
  'primaryColor': '#E8F3FF',
  'primaryBorderColor': '#3478A8',
  'secondaryColor': '#FFF0C9',
  'tertiaryColor': '#E7F6EC',
  'lineColor': '#536878'
}}}%%
flowchart LR
    P("🎯 Problem") --> G("🧩 Gap")
    G --> Q("❓ RQ / H")
    Q --> D("🧭 Design")
    D --> E("📊 Evidence")
    E --> V("🛡️ Validation")
    V --> C("💡 Claim")
    C --> N("🏆 Contribution")
    N --> S("📘 Dissertation")
    S --> Z("🗣️ Defense")
    V -.-> Q
    E -.-> G
    C -.-> D
```

# 21. Головний принцип

У методологічно сильній дисертації можна поставити запитання до будь-якого ключового твердження:

> *Чому я маю в це повірити?*

І отримати простежувану відповідь:

$$
Claim
\rightarrow
Evidence
\rightarrow
Validation
\rightarrow
Research\ Artifact
\rightarrow
Protocol.
$$

А до будь-якого експерименту можна поставити зворотне запитання:

> *Навіщо він потрібний?*

І пройти:

$$
Experiment
\rightarrow
Evidence
\rightarrow
Claim
\rightarrow
Contribution
\rightarrow
Research\ Gap.
$$

Якщо обидва маршрути працюють, окремі дослідницькі дії починають утворювати *цілісну програму PhD-дослідження*.


# 22. Підготовка до ЛР8

У фінальній лабораторній роботі потрібно буде інтегрувати артефакти ЛР1--ЛР7.

Ключовими результатами стануть:

1. `Final Methodological Framework v1`;
2. `Dissertation Evidence Map v1`;
3. `Publication--Claim Map v1`;
4. `Scope & Generalizability Map v1`;
5. `Integration Audit / Checkpoint 2`;
6. короткий пакет для *методологічного захисту*.

ЛР8 не повинна бути механічним копіюванням попередніх таблиць. Її задача -- перевірити, чи утворюють вони *одну систему*.


# 23. Контрольні запитання

1. Чим інтеграція результатів відрізняється від їх простого переліку?
2. Чому Result не тотожний Claim?
3. Що додається до Result під час переходу до Claim?
4. Для чого потрібна Claim Map?
5. Що таке evidence gap?
6. Що може означати Evidence без пов'язаного Claim?
7. Чим науковий внесок відрізняється від артефакту?
8. Як перевірити відповідність задач дисертації результатам?
9. Що таке «результат-сирота»?
10. Навіщо явно визначати межі застосовності Claim?
11. Чому generalizability не означає універсальність?
12. Яку роль можуть відігравати негативні результати?
13. Чому набір публікацій автоматично не утворює дисертацію?
14. Для чого потрібна Publication--Claim Map?
15. Яким має бути повний ланцюг traceability для Contribution?
16. Що означає логічна замкненість методологічного каркаса?
17. Чим контрольна точка 2 відрізняється від контрольної точки 1?
18. Що саме захищається під час методологічного захисту?
19. Які «червоні прапорці» свідчать про розрив методологічної архітектури?
20. Що означає готовність дослідницької програми наприкінці першого семестру PhD?


# 24. Підсумок

Методологія дисертаційного дослідження завершується не вибором методу і не отриманням метрики.

Її інтегрована логіка має вигляд:

$$
\text{Problem}
\rightarrow
\text{Gap}
\rightarrow
\text{RQ/H}
\rightarrow
\text{Design}
\rightarrow
\text{Evidence}
\rightarrow
\text{Validation}
\rightarrow
\text{Claim}
\rightarrow
\text{Contribution}.
$$

А на рівні дисертації до цього додаються:

$$
\text{Limitations}
+
\text{Reproducibility}
+
\text{Traceability}
+
\text{Publications}
+
\text{Defense}.
$$

Завдання дослідника -- не «зібрати» ці компоненти, а зробити їх *логічно узгодженими та доказово пов'язаними*.

Саме така система створює основу для подальшої багаторічної роботи над дисертацією.
