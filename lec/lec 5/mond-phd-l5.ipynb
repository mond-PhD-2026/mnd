# Вступ

У комп'ютерних науках легко почати дослідження з інструмента:

> «Я використаю Transformer».

> «Я навчу класифікатор».

> «Я проведу експеримент на трьох наборах даних».

> «Я порівняю точність моделей».

Але *метод не повинен передувати дослідницькому питанню*.

На рівні PhD логіка має бути протилежною:

$$
\boxed{
\text{Research Question}
\rightarrow
\text{Scientific Claim}
\rightarrow
\text{Research Design}
\rightarrow
\text{Method}
\rightarrow
\text{Data}
\rightarrow
\text{Analysis}
\rightarrow
\text{Evidence}
}
$$

Тобто ми не шукаємо задачу для улюбленого алгоритму. Ми проєктуємо таку систему дослідження, яка дозволить отримати *доказ, релевантний конкретному науковому твердженню*.


# Тема 5. Дизайн дослідження та система методів

## 1. Що таке дизайн дослідження

*Дизайн дослідження* (research design) -- це логічна конструкція, яка визначає, *яким способом дослідницьке питання буде пов'язане з даними, методами аналізу, процедурою перевірки та майбутнім висновком*.

Його не слід зводити до:

- переліку алгоритмів;
- списку бібліотек;
- опису набору даних;
- схеми програмної системи;
- переліку експериментів без пояснення їх ролі.

Дизайн відповідає на сильніше питання:

> *Чому саме така процедура дозволяє отримати доказ, достатній для підтримки або спростування конкретного Claim?*


Дизайн дослідження визначає не лише *що робити*, а й *яка логіка виведення* пов'язує спостереження з висновком.


# 2. Від задачі дослідження до типу наукового твердження

У НРП зафіксована магістральна логіка:

$$
\text{Задача дослідження}
\rightarrow
\text{тип наукового твердження}
\rightarrow
\text{план дослідження}
\rightarrow
\text{метод}
\rightarrow
\text{дані}
\rightarrow
\text{аналіз}.
$$

Це важливо, тому що різні твердження потребують *різних типів доказів*.

Наприклад, твердження:

> «Ознака $X$ асоційована з ботоподібною поведінкою»

не тотожне твердженню:

> «Ознака $X$ спричиняє зміну поведінки».

А твердження:

> «Модель прогнозує клас із заданою якістю»

не є автоматично поясненням механізму, який породжує спостережуваний ефект.


## 2.1. П'ять типів дослідницьких задач

Для практичного проєктування корисно розрізняти щонайменше п'ять типів.

| Тип задачі | Основне питання | Типовий результат | Чого не можна автоматично стверджувати |
|---|---|---|---|
| *Опис* (description) | Що спостерігається? | структура, розподіл, патерн | чому це виникло |
| *Асоціація* (association) | Що з чим пов'язано? | статистичний/структурний зв'язок | причинність |
| *Прогнозування* (prediction) | Наскільки добре можна передбачити $Y$? | прогнозна якість | механізм або причинний ефект |
| *Пояснення* (explanation) | Який механізм або структура пояснює явище? | обґрунтована пояснювальна модель | причинність без відповідного дизайну |
| *Причинне виведення* (causal inference) | Що зміниться в $Y$, якщо змінити $X$? | оцінка причинного ефекту | універсальність поза умовами ідентифікації |

::: callout-warning
## Важливо

*Тип аналізу повинен відповідати силі майбутнього твердження.*

Висока прогнозна точність не перетворює прогнозну модель на причинну. Кореляція не стає причинністю через складність алгоритму.
:::


## 2.2. Один об'єкт -- різні дослідницькі питання

Розглянемо умовну тему виявлення ботоподібної активності.

### Опис

> Які типи поведінкових патернів спостерігаються в дискусіях?

### Асоціація

> Які поведінкові ознаки статистично пов'язані з ботоподібною активністю?

### Прогнозування

> Наскільки точно за поведінковими та текстовими ознаками можна передбачити ботоподібну активність?

### Пояснення

> Які групи ознак і взаємодії між ними пояснюють стійкість класифікації при зміні тематики?

### Причинне виведення

> Як зміна конкретного механізму взаємодії впливає на поширення повідомлень за інших рівних умов?

Тема може залишатися тією самою, але *дизайн, дані, методи і допустимі висновки* будуть різними.


# 3. Типи досліджень у комп'ютерних науках

У дисертації з комп'ютерних наук часто поєднуються кілька типів досліджень.

## 3.1. Теоретичне дослідження

Спрямоване на отримання формальних результатів:

- означень;
- властивостей;
- теорем;
- оцінок складності;
- доведень коректності;
- формальних меж застосовності.

Типовий доказ -- *логічне або математичне доведення*.

## 3.2. Експериментальне дослідження

Передбачає контрольоване варіювання умов і спостереження наслідків.

Типові компоненти:

- фактор;
- рівні фактора;
- контрольна умова;
- повторення;
- вимірювана відповідь;
- протокол аналізу.

## 3.3. Модельне дослідження

Вивчає систему через її абстрактну модель:

- математичну;
- імітаційну;
- агентну;
- графову;
- імовірнісну;
- логічну.

Ключове питання -- *наскільки висновки про модель переносяться на досліджувану систему*.

## 3.4. Обчислювальне дослідження

Використовує обчислювальний експеримент для дослідження алгоритмів, моделей, даних або складних систем.

У сучасних Computer Science дисертаціях саме тут часто виникають:

- benchmark experiments;
- порівняння моделей;
- абляції;
- симуляції;
- аналіз чутливості;
- експерименти з перенесенням між доменами.


## 3.5. Дисертація як система взаємопов'язаних досліджень

Сильна дисертація не обов'язково належить до одного типу.

```{mermaid}
%%{init: {'theme': 'base', 'themeVariables': {
  'fontSize': '17px',
  'fontFamily': 'Trebuchet MS, Verdana, Arial, sans-serif',
  'primaryColor': '#E8F3FF',
  'primaryBorderColor': '#3478A8',
  'secondaryColor': '#FFF0C9',
  'tertiaryColor': '#E7F6EC',
  'lineColor': '#536878'
}}}%%
flowchart TB
    G("🧩 Research Gap") --> RQ("❓ Research Questions")
    RQ --> T("📐 Теоретичне<br/>дослідження")
    RQ --> X("🧪 Експериментальне<br/>дослідження")
    RQ --> M("🧩 Модельне<br/>дослідження")
    RQ --> C("💻 Обчислювальне<br/>дослідження")
    T --> E("🔎 Сукупна доказова база")
    X --> E
    M --> E
    C --> E
    E --> CL("💬 Scientific Claims")
```

1. формального означення;
2. теоретичного аналізу властивостей;
3. обчислювального порівняння з базовими рівнями;
4. аналізу стійкості;
5. перевірки на незалежних даних.


# 4. Метод не дорівнює дизайну дослідження

Розглянемо твердження:

> «У роботі використано Random Forest, XGBoost і Transformer».

Це *перелік методів*, але не дизайн.

Дизайн з'являється тоді, коли ми можемо пояснити:

- яке Research Question перевіряється;
- який Claim очікується;
- які дані потрібні;
- чому вибрано саме ці порівняння;
- що є базовим рівнем;
- які фактори контролюються;
- яка метрика відповідає твердженню;
- який аналіз дозволить відокремити альтернативні пояснення;
- за яких умов висновок буде визнано підтриманим.


## 4.1. Приклад: від алгоритму до дослідницького дизайну

### Слабка постановка

> Навчити Transformer і порівняти його з LSTM.

### Сильніша постановка

> Перевірити, чи зберігає Transformer перевагу над LSTM при зменшенні повноти спостережень за однакових умов навчання, однакового бюджету даних і спільного протоколу оцінювання.

Тут уже визначено:

- *фактор* -- повнота спостережень;
- *порівнювані підходи* -- Transformer і LSTM;
- *контроль* -- однакові умови;
- *об'єкт вимірювання* -- зміна якості;
- *майбутній Claim* -- залежність від рівня інформаційної неповноти.

Отже, алгоритми є лише частиною ширшої логіки.


# 5. Операціоналізація: як зробити поняття вимірюваним

Багато дисертаційних понять не спостерігаються безпосередньо:

- надійність;
- стійкість;
- пояснюваність;
- складність;
- якість взаємодії;
- ботоподібність;
- ситуаційна обізнаність;
- семантична подібність.

*Операціоналізація* (operationalization) -- це перехід від теоретичного конструкта до спостережуваних змінних, індикаторів і процедур вимірювання.

Схематично:

$$
\text{Construct}
\rightarrow
\text{Operational Definition}
\rightarrow
\text{Variable / Indicator}
\rightarrow
\text{Measurement}.
$$


## 5.1. Приклад операціоналізації

Припустимо, Claim стосується *стійкості моделі до тематичного зсуву*.

Слово «стійкість» саме по собі не є вимірюванням.

Можлива операціоналізація:

$$
\Delta M = M_{\text{in-domain}} - M_{\text{cross-domain}},
$$

де $M$ -- заздалегідь визначена метрика якості.

Тоді менше значення $\Delta M$ може інтерпретуватися як вища стійкість до зміни домену.

Але навіть тут потрібно зафіксувати:

- яку саме метрику $M$ використовуємо;
- які домени вважаємо різними;
- як формуємо train/test;
- чи однакові класи;
- чи порівнювані обсяги даних;
- чи повторюється експеримент;
- як оцінюємо невизначеність.

Отже, операціоналізація -- це частина *дизайну доказу*.


# 6. Дані як частина дослідницького дизайну

Набір даних (dataset) не є нейтральним контейнером.

Він визначає, *про яку популяцію, середовище, період і умови ми взагалі можемо робити висновки*.

Для кожного набору даних потрібно описати щонайменше:

1. походження;
2. одиницю спостереження;
3. цільову популяцію;
4. часовий період;
5. спосіб формування вибірки;
6. критерії включення та виключення;
7. розмір;
8. пропуски;
9. спосіб отримання міток;
10. потенційні зміщення (bias);
11. етичні та правові обмеження;
12. межі репрезентативності.


## 6.1. Популяція, вибірка та одиниця аналізу

Три поняття не слід змішувати.

*Популяція* (population) -- сукупність об'єктів, щодо якої ми хочемо робити висновок.

*Вибірка* (sample) -- фактично досліджувана підмножина.

*Одиниця аналізу* (unit of analysis) -- елемент, для якого формуються змінні та висновки.

Наприклад, у дослідженні соціальних медіа одиницею аналізу може бути:

- повідомлення;
- користувач;
- дискусія;
- часовий інтервал;
- ребро графа взаємодій;
- ціла мережа.

::: callout-warning
## Типова помилка

Якщо train/test містять повідомлення тих самих користувачів, а Claim сформульовано про здатність моделі узагальнюватися на *нових користувачів*, звичайний випадковий поділ може створити витік інформації (data leakage).
:::


# 7. Вибірка та стратегія поділу даних

Поділ даних -- це не лише технічна операція.

Він повинен моделювати *майбутню ситуацію застосування або перевірки наукового твердження*.

| Claim | Більш релевантний поділ |
|---|---|
| узагальнення на нові об'єкти з тієї самої популяції | випадковий поділ за незалежними об'єктами |
| узагальнення на нових користувачів | group split за користувачем |
| прогноз майбутнього | часовий поділ |
| переносимість між темами | cross-domain split |
| переносимість між джерелами | leave-one-source-out |
| стабільність за зміни середовища | кілька незалежних зовнішніх тестів |

Таким чином:

$$
\boxed{
\text{Split Strategy}
\text{ повинна відповідати }
\text{Generalization Claim}
}
$$


# 8. Базові рівні порівняння

*Базовий рівень порівняння* (baseline) -- це обґрунтована точка відліку, відносно якої оцінюється новий результат.

Слабкий експеримент часто має лише одну «зручну» базову модель.

Сильніший дизайн може включати:

- простий наївний baseline;
- класичний метод;
- сучасний сильний метод;
- попередній варіант власного підходу;
- верхню або нижню межу, якщо її можна визначити;
- контрольну умову без ключового компонента.

Baseline потрібний не для того, щоб «перемогти слабку модель», а щоб відповісти:

> *Який саме додатковий ефект створює запропонований внесок?*


## 8.1. Baseline, benchmark і state of the art

Ці поняття пов'язані, але не тотожні.

| Поняття | Роль |
|---|---|
| *Базовий рівень порівняння* (baseline) | точка відліку в конкретному експерименті |
| *Еталонне завдання/набір* (benchmark) | стандартизована основа для порівняння |
| *Сучасний рівень* (state of the art) | найсильніші релевантні результати або підходи, відомі з доказової бази |

Новий метод не обов'язково має перевершити всі state-of-the-art результати, щоб мати науковий внесок.

Внесок може полягати, наприклад, у:

- меншій обчислювальній складності;
- кращій стійкості;
- новому поясненні;
- роботі за слабших припущень;
- встановленні меж застосовності;
- виявленні ефекту, який попередні benchmark-порівняння приховували.


# 9. Контрольні умови

*Контрольна умова* (control condition) потрібна, щоб відокремити ефект досліджуваного фактора від альтернативних пояснень.

Наприклад, якщо до моделі додано новий модуль і якість зросла, можливі пояснення:

1. модуль реалізує корисний механізм;
2. просто збільшилася кількість параметрів;
3. змінився режим навчання;
4. використано інший preprocessing;
5. випадково вдаліший seed;
6. змінилася вибірка.

Тому сильний дизайн прагне змінювати *один концептуально важливий фактор*, контролюючи решту.


# 10. Експериментальний протокол

*Експериментальний протокол* (experimental protocol) -- це заздалегідь визначена процедура проведення дослідження, достатня для однозначного розуміння того, *що саме було зроблено і чому*.

Мінімально він має визначати:

- дослідницьке питання;
- гіпотезу або Claim;
- дані;
- фактори та контрольні умови;
- baselines;
- процедуру поділу;
- preprocessing;
- навчання/налаштування;
- метрики;
- кількість повторень;
- правило агрегації результатів;
- план статистичного аналізу;
- критерій підтримки або відхилення робочої гіпотези;
- програмне та обчислювальне середовище.


## 10.1. Протокол як захист від «підгонки історії»

Якщо правила аналізу визначаються *після* перегляду результатів, виникає ризик несвідомо вибрати:

- найкращу метрику;
- найвигідніший split;
- найвдаліший seed;
- лише успішні експерименти;
- зручну підгрупу;
- формулювання гіпотези, що відповідає вже побаченим даним.

Тому доцільно розрізняти:

$$
\text{Exploratory Analysis}
\neq
\text{Confirmatory Analysis}.
$$

*Розвідувальний аналіз* (exploratory analysis) допомагає породжувати гіпотези.

*Підтверджувальний аналіз* (confirmatory analysis) перевіряє заздалегідь сформульовані гіпотези за визначеним протоколом.

Обидва типи корисні, але їх не слід маскувати один під одного.


# 11. Фактори, змінні та конфаундери

Для експериментальної логіки важливо розрізняти:

- *незалежну змінну* (independent variable) -- фактор, який змінюємо або порівнюємо;
- *залежну змінну* (dependent variable) -- вимірюваний результат;
- *контрольовані змінні* (control variables) -- умови, які прагнемо утримувати сталими;
- *конфаундер* (confounder) -- фактор, пов'язаний і з досліджуваною умовою, і з результатом, який може створити альтернативне пояснення.

Наприклад:

> Модель A краща за модель B.

Якщо A навчалась на більшому обсязі даних, ми не знаємо, що саме створило перевагу:

$$
\text{архітектура}
\quad \text{чи} \quad
\text{більший обсяг даних}.
$$

Це проблема *ідентифікації ефекту*.


Ключовий принцип: *test set не повинен перетворюватися на інструмент багаторазового вибору моделі*.


## 12.1. Перехресна перевірка і повторення

Перехресна перевірка (cross-validation) корисна для оцінювання варіативності результату та ефективнішого використання обмежених даних.

Але вона не вирішує автоматично проблеми:

- залежних спостережень;
- часової структури;
- групового витоку;
- доменного зсуву;
- неправильного формування ознак до split.

Отже, питання не лише:

> «Чи використали ми cross-validation?»

а:

> *«Чи відповідає схема перевірки структурі даних і силі Claim?»*


# 13. Дизайн для асоціації та пояснення

Якщо мета -- встановити зв'язок між змінними, потрібно визначити:

1. які змінні вимірюються;
2. яка форма зв'язку очікується;
3. які альтернативні фактори можуть пояснити асоціацію;
4. які припущення має модель;
5. чи достатній обсяг і діапазон даних;
6. як оцінюється невизначеність.

Для пояснювального Claim цього ще недостатньо.

Потрібно показати, *чому знайдена структура має змістовне пояснення*, а не є лише статистичною закономірністю.


# 14. Дизайн для причинного виведення

Причинне питання має іншу форму:

> *Що сталося б з результатом $Y$, якби ми змінили $X$, залишивши інші релевантні умови незмінними?*

Це контрфактична логіка.

Позначимо потенційні результати:

$$
Y(1), \qquad Y(0).
$$

Індивідуальний причинний ефект:

$$
\tau_i = Y_i(1) - Y_i(0).
$$

Але для одного й того самого об'єкта ми не можемо одночасно спостерігати $Y_i(1)$ і $Y_i(0)$.

Саме тому причинне виведення потребує спеціального дизайну та припущень.


## 14.1. Рандомізований і спостережний дизайн

У рандомізованому експерименті випадкове призначення умов допомагає зробити групи порівнюваними.

У спостережному дослідженні такого механізму немає, тому потрібне явне обґрунтування:

- які конфаундери враховано;
- які залишилися невиміряними;
- чому ідентифікаційні припущення правдоподібні;
- які перевірки чутливості виконано.

::: callout-warning
## Важливо

Ні нейронна мережа, ні велика вибірка, ні низьке $p$-value самі по собі не створюють причинного дизайну.
:::


# 15. Аналітичний план

*Аналітичний план* (analysis plan) визначає, як із первинних результатів буде отримано доказ для конкретного Claim.

Він має відповісти:

- які змінні аналізуються;
- які перетворення допустимі;
- які метрики обчислюються;
- як агрегуються повторення;
- як оцінюється невизначеність;
- які порівняння є основними;
- які -- додатковими;
- як працюємо з пропусками;
- як працюємо з викидами;
- які припущення перевіряємо;
- як інтерпретуємо негативний або нульовий результат.

У лекції 6 ця логіка буде продовжена через *метрики, розмір ефекту, статистичну і практичну значущість, робастність, абляції, чутливість та загрози валідності*.


# 16. План роботи з даними

План роботи з даними (data plan) є частиною дизайну, а не технічним додатком.

Корисно зафіксувати ланцюг:

$$
\text{Source}
\rightarrow
\text{Acquisition}
\rightarrow
\text{Cleaning}
\rightarrow
\text{Transformation}
\rightarrow
\text{Split}
\rightarrow
\text{Analysis}
\rightarrow
\text{Derived Data}.
$$

Для кожного кроку потрібно знати:

- що змінюється;
- навіщо;
- чи може процедура створити bias;
- чи використовує вона інформацію з test set;
- чи можна її відтворити;
- яка версія даних є вхідною і вихідною.

Це стане основою для теми 7, де ми окремо розглядатимемо *походження та історію даних* (data provenance) і простежуваність.


# 17. Дослідницька програма як набір Study Units

Для складної дисертації зручно проєктувати не один «великий експеримент», а систему окремих досліджень.

Кожне дослідження (Study Unit) має власну функцію:

| Study | Питання | Claim | Дизайн | Основний доказ |
|---|---|---|---|---|
| S1 | Які ознаки стійкі? | C1 | cross-domain observational | оцінка стабільності |
| S2 | Чи покращує відбір за стійкістю переносимість? | C2 | controlled computational experiment | порівняння з baselines |
| S3 | Який компонент створює ефект? | C3 | ablation design | зміна результату після вилучення компонента |
| S4 | Де метод перестає працювати? | C4 | sensitivity/stress testing | межі застосовності |

Така структура робить дисертацію *системою доказів*, а не послідовністю несумісних експериментів.


## 17.1. Матриця «Claim -- Study -- Evidence»

Ще до отримання результатів можна побудувати робочу матрицю.

| Claim ID | Що плануємо стверджувати | Study | Дані | Порівняння | Evidence |
|---|---|---|---|---|---|
| C1 | існує стійка група ознак | S1 | D1--D3 | між доменами | розподіл $\Delta M$ |
| C2 | запропонований відбір підвищує переносимість | S2 | D1--D3 | baselines | ефект + невизначеність |
| C3 | компонент R є необхідним для ефекту | S3 | D1--D3 | full vs ablated | ablation effect |
| C4 | ефект зберігається до рівня шуму $q$ | S4 | perturbed D | рівні $q$ | sensitivity curve |

Ця матриця є попередником повної матриці *«твердження -- доказ -- валідація»*, яку будемо будувати у темі 6.


# 18. Приклад повного переходу від Gap до Design

Припустимо, після теми 4 сформульовано робочу прогалину:

> Наявні моделі демонструють високу якість у межах окремих тематичних доменів, однак недостатньо встановлено, які поведінкові ознаки зберігають дискримінативність за тематичного зсуву.

Тоді:

### Research Question

> Які групи поведінкових ознак зберігають дискримінативність при переході між тематично різними дискусіями?

### Робочий Claim

> Ознаки групи $F_s$ демонструють менше падіння дискримінативності при cross-domain transfer, ніж ознаки групи $F_c$.

### Дизайн

- кілька тематично різних доменів;
- навчання на одному/кількох доменах;
- тестування на невикористаному домені;
- однакова модель-класифікатор;
- однаковий preprocessing;
- однаковий бюджет даних;
- повторення для кількох seed;
- порівняння груп ознак;
- оцінка зміни метрики та невизначеності.

Тут метод випливає з Claim, а не навпаки.


## 18.1. Схема прикладу

```{mermaid}
%%{init: {'theme': 'base', 'themeVariables': {
  'fontSize': '17px',
  'fontFamily': 'Trebuchet MS, Verdana, Arial, sans-serif',
  'primaryColor': '#E8F3FF',
  'primaryBorderColor': '#3478A8',
  'secondaryColor': '#FFF0C9',
  'tertiaryColor': '#E7F6EC',
  'lineColor': '#536878'
}}}%%
flowchart LR
    G("🧩 Gap:<br/>невідома стійкість ознак") --> RQ("❓ RQ:<br/>які ознаки стійкі?")
    RQ --> C("💬 Claim:<br/>Fs стійкіші за Fc")
    C --> D("🧭 Cross-domain<br/>design")
    D --> X("🧪 Однаковий<br/>класифікатор")
    D --> S("🔀 Domain-based<br/>split")
    D --> M("📊 Δ метрики +<br/>невизначеність")
    X --> E("🔎 Evidence")
    S --> E
    M --> E
```


# 19. Типові помилки дизайну дослідження

## Помилка 1. Метод шукає проблему

> «Хочу використати LLM, тому треба знайти, що ним досліджувати».

## Помилка 2. Claim ширший за дизайн

Один dataset і випадковий split використовуються для твердження про загальну переносимість методу.

## Помилка 3. Немає сильного baseline

Новий метод порівнюється лише зі слабким або застарілим рішенням.

## Помилка 4. Змінюється одразу багато факторів

Інша архітектура, інші дані, preprocessing і tuning -- неможливо встановити джерело ефекту.

## Помилка 5. Test set використовується для розробки

Після кожної зміни моделі дивимося на test metric і продовжуємо налаштування.

## Помилка 6. Метрика вибрана за звичкою

Accuracy використовується для незбалансованої задачі без пояснення, який аспект Claim вона вимірює.

## Помилка 7. Порівняння несиметричне

Один метод ретельно налаштований, інший запускається з default settings.

## Помилка 8. Висновок причинний, дизайн асоціативний

Зі статистичного зв'язку робиться висновок про вплив.


# 20. Тест узгодженості Research Design

Для кожного запланованого дослідження поставимо десять питань.

1. *RQ:* на яке дослідницьке питання воно відповідає?
2. *Claim:* яке твердження може бути підтримане результатом?
3. *Design:* чому обраний дизайн відповідає цьому Claim?
4. *Data:* яку популяцію представляють дані?
5. *Unit:* яка одиниця аналізу?
6. *Comparison:* відносно чого оцінюється ефект?
7. *Control:* які альтернативні пояснення контролюються?
8. *Analysis:* як саме результат перетворюється на Evidence?
9. *Boundary:* де закінчуються межі допустимого висновку?
10. *Traceability:* чи можна простежити зв'язок від Gap до майбутнього Evidence?

Якщо відповідь на кілька пунктів відсутня, експеримент може бути технічно виконуваним, але методологічно ще не спроєктованим.


# 21. Що аспірант повинен мати після цієї лекції

До лабораторної роботи 5 необхідно підготувати:

1. актуальну версію *Research Gap*;
2. 1--3 основні *Research Questions*;
3. робочі *Contribution Claims / Novelty Claims v1*;
4. для кожного Claim -- бажаний тип доказу;
5. перелік Study Units, необхідних для отримання цих доказів;
6. для кожного Study Unit -- тип дослідження;
7. джерела та структуру даних;
8. одиницю аналізу та стратегію вибірки;
9. baselines і контрольні умови;
10. попередній експериментальний та аналітичний протокол.

У ЛР5 ці компоненти будуть інтегровані у *Research Design v1* та *Experimental/Analytical Plan v1*.

# 23. Підсумки

1. *Дизайн дослідження* (research design) -- це логічний міст між дослідницьким питанням і доказом.
2. Проєктування доцільно вести за ланцюгом: *задача → Claim → дизайн → метод → дані → аналіз → Evidence*.
3. Опис, асоціація, прогнозування, пояснення і причинне виведення потребують різних типів доказів.
4. У комп'ютерних науках теоретичні, експериментальні, модельні та обчислювальні дослідження можуть утворювати єдину дослідницьку програму.
5. Метод є інструментом усередині дизайну, а не заміною дизайну.
6. Дані, вибірка, одиниця аналізу і стратегія поділу визначають межі майбутнього Claim.
7. Baselines і контрольні умови потрібні для інтерпретації джерела ефекту.
8. Експериментальний протокол має бути визначений настільки явно, щоб відділити заплановану перевірку від післяекспериментальної інтерпретації.
9. Аналітичний план визначає, як результат перетворюється на Evidence.
10. Дисертаційну програму доцільно проєктувати як систему *Study Units*, кожен з яких підтримує конкретні наукові твердження.

Після цієї теми магістральна логіка курсу набуває вигляду:

$$
\text{Research Gap}
\rightarrow
\text{RQ}
\rightarrow
\text{Claim}
\rightarrow
\text{Research Design}
\rightarrow
\text{Study}
\rightarrow
\text{Evidence}.
$$

Наступна тема поставить ще сильніше питання: *коли отриманий Evidence є достатньо надійним, щоб підтримати Claim?*
