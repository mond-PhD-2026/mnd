# Вступ

Фраза *"я переглянув багато статей"* ще нічого не говорить про якість літературного аналізу.

Для дисертаційного дослідження важливо мати можливість відповісти принаймні на п'ять питань:

1. *Де і як шукали?*
2. *За якими правилами відбирали джерела?*
3. *Які саме твердження підтверджує кожне джерело?*
4. *Наскільки надійними та узгодженими є знайдені докази?*
5. *Який стан знання випливає із сукупності цих доказів?*

Отже, літературний огляд рівня PhD -- це не бібліографічний список і не послідовний переказ статей. Це *дослідницька процедура*, яка має власний протокол, критерії, артефакти та логіку синтезу.


# 2. Пошук як відтворювана дослідницька процедура

*Систематичний пошук (systematic search)* -- це пошук, правила якого визначені настільки явно, щоб інший дослідник міг зрозуміти, де, коли, що і за якими критеріями шукалося.

Це не означає, що кожна дисертація повинна містити повний систематичний огляд. Але пошукова частина дисертаційного дослідження повинна бути *прозорою та простежуваною*.

Мінімальний протокол пошуку має фіксувати:

| Компонент | Приклад |
|---|---|
| Дослідницьке питання | Які підходи забезпечують стійкість детекції ботоподібної активності до тематичного зсуву? |
| Джерела пошуку | Scopus, Web of Science, IEEE Xplore, ACM Digital Library |
| Час пошуку | дата виконання або інтервал |
| Пошукові поля | title, abstract, keywords |
| Пошуковий запит | логічна комбінація термінів |
| Часові межі | наприклад, 2019--2026 |
| Критерії включення | релевантність задачі, наявність експерименту тощо |
| Критерії виключення | дублікати, нерелевантний домен тощо |
| Додатковий пошук | прямий і зворотний ланцюговий пошук |


## 2.1. Від дослідницького питання до пошукових концептів

Пошуковий запит не варто будувати як одне дуже довге речення природною мовою. Спочатку виділяємо *концепти пошуку (search concepts)*.

### Приклад

Дослідницьке питання:

> Які методи виявлення ботоподібної активності зберігають якість за зміни тематики або часової структури соціомедійних даних?

Можемо виділити:

- *об'єкт*: social media bots, bot detection, coordinated behavior;
- *методичний аспект*: detection, classification, machine learning;
- *умова стійкості*: domain shift, topic shift, temporal shift, generalization, robustness.

Умовно:

$$
Q = C_1 \land C_2 \land C_3,
$$

де всередині одного концепту синоніми поєднуються оператором OR:

$$
C_1 = t_{11} \lor t_{12} \lor \ldots \lor t_{1m}.
$$

Тоді пошуковий запит має логічну структуру:

```text
("social bot" OR "bot detection" OR "coordinated behavior")
AND
("machine learning" OR classification OR detection)
AND
("domain shift" OR "topic shift" OR robustness OR generalization)
```

::: callout-warning
## Запит -- це модель предметної області

Надто вузький запит створює ризик пропустити релевантні роботи. Надто широкий -- породжує великий обсяг шуму. Тому пошуковий запит майже завжди потребує кількох пілотних ітерацій.
:::


## 2.2. Чутливість і специфічність пошуку

Інтуїтивно пошукову стратегію можна розглядати як задачу інформаційного пошуку.

Нехай:

- $R$ -- множина всіх релевантних робіт;
- $F$ -- множина знайдених робіт;
- $R \cap F$ -- релевантні роботи, які ми знайшли.

Тоді *повнота пошуку (recall)*:

$$
Recall = \frac{|R \cap F|}{|R|}.
$$

А *точність пошуку (precision)*:

$$
Precision = \frac{|R \cap F|}{|F|}.
$$

У реальному літературному пошуку множина $R$ повністю невідома, тому абсолютну повноту ми зазвичай обчислити не можемо. Але ця модель пояснює компроміс:

- розширення запиту підвищує шанс не пропустити важливу роботу;
- надмірне розширення різко збільшує кількість нерелевантних результатів.

Для дисертаційного дослідження небезпечно оптимізувати лише *precision*: дуже "чистий" список із 20 статей може приховувати систематичні пропуски.


# 3. Критерії включення та виключення

*Критерії включення (inclusion criteria)* і *критерії виключення (exclusion criteria)* задаються *до* остаточного відбору джерел.

Їхнє призначення -- зменшити довільність рішень дослідника.

### Приклад

*Включаємо*, якщо робота:

1. досліджує автоматизоване виявлення ботоподібної або координованої активності;
2. містить опис методу та експериментальної перевірки;
3. використовує реальні або обґрунтовано змодельовані соціомедійні дані;
4. містить достатню інформацію для вилучення ключових результатів.

*Виключаємо*, якщо:

1. робота є дублікатом;
2. повний текст недоступний і ключові результати неможливо перевірити;
3. "bot" використовується в іншому значенні;
4. публікація не містить власного дослідження, якщо оглядові роботи аналізуються окремо;
5. робота не відповідає визначеним часовим або мовним межам.

::: callout-important
## Критерії не повинні "підганяти" літературу під очікуваний висновок

Якщо правило відбору з'являється лише після того, як дослідник побачив небажані результати, виникає ризик *упередження відбору (selection bias)*.
:::


## 3.1. Етапи скринінгу

Відбір доцільно виконувати щонайменше у два проходи:

1. *первинний скринінг (title/abstract screening)* -- назва, анотація, ключові слова;
2. *повнотекстовий скринінг (full-text screening)* -- перевірка роботи за всіма критеріями.

Ланцюговий пошук особливо корисний для:

- знаходження класичних робіт, термінологія яких відрізняється від сучасної;
- простеження розвитку методу;
- виявлення серії робіт однієї дослідницької групи;
- перевірки того, як спільнота відреагувала на певний результат.

Але знайдені таким способом джерела проходять *ті самі критерії включення/виключення*.


# 5. Який тип огляду нам потрібен?

Не кожен огляд має однакову мету.

| Підхід | Основне питання | Результат |
|---|---|---|
| Систематичний огляд (systematic review) | Що свідчить сукупність відібраних досліджень щодо чітко сформульованого питання? | відтворюваний відбір + критичний синтез |
| Огляд обсягу предметного поля (scoping review) | Що взагалі досліджувалося, якими методами і де є незаповнені області? | карта поля |
| Наративний синтез (narrative synthesis) | Як узгодити різнорідні результати, які не можна коректно агрегувати чисельно? | структурована аргументація |
| Метааналіз (meta-analysis) | Який сукупний кількісний ефект випливає з порівнюваних досліджень? | агрегована оцінка ефекту |

Для дисертації з комп'ютерних наук часто потрібна *комбінація*:

> систематизований пошук → картування поля → критичний наративний синтез → за наявності порівнюваних кількісних результатів метааналіз.

::: callout-warning
## Метааналіз -- не "усереднення accuracy"

Якщо дослідження використовують різні датасети, різні розбиття, різні класи, різні метрики та експериментальні протоколи, пряме усереднення значень метрик може бути методологічно некоректним.
:::


# 6. Вилучення даних із джерел

Після відбору починається *вилучення даних (data extraction)*.

На цьому етапі ми переводимо неструктурований текст публікацій у порівнювану структуру.

Для досліджень у Computer Science корисно вилучати щонайменше:

| Група | Поля |
|---|---|
| Ідентифікація | автори, рік, DOI, venue |
| Дослідницька задача | проблема, research question |
| Дані | джерело, розмір, період, вибірка |
| Метод | алгоритм, модель, архітектура |
| Порівняння | baseline, competing methods |
| Протокол | train/test split, cross-validation, external validation |
| Метрики | F1, AUC, MAE, RMSE тощо |
| Результати | числові оцінки, ефекти, порівняння |
| Обмеження | threats to validity, limitations |
| Релевантність | яке наше твердження підтримує або заперечує |

Одна з найважливіших дисциплінарних навичок аспіранта -- відрізняти *результат, який реально показано*, від *інтерпретації автора*.


## 6.1. Простий приклад: що саме ми вилучаємо

Припустимо, у статті сказано:

> Модель A досягла F1 = 0.91 на Dataset X, перевищивши baseline B з F1 = 0.86.

Неправильне узагальнення:

> *Модель A краща за B.*

Коректніше:

> *У наведеному авторами експериментальному протоколі на Dataset X модель A показала F1 = 0.91 проти 0.86 для baseline B.*

Чому друга форма сильніша?

Бо вона зберігає *контекст доказу*:

$$
Evidence = Result + Conditions + Method + Source.
$$

Без умов отримання результату доказ легко перетворюється на надмірне узагальнення.


# 7. Матриця доказів

*Матриця доказів (evidence matrix)* -- центральний робочий артефакт цієї теми.

Вона зв'язує:

$$
\text{джерело} \rightarrow \text{метод} \rightarrow \text{дані} \rightarrow \text{результат} \rightarrow \text{обмеження} \rightarrow \text{наше твердження}.
$$

### Мінімальна структура

| ID | Джерело | Задача | Дані | Метод | Ключовий результат | Обмеження | Підтримує твердження |
|---|---|---|---|---|---|---|---|
| S01 | Author et al., 2024 | bot detection | Dataset A | Transformer | ... | один домен | C1 |
| S02 | Author et al., 2025 | cross-domain detection | Dataset B/C | GNN | ... | мала вибірка | C1, C2 |
| S03 | Author et al., 2023 | temporal robustness | Dataset D | RF | ... | короткий період | C2 |

Матриця потрібна не "для таблиці у дисертації". Це *інструмент мислення*, який дозволяє побачити:

- які твердження мають багато незалежних підтверджень;
- де докази суперечать один одному;
- які методи порівнювалися некоректно;
- які датасети домінують у літературі;
- які умови майже не досліджені.


## 7.1. Матриця "твердження -- джерело"

Для попереднього аналізу можна побудувати бінарну матрицю

$$
E = [e_{ij}],
$$

де

$$
e_{ij} =
\begin{cases}
1, & \text{якщо джерело } S_i \text{ містить релевантний доказ щодо } C_j,\\
0, & \text{інакше.}
\end{cases}
$$

Наприклад:

| Джерело | C1: тематичний зсув впливає на якість | C2: поведінкові ознаки стійкіші | C3: потрібна зовнішня валідація |
|---|---:|---:|---:|
| S1 | 1 | 0 | 1 |
| S2 | 1 | 1 | 0 |
| S3 | 0 | 1 | 1 |
| S4 | 1 | 0 | 0 |

Це ще не доводить істинність C1--C3. Кількість публікацій не дорівнює силі доказу. Але матриця допомагає перевірити *покриття аргументації*.


# 8. Якість доказів

Не всі джерела мають однакову доказову силу.

У Computer Science якість емпіричного доказу часто залежить від таких характеристик:

1. *відповідність дизайну поставленому питанню*;
2. *репрезентативність даних*;
3. *коректність baseline-порівнянь*;
4. *відсутність витоку даних (data leakage)*;
5. *коректність метрик*;
6. *оцінка невизначеності*;
7. *зовнішня валідація*;
8. *відтворюваність опису та артефактів*;
9. *аналіз обмежень і загроз валідності*.

Тому не варто вводити механічну формулу типу

$$
Quality = 0.2Q_1 + 0.2Q_2 + \ldots
$$

без методологічного обґрунтування ваг.

На ранньому етапі кориснішою може бути прозора якісна шкала:

- *висока довіра*;
- *помірна довіра*;
- *обмежена довіра*;

із явним поясненням причин такої оцінки.


## 8.1. Суперечливі результати -- це не проблема огляду

Припустимо:

- S1 і S2 показують, що метод A кращий за B;
- S3 показує, що B кращий за A.

Слабкий огляд обере "більшість":

$$
2 > 1 \Rightarrow A \text{ кращий}.
$$

Методологічно сильніший аналіз запитає:

- чи однакові датасети?
- чи однакова цільова змінна?
- чи однакові метрики?
- чи однакові умови train/test?
- чи однаковий часовий горизонт?
- чи є domain shift?
- чи відрізняється preprocessing?
- чи достатні розміри вибірок?

Тоді суперечність може перетворитися на нове знання:

> Перевага A над B спостерігається за умов $Z_1$, але не відтворюється за $Z_2$.

Саме такі *умовні твердження* часто є набагато ціннішими за спрощене "метод A найкращий".


# 9. Від матриці доказів до карти сучасного стану досліджень

*Карта сучасного стану досліджень (state-of-the-art map)* повинна показувати не список робіт, а структуру знання у предметній області.

Для кожного напряму доцільно фіксувати:

- які задачі розв'язуються;
- які класи методів домінують;
- на яких даних вони перевіряються;
- які результати є достатньо стійкими;
- де результати суперечливі;
- які методологічні обмеження повторюються;
- які питання залишаються відкритими.



## 9.1. Тематичний і методологічний синтез

Для дисертації корисно виконувати щонайменше два взаємодоповнювальні зрізи.

### Тематичний синтез (thematic synthesis)

Групуємо дослідження за *тим, що вони вивчають*:

- задачами;
- підпроблемами;
- типами даних;
- дослідницькими контекстами.

### Методологічний синтез (methodological synthesis)

Групуємо за *тим, як вони отримують знання*:

- класами моделей;
- дизайнами експериментів;
- способами формування вибірки;
- схемами валідації;
- baseline;
- метриками.

Це важливе розрізнення. Дві статті можуть досліджувати ту саму проблему, але давати різні висновки через різні методологічні рішення.


# 10. Достатність доказової бази

Питання *"скільки статей достатньо?"* не має універсальної числової відповіді.

Достатність визначається не лише кількістю, а *покриттям дослідницького питання*.

Корисно перевірити:

1. чи представлені основні напрями досліджень;
2. чи охоплено ключові методи;
3. чи враховано класичні та сучасні роботи;
4. чи присутні результати, що суперечать домінантній позиції;
5. чи простежено ключові твердження до першоджерел;
6. чи не залежить висновок від одного датасету або однієї дослідницької групи;
7. чи нові знайдені роботи суттєво змінюють карту категорій і висновків.

Останній пункт пов'язаний з ідеєю *насичення (saturation)*: додаткові джерела перестають істотно змінювати структуру синтезу.

Але насичення не є виправданням для припинення пошуку "коли набридло". Воно має оцінюватися відносно заздалегідь визначеного дослідницького питання і пошукової стратегії.


# 11. Пошук за підтримки штучного інтелекту

Штучний інтелект може бути корисним у літературному пошуку, але змінює не вимоги до доказів, а лише *спосіб навігації*.

*Пошук за підтримки ШІ (AI-assisted search)* доцільно використовувати для:

- генерації синонімів і пошукових концептів;
- первинного тематичного групування;
- пошуку потенційно релевантних праць;
- побудови чернеткової карти понять;
- формування питань до тексту статті;
- допомоги у вилученні структурованих полів.

Але модель не є першоджерелом.

```{mermaid}
%%{init: {'theme': 'base', 'themeVariables': {
  'fontSize': '17px',
  'fontFamily': 'Trebuchet MS, Verdana, Arial, sans-serif',
  'primaryColor': '#EAF4FF',
  'primaryBorderColor': '#3478A8',
  'secondaryColor': '#FFF2CC',
  'tertiaryColor': '#E8F7EE',
  'lineColor': '#526A7A'
}}}%%
flowchart LR
    AI("🤖 AI suggestion") --> ID("🔗 DOI / назва / автори")
    ID --> DB("📚 База / видавець")
    DB --> SRC("📄 Першоджерело")
    SRC --> V("✅ Перевірений доказ")
    AI -.->|не цитуємо як доказ| X("⛔ Неперевірене твердження")
```

## 11.1. Що саме потрібно перевіряти

Для кожної запропонованої ШІ публікації перевіряємо:

- чи існує вона;
- чи збігаються автори, назва, рік і venue;
- чи коректний DOI;
- чи справді текст містить приписаний йому результат;
- чи не вирваний висновок із контексту;
- чи відповідають дані, метрики та умови експерименту;
- чи не цитуємо вторинне джерело замість першоджерела.

Корисна модель:

$$
AI\ output \rightarrow Candidate\ evidence \rightarrow Verification \rightarrow Evidence.
$$

До завершення етапу *Verification* інформація залишається лише *кандидатом у доказ*.


# 12. Типові дефекти літературного аналізу

### Дефект 1. Пошук без протоколу

*"Я шукав у Google Scholar за різними ключовими словами."*

Проблема -- неможливо зрозуміти межі та відтворити процедуру.

### Дефект 2. Огляд "по авторах"

Кожен абзац присвячено окремій статті, але синтез відсутній.

### Дефект 3. Citation counting

Велика кількість цитувань використовується як заміна оцінки методологічної якості.

### Дефект 4. Cherry-picking

Відбираються переважно роботи, які підтверджують очікуваний висновок.

### Дефект 5. Порівняння непорівнюваного

Значення accuracy/F1/AUC із різних датасетів і протоколів ставляться в одну таблицю як прямі конкуренти.

### Дефект 6. Вторинне цитування

Твердження переноситься з оглядової статті без перевірки первинної роботи.

### Дефект 7. AI-generated bibliography

До списку потрапляють неперевірені або неіснуючі джерела, згенеровані LLM.


# 13. Наскрізний приклад

Повернімося до умовної теми:

> *Методи раннього виявлення ботоподібної активності у дискусіях соціальних медіа.*

Після лекції 2 маємо попереднє твердження:

> *C1:* наявні методи недостатньо стійкі до зміни тематики дискусії.

На цьому етапі C1 -- ще не встановлений факт.

Щоб перевести його в доказово обґрунтоване твердження, потрібно:

1. сформулювати пошукові концепти;
2. виконати систематизований пошук;
3. відібрати дослідження за явними критеріями;
4. вилучити дані про domain/topic shift;
5. зіставити експериментальні протоколи;
6. знайти як підтверджувальні, так і суперечливі результати;
7. оцінити обмеження;
8. сформулювати висновок лише в межах доказів.

Можливий результат синтезу:

> Наявні дослідження демонструють зниження якості частини моделей за міждоменного або часового перенесення, однак сила ефекту залежить від типу ознак, платформи та протоколу оцінювання.

Це значно сильніше за фразу:

> *Сучасні методи погано працюють на нових даних.*

Перше твердження має структуру, яку можна перевіряти і надалі уточнювати.

# Висновки

1. *Літературний огляд рівня PhD -- це дослідницька процедура*, а не перелік прочитаних джерел.
2. *Пошук має бути протокольованим*: бази, запити, часові межі та критерії відбору повинні бути явними.
3. *Критерії включення та виключення* зменшують довільність і ризик упередження відбору.
4. *Ланцюговий пошук* доповнює пошук у бібліографічних базах, але не скасовує критерії відбору.
5. Центральною одиницею синтезу є *наукове твердження та доказ*, а не окрема стаття.
6. *Матриця доказів* забезпечує простежуваність між джерелами, методами, даними, результатами, обмеженнями та нашими твердженнями.
7. *Суперечливі результати* потрібно пояснювати через умови отримання доказів, а не усувати голосуванням більшості.
8. *Карта сучасного стану досліджень* повинна відображати структуру знання: напрями, методи, дані, стійкі результати, суперечності, обмеження та відкриті питання.
9. *Пошук за підтримки ШІ* допустимий як інструмент навігації, але кожне доказове твердження перевіряється за першоджерелом.
10. Доказова база є механізмом *перевірки та уточнення Conceptual Framework v1*.
11. Дослідницька прогалина не декларується заздалегідь -- вона має випливати з критичного синтезу доказів.
