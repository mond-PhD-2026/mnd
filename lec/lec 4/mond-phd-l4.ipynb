# Вступ

На початку PhD-дослідження легко сформулювати щось, що *звучить новим*:

> «Запропоновано нову модель на основі Transformer для виявлення ботів».

Але з цього формулювання ще не випливає:

- що саме невідомо науці;
- чому наявних підходів недостатньо;
- яке нове знання буде отримано;
- чому заміна однієї архітектури на іншу є науковим внеском;
- якими доказами можна підтвердити новизну.

Для дисертаційного дослідження недостатньо твердження:

> *«Такого ще ніхто не робив».*

Потрібна сильніша конструкція:

> *«На підставі систематизованої доказової бази встановлено конкретну невирішену проблему; запропонований результат усуває або зменшує цю невизначеність і створює нове, перевірюване та науково значуще знання».*

Отже, у центрі цієї лекції -- не «красиве формулювання новизни», а *доказова логіка наукового внеску*.


# Тема 4. Дослідницька прогалина, наукова новизна та оригінальний науковий внесок

## 1. Чотири поняття, які не можна ототожнювати

У науковому тексті поруч часто вживаються слова *нове*, *новизна*, *оригінальність* і *внесок*. Вони пов'язані, але не тотожні.

| Поняття | Робоче розуміння | Контрольне питання |
|---|---|---|
| *Нове* (new) | те, чого раніше не було в певному контексті | Чи зустрічався такий об'єкт/варіант раніше? |
| *Новаторське* (novel) | нове рішення або знання, яке змістовно відрізняється від наявного стану | У чому суттєва відмінність? |
| *Наукова новизна* (scientific novelty) | властивість наукового результату містити раніше не встановлене знання | Що саме стало відомо завдяки дослідженню? |
| *Науковий внесок* (research contribution) | те, що дослідження додає до сукупного знання, методів, моделей, доказів або дослідницької практики | Що змінилося в знанні після цієї роботи? |

::: callout-warning
## Важливо

*Технічна новизна не гарантує наукової новизни.*

Новий програмний продукт, нова комбінація бібліотек, використання нової версії LLM або перенесення відомого алгоритму на інший набір даних (dataset) можуть бути інженерно корисними, але самі по собі ще не утворюють нового наукового знання.
:::


## 1.1. «Нове» не означає «науково нове»

Розглянемо кілька тверджень.

> *A.* Уперше застосовано модель X до набір даних (dataset) Y.

Це може бути *новим застосуванням*, але ми ще не знаємо, чи отримано нове знання.

> *B.* Встановлено, що перевага моделі X над моделлю Y виникає лише за певного рівня інформаційної неповноти, а зі збільшенням повноти спостережень ефект зникає.

Тут уже з'являється потенційний науковий результат: встановлена *залежність*, яку можна перевірити, відтворити та пояснити.

> *C.* Створено вебзастосунок для демонстрації моделей X та Y.

Це корисний *артефакт* (artifact), але його науковий статус залежить від того, *яке знання стало можливим завдяки його створенню та дослідженню*.

Отже:

$$
\text{New artifact} \not\Rightarrow \text{Scientific novelty}.
$$

Натомість більш корисною є логіка:

$$
\text{Evidence} + \text{Unresolved question} + \text{New result} + \text{Validation}
\Rightarrow \text{Potential scientific contribution}.
$$


# 2. Дослідницька прогалина

*Дослідницька прогалина* (research gap) -- це обґрунтована доказами невирішена, недостатньо досліджена, суперечливо досліджена або методологічно обмежена частина наукової проблеми, яка створює підставу для нового дослідження.

Ключове слово тут -- *обґрунтована*.

Прогалина не виникає тому, що автор:

- не знайшов статтю;
- не зустрів конкретну комбінацію технологій;
- вважає тему цікавою;
- хоче застосувати певний метод;
- побачив фразу *future work* в одній публікації.

Вона повинна випливати із *сукупності релевантних доказів*.


Отже, *research gap* -- не стартова декларація, а *результат аналітичного синтезу доказів*.


## 2.2. Типові форми прогалин

Дослідницька прогалина може мати різну природу.

### 1. Прогалина знання (knowledge gap)

Певна залежність, механізм або властивість недостатньо встановлені.

> Невідомо, які поведінкові ознаки зберігають дискримінативність при переході між тематично різними дискусіями.

### 2. Прогалина доказів (evidence gap)

Твердження існує, але доказова база недостатня, суперечлива або побудована на слабких протоколах.

> Різні роботи повідомляють протилежні результати щодо переваги Transformer над LSTM, однак використовують різні набори даних і непорівнювані експериментальні умови.

### 3. Методологічна прогалина (methodological gap)

Відомі методи не дозволяють надійно відповісти на важливе дослідницьке питання.

> Переважна частина оцінювання виконується на випадковому train/test split, що не дозволяє перевірити стійкість моделі до часової або тематичної зміни розподілу.

### 4. Прогалина контексту (contextual gap)

Відомі результати не перевірені в суттєво іншому середовищі, де теоретично очікується інша поведінка.

Тут потрібна обережність: *інший набір даних (dataset) сам по собі не створює наукову прогалину*.

### 5. Прогалина суперечності (inconsistency gap)

Якісні дослідження дають несумісні результати, а причина відмінностей не встановлена.

### 6. Прогалина пояснення (explanatory gap)

Ми знаємо, *що* ефект існує, але недостатньо знаємо, *чому* і *за яких умов* він виникає.


## 2.3. Прогалина ≠ відсутність роботи з точно такою самою назвою

Одна з найнебезпечніших логічних помилок:

> «Я не знайшов досліджень, де метод A застосовано до об'єкта B, отже це research gap».

Це лише *відсутність знайденого поєднання*.

Для перетворення її на дослідницьку прогалину потрібно показати щонайменше:

1. чому контекст B створює *науково суттєві* умови;
2. чому результати з інших контекстів не можна механічно перенести;
3. яка невизначеність залишається;
4. яке знання може усунути ця невизначеність.

Схематично:

$$
\text{Not found}(A,B) \neq \text{Research Gap}.
$$

Але:

$$
\text{Established limitation}
+ \text{important uncertainty}
+ \text{insufficient evidence}
\rightarrow \text{defensible gap}.
$$


# 3. Твердження про прогалину (Gap Claim)

*Твердження про прогалину* (Gap Claim) -- це коротке наукове твердження про те, *якого знання або доказу бракує* у сучасному стані досліджень.

Сильне Gap Claim має відповідати на три питання:

1. *Що вже відомо?*
2. *Що залишається невстановленим або недостатньо обґрунтованим?*
3. *Чому ця невизначеність є суттєвою?*

Корисна структура:

> *Відомо X*, однак наявні дослідження *обмежені Y*, унаслідок чого залишається невстановленим *Z*, що ускладнює/не дозволяє *Q*.

Це не шаблон для механічного копіювання, а спосіб перевірити логіку аргументу.


## 3.1. Від слабкого до сильнішого Gap Claim

### Слабко

> Недостатньо досліджено використання Transformer для виявлення ботів.

Проблеми:

- «недостатньо» не операціоналізовано;
- не зрозуміло, які дослідження вже є;
- не визначено, чого саме ми не знаємо;
- технологія підміняє проблему.

### Краще

> Наявні підходи до виявлення ботоподібної активності демонструють високу якість у межах окремих наборів даних, однак їхня стійкість до тематичного зсуву оцінюється нерегулярно та за непорівнюваними протоколами. Тому недостатньо встановлено, які групи поведінкових і текстових ознак зберігають дискримінативність за зміни тематики дискусії.

Тут уже можна побачити:

*відомий стан → обмеження доказів → конкретне невідоме.*


## 3.2. Доказова структура Gap Claim

```{mermaid}
%%{init: {'theme': 'base', 'themeVariables': {
  'fontSize': '18px',
  'fontFamily': 'Trebuchet MS, Verdana, Arial, sans-serif',
  'primaryColor': '#E8F3FF',
  'primaryBorderColor': '#3478A8',
  'secondaryColor': '#FFF0C9',
  'tertiaryColor': '#E7F6EC',
  'lineColor': '#536878'
}}}%%
flowchart LR
    K("📚 Що відомо?<br/>Evidence") --> L("⚠️ Яке обмеження<br/>виявлено?")
    L --> U("❓ Що залишається<br/>невідомим?")
    U --> I("🎯 Чому це<br/>важливо?")
    I --> G("🧩 Gap Claim")
```

# 4. Від прогалини до дослідницького питання

Прогалина описує *дефіцит знання*. Дослідницьке питання (Research Question, RQ) перетворює цей дефіцит на *перевірюваний запит*.

Наприклад:

*Gap Claim*

> Недостатньо встановлено, які групи ознак зберігають дискримінативність за тематичного зсуву.

*RQ*

> Які поведінкові та текстові групи ознак і якою мірою зберігають дискримінативність для виявлення ботоподібної активності при переході між тематично відмінними дискусіями?

Отже:

$$
\text{Gap} \rightarrow \text{RQ} \rightarrow \text{Research Design} \rightarrow \text{Evidence}.
$$

Цей ланцюг буде деталізовано в наступному змістовому модулі.


# 5. Науковий внесок

*Науковий внесок* (research contribution) -- це змістовне доповнення або уточнення наукового знання, яке виникає внаслідок дослідження та підтримується належними доказами.

Важливо розрізняти:

| Результат | Може бути внеском? | За якої умови |
|---|---:|---|
| Алгоритм | так | якщо обґрунтовано нову ідею/механізм і показано її ефект |
| Модель | так | якщо модель створює нове пояснення, властивість або перевірюваний результат |
| Архітектура | так | якщо структурна зміна має наукове обґрунтування та доведений ефект |
| набір даних (dataset) | так | якщо його створення усуває значущу доказову прогалину й забезпечує нові дослідницькі можливості |
| Метрика | так | якщо вона вимірює суттєву властивість, яку наявні метрики не відображають належно |
| Програмний артефакт | іноді | якщо артефакт є носієм/інструментом перевірки нового знання |
| Застосування відомого методу | іноді | якщо новий контекст породжує науково значущий результат, а не лише демонстрацію |


## 5.1. Contribution Claim

*Твердження про внесок* (Contribution Claim) описує, *що саме дослідження додає до наявного знання*.

Його не варто формулювати як:

> «Розроблено програму...»

або

> «Використано Transformer...»

Сильніше:

> «Встановлено умови, за яких поведінкові ознаки зберігають стійкість до тематичного зсуву, та запропоновано спосіб їх відбору, що враховує цю властивість».

Тут присутні два можливі рівні внеску:

- *емпіричний* -- встановлено закономірність;
- *методичний/методологічний* -- запропоновано спосіб використання цієї закономірності.

```{mermaid}
    C --> V("🔍 Comparison<br/>with prior work")
    V --> N("🏆 Novelty Claim")
```

Отже, на початку першого року PhD ми формуємо *робочу гіпотезу щодо новизни*, а не проголошуємо остаточно доведену новизну.


# 7. Gap Claim, Contribution Claim і Novelty Claim

Ці три твердження виконують різні функції.

| Твердження | Головне питання | Часова логіка |
|---|---|---|
| *Gap Claim* | Чого бракує сучасному знанню? | до основного дослідження |
| *Contribution Claim* | Що наше дослідження додає? | уточнюється в процесі отримання результатів |
| *Novelty Claim* | У чому отриманий результат новий відносно наявний доробок (prior art)? | остаточно перевіряється після отримання результатів |

Їхня узгодженість може бути представлена як:

$$
G \xrightarrow{\text{research}} C,
$$

де $G$ -- прогалина, а $C$ -- внесок.

Новизна визначається порівнянням внеску з наявним знанням $K$:

$$
N = C \setminus K.
$$

Це не математична операція в строгому сенсі, а *концептуальна модель*: новизна -- та частина обґрунтованого внеску, якої немає у підтвердженому prior knowledge.


# 8. Які докази потрібні для різних тверджень

Не всі твердження доводяться однаково.

| Claim | Типові необхідні докази |
|---|---|
| «Проблема актуальна» | масштаб/наслідки проблеми, авторитетні джерела, емпіричні дані |
| «Наявні методи мають обмеження» | порівняльні дослідження, systematic/scoping review, реплікації |
| «Існує research gap» | систематизована карта літератури та відсутність достатньої відповіді на конкретне питання |
| «Метод кращий» | коректний базовий рівень порівняння (baseline), однаковий протокол, метрики, статистична/практична оцінка |
| «Встановлено закономірність» | дизайн, що дозволяє перевірити залежність, повтори, аналіз невизначеності |
| «Результат новий» | зіставлення з релевантним наявний доробок (prior art) |
| «Результат узагальнюється» | зовнішня валідація, інші набори даних (datasets)/умови/домени |

::: callout-important
## Принцип

*Сила Claim не повинна перевищувати силу Evidence.*

Якщо дослідник провів один експеримент на одному наборі даних, він не може без додаткових доказів робити універсальне твердження про перевагу методу.
:::


# 9. Технічна новизна та нове наукове знання

Розглянемо типовий для Computer Science ланцюг.

### Варіант A -- переважно інженерний

> Взяли відому CNN → додали attention → отримали +1.2% accuracy.

Це може бути корисним результатом, але науковий внесок поки неочевидний.

### Варіант B -- потенційно науковий

> Висунуто припущення, що виграш attention залежить від просторової неоднорідності інформативних ознак. Сформовано контрольований експеримент, у якому цей фактор варіюється незалежно. Встановлено закономірність зміни ефекту та визначено область, де модифікація дає стабільний виграш.

Тут уже є:

- теоретично/методологічно мотивована ідея;
- фактор;
- перевірюване твердження;
- контрольований доказ;
- межі застосування.

Саме *перехід від «ми зробили» до «ми встановили»* часто відділяє інженерний результат від наукового.


## 9.1. Артефакт як носій внеску

У Computer Science програмний або обчислювальний артефакт може бути центральним результатом дослідження.

Але важливо розділити:

$$
\text{Artifact} \neq \text{Contribution}.
$$

Артефакт може *реалізовувати* внесок:

```{mermaid}
%%{init: {'theme': 'base', 'themeVariables': {
  'fontSize': '18px',
  'fontFamily': 'Trebuchet MS, Verdana, Arial, sans-serif',
  'primaryColor': '#E8F3FF',
  'primaryBorderColor': '#3478A8',
  'secondaryColor': '#FFF0C9',
  'tertiaryColor': '#E7F6EC',
  'lineColor': '#536878'
}}}%%
flowchart LR
    I("💡 Наукова<br/>ідея") --> A("🛠️ Artifact")
    A --> X("🧪 Experiment")
    X --> E("📊 Evidence")
    E --> K("🧠 New Knowledge")
```

# 10. Наскрізний приклад: виявлення ботоподібної активності

Повернімося до прикладу, який використовувався в попередніх лекціях.

### Крок 1. Вихідна проблема

Моделі можуть демонструвати високу якість на даних, подібних до навчальних, але втрачати її за зміни тематики дискусій.

### Крок 2. Сучасний стан

У літературі використовуються:

- текстові ознаки;
- поведінкові ознаки;
- мережеві ознаки;
- профільні метадані;
- гібридні представлення;
- різні ML/DL/LLM-підходи.

Однак протоколи оцінювання між роботами суттєво відрізняються.

### Крок 3. Робочий Gap Claim

> Недостатньо встановлено, які групи поведінкових і текстових ознак зберігають дискримінативність за тематичного зсуву та як цю стійкість доцільно враховувати при побудові моделі.

### Крок 4. Очікуваний Contribution Claim

> Встановити закономірності зміни дискримінативності груп ознак за тематичного зсуву та на їх основі запропонувати спосіб формування стійкого ознакового представлення.

### Крок 5. Робочий Novelty Claim

> Новизна потенційно полягає не в застосуванні конкретного класифікатора, а у *встановленні та формалізації властивості міжтематичної стійкості ознак* і використанні цієї властивості як принципу їх відбору.

Зверніть увагу: ми поки не пишемо *«вперше встановлено»*. Для цього потрібні результати дослідження та повторна перевірка наявний доробок (prior art).


# 11. Карта наукових внесків (Contribution Map)

Для дисертації корисно не шукати одну «головну фразу новизни», а побудувати *карту потенційних внесків*.

Приклад:

| ID | Research Gap | Очікуваний результат | Тип внеску | Необхідний доказ |
|---|---|---|---|---|
| C1 | невідомо, які ознаки стійкі до thematic shift | оцінені закономірності стійкості | емпіричний | міждоменний експеримент (cross-domain experiment) |
| C2 | немає способу врахувати стійкість при відборі | метод відбору ознак | методичний | порівняння з базові рівні порівняння (baselines) + ablation |
| C3 | межі переносимості не встановлені | умови та межі застосування | пояснювальний | аналіз чутливості та стійкості (sensitivity/robustness analysis) |

Це значно сильніша конструкція, ніж список:

> «розроблено метод; розроблено модель; створено програму».

Кожний внесок пов'язаний із *конкретною прогалиною та конкретним доказом*.


Таким чином, дисертація розглядається як *система взаємопов'язаних наукових тверджень і внесків*, а не як великий текст навколо однієї технології.


# 12. Типові помилки при формулюванні новизни

## Помилка 1. «Уперше застосовано...»

Сам факт першого застосування технології до конкретного об'єкта не гарантує наукової новизни.

## Помилка 2. «Удосконалено метод шляхом додавання...»

Додавання компонента ще не пояснює:

- чому саме він потрібний;
- яку властивість змінює;
- який ефект створює;
- чи є ефект стабільним;
- чи не пояснюється він збільшенням кількості параметрів.

## Помилка 3. Новизна через назву технології

> «Новизна полягає у використанні LLM».

LLM -- інструмент або клас моделей. Новизна має бути у *результаті, методі, закономірності, поясненні або доказі*.

## Помилка 4. Надмірний Claim

> «Запропонований метод забезпечує високу точність виявлення ботів у соціальних мережах».

Якщо експеримент виконано лише на одному набір даних (dataset), твердження значно ширше за доказову базу.

## Помилка 5. Gap, Contribution і Novelty не пов'язані

Прогалина стосується переносимості, метод оптимізує швидкість, а новизна заявляє точність. Це три різні історії.

## Помилка 6. Новизна формулюється до огляду

Тоді огляд літератури перетворюється на пошук підтвердження вже прийнятого рішення -- форму *confirmation bias*.


# 13. Тест на захищуваність наукового внеску

Для кожного потенційного внеску поставимо сім питань.

1. *Gap:* яку конкретну прогалину він закриває?
2. *наявний доробок (prior art):* з чим саме його потрібно порівнювати?
3. *Difference:* у чому змістовна відмінність?
4. *Mechanism:* чому очікується ефект?
5. *Evidence:* яким дизайном цей ефект буде перевірено?
6. *Boundary:* за яких умов Claim перестає бути справедливим?
7. *Value:* яке нове знання отримує наукова спільнота?

Якщо на більшість питань немає відповіді, внесок ще *не сформований методологічно*, навіть якщо реалізація вже працює.


# 14. Контрольна точка перегляду 1

Після тем 1--4 ми вже можемо виконати першу серйозну ревізію початкової концепції дисертації.

На старті було:

$$
\text{Problem v1} \rightarrow
\text{Object} \rightarrow
\text{Subject} \rightarrow
\text{Aim} \rightarrow
\text{Tasks}.
$$

Після формування доказової бази додаються:

$$
\text{State of the Art} \rightarrow
\text{Research Gap} \rightarrow
\text{Expected Contributions} \rightarrow
\text{Novelty Claims v1}.
$$

Тепер початковий концептуальний каркас *повинен бути переглянутий*, якщо докази не підтримують його первинну версію.


## 14.1. Повний каркас після змістового модуля 2

```{mermaid}
%%{init: {'theme': 'base', 'themeVariables': {
  'fontSize': '17px',
  'fontFamily': 'Trebuchet MS, Verdana, Arial, sans-serif',
  'primaryColor': '#E8F3FF',
  'primaryBorderColor': '#3478A8',
  'secondaryColor': '#FFF0C9',
  'tertiaryColor': '#E7F6EC',
  'lineColor': '#536878'
}}}%%
flowchart TB
    P("🎯 Research Problem") --> O("🌐 Object / Subject")
    O --> A("🏁 Aim / Tasks")
    P --> E("📚 Evidence Base")
    E --> S("🗺️ State of the Art")
    S --> G("🧩 Research Gap")
    G --> R("❓ Research Questions<br/>v1")
    G --> C("💡 Expected Contributions")
    C --> N("🏆 Novelty Claims<br/>v1")
    R -. "уточнюють" .-> A
    G -. "перевіряє" .-> P
```

Саме на цьому етапі може з'ясуватися, що:

- проблема сформульована занадто широко;
- предмет треба звузити;
- одна із задач не підтримується доказами;
- початково обрана технологія не є принциповою;
- очікуваний внесок уже частково реалізований іншими авторами;
- натомість у доказовій базі виявлено іншу, сильнішу прогалину.

Це *нормальний результат наукового пошуку*, а не помилка планування.

# 17. Підсумки

1. *Дослідницька прогалина* (research gap) -- це не просто «те, чого не знайшли», а обґрунтований доказами дефіцит знання.
2. *Gap Claim* фіксує, що відомо, яке обмеження існує і що залишається невстановленим.
3. *Науковий внесок* (research contribution) описує те, що дослідження додає до наявного знання.
4. *Наукова новизна* (scientific novelty) визначається лише відносно встановленого *prior knowledge*.
5. Новий алгоритм, модель, набір даних (dataset) або програмний артефакт можуть бути носіями внеску, але їхня технічна новизна не є автоматично науковою.
6. *Contribution Claim* має відповідати конкретному Research Gap.
7. *Novelty Claim* потребує зіставлення результату з релевантним наявний доробок (prior art) і не повинен бути ширшим за доказову базу.
8. Для PhD важливий перехід від *«ми зробили»* до *«ми встановили, обґрунтували й перевірили»*.
9. Після формування доказової бази початковий концептуальний каркас дисертації потрібно критично переглянути.
10. Результатом змістового модуля 2 стає ланцюг:

$$

\text{Research Gap}

$$
