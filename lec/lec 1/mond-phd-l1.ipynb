# Вступ

На магістерському рівні ми вже ставили питання:

> *Де саме в технічно складній роботі виникає наукове знання?*

На рівні PhD цього вже недостатньо. Тут виникає сильніше питання:

> **Яку систему взаємопов'язаних досліджень потрібно побудувати, щоб обґрунтовано претендувати на оригінальний науковий внесок?**

Дисертація не стає PhD-дослідженням лише тому, що:

- використано сучасну модель;
- створено складне програмне забезпечення;
- зібрано великий dataset;
- отримано високу метрику;
- опубліковано кілька статей.

Усе це може бути важливою частиною роботи. Але центральним залишається інше:

$$
\boxed{
\text{Що нового стало відомо}
+
\text{чому цьому можна довіряти}
+
\text{який внесок це робить у знання}
}
$$


# Тема 1. Наукове дослідження рівня PhD у комп'ютерних науках

## 1. Від виконання дослідження до створення нового знання

Наукове дослідження можна подати як систематичний процес зменшення невизначеності.

На рівні PhD до цього додається принципова вимога: дослідження має приводити до *оригінального внеску* (*original contribution*) у відповідну галузь знань.


Тому ключовою одиницею PhD-дослідження є не окремий алгоритм, модель або експеримент, а *обґрунтоване наукове твердження*, включене в цілісну систему знання.


## 2. Research, original research і doctoral research

Три поняття доцільно розрізняти.

### Research (дослідження)

Систематична діяльність, спрямована на отримання обґрунтованої відповіді на дослідницьке питання.

### Original research (оригінальне дослідження)

Дослідження, у якому отримано результат, що не є простим повторенням уже відомого і додає новий елемент до наявного знання.

### Doctoral research (дисертаційне дослідження рівня PhD)

Оригінальне дослідження, у якому аспірант:

1. самостійно формулює або істотно уточнює наукову проблему;
2. позиціонує її відносно *state of the art* (сучасного стану досліджень);
3. будує методологічно узгоджену програму дослідження;
4. отримує й перевіряє систему наукових тверджень;
5. демонструє оригінальний внесок;
6. критично визначає межі отриманого знання;
7. забезпечує достатню прозорість і простежуваність дослідницького процесу.


### 2.1. PhD -- це не "більша магістерська робота"

Якісна відмінність полягає не лише в обсязі.

| Вимір | Магістерський рівень | PhD-рівень |
|---|---|---|
| Вихідна задача | Часто задана або добре окреслена | Має бути проблематизована й обґрунтована |
| Робота з літературою | Показати володіння полем | Довести стан знання і наявність прогалини |
| Метод | Коректно обрати й застосувати | Обґрунтувати систему методів щодо claims |
| Експеримент | Отримати й проаналізувати результат | Побудувати доказову архітектуру |
| Новизна | Може бути локальною | Має підтримувати оригінальний внесок |
| Результат | Завершене дослідження / розробка | Система нових обґрунтованих тверджень |
| Критичність | Аналіз обмежень | Системний аудит альтернативних пояснень і меж |
| Відтворюваність | Бажана | Частина дослідницької якості |

::: callout-important
На PhD-рівні недостатньо показати, що запропоноване рішення *працює*. Потрібно встановити, *що саме з цього випливає як нове знання* і *якими доказами це підтримується*.
:::


## 3. Три близькі, але не тотожні поняття: originality, novelty, contribution

У дисертаційній роботі ці поняття часто змішують.

### Originality (оригінальність)

Характеристика дослідження або підходу, яка вказує, що робота не є простим відтворенням уже виконаного.

### Novelty (новизна)

Характеристика конкретного результату або твердження щодо того, *що саме є новим* порівняно з наявним знанням.

### Contribution (науковий внесок)

Те, *що дослідження додає до знання або дослідницької практики* і чому це додавання має значення для наукової спільноти.

У спрощеній формі:

$$
\text{Originality}
\neq
\text{Novelty}
\neq
\text{Contribution}.
$$

Але в сильній дисертації вони мають бути логічно пов'язані.


### Приклад

Аспірант запропонував нову нейромережеву архітектуру для класифікації повідомлень соціальних медіа.

Факт того, що архітектура раніше не публікувалася, може підтримувати *novelty claim* (твердження про новизну).

Але цього ще недостатньо для сильного *contribution claim* (твердження про внесок).

Потрібно встановити, наприклад:

- яку невирішену проблему ця архітектура адресує;
- чому наявні підходи недостатні;
- за яких умов нова архітектура дає перевагу;
- чи є ця перевага статистично та практично значущою;
- якою є ціна покращення;
- чи стійкий результат до зміни даних і протоколу;
- які межі узагальнення.

Тоді ланцюг стає таким:

$$
\text{New Artifact}
\rightarrow
\text{Validated Properties}
\rightarrow
\text{Scientific Claim}
\rightarrow
\text{Contribution}.
$$


# 4. Дослідження та інженерна розробка

У Computer Science межа між *research* (дослідженням) та *engineering* (інженерією) особливо важлива.

### Інженерна логіка

> Який артефакт потрібно створити, щоб задовольнити вимоги?

### Дослідницька логіка

> Яке невідоме твердження про метод, систему, алгоритм, дані або явище потрібно встановити і як отримати для нього достатні evidence?

```{mermaid}
%%{init: {'theme': 'base', 'themeVariables': {
  'fontSize': '18px',
  'fontFamily': 'Trebuchet MS, Verdana, Arial, sans-serif',
  'lineColor': '#6366f1'
}}}%%
flowchart LR
    N("🎯 Потреба")
    A("🧩 Artifact")
    F("⚙️ Функціонування")
    Q("❓ Research Question")
    EV("🔎 Evidence")
    CL("💬 Claim")

    N --> A --> F
    A --> Q --> EV --> CL

    classDef eng fill:#dbeafe,stroke:#2563eb,stroke-width:2px,color:#1e3a8a
    classDef res fill:#fef3c7,stroke:#d97706,stroke-width:2px,color:#78350f
    classDef out fill:#dcfce7,stroke:#16a34a,stroke-width:2px,color:#14532d

    class N,A,F eng
    class Q,EV res
    class CL out
```

## 4.1. Artifact ≠ Claim

Розрізнення *artifact* (артефакту) і *claim* (твердження) є одним із центральних для курсу.

| Artifact | Можливий scientific claim |
|---|---|
| Новий алгоритм | Алгоритм має властивість $P$ за умов $C$ |
| ML-модель | Модель стійкіша до dataset shift у визначеному класі сценаріїв |
| Dataset | Запропонований протокол формування даних зменшує певний тип bias |
| Архітектура системи | Архітектурне рішення забезпечує вимірюваний компроміс між latency і accuracy |
| Метод пояснення | Метод дає стабільніші пояснення за визначеним критерієм |
| Програмний прототип | Реалізація демонструє здійсненність, але сама по собі ще не доводить загального claim |

::: callout-tip
Корисне контрольне питання для аспіранта:

> *Якщо прибрати назву створеного програмного продукту, яке нове речення про світ, метод або систему я все ще можу науково захищати?*
:::


# 5. Які типи досліджень співіснують у Computer Science

Методологія CS є плюралістичною. Одна дисертація може поєднувати кілька типів досліджень.

У НРП цього курсу ми спеціально виділяємо:

- *theoretical research* (теоретичне дослідження);
- *empirical research* (емпіричне дослідження);
- *experimental research* (експериментальне дослідження);
- *computational research* (обчислювальне дослідження);
- *design science* (наука про проєктування).

Вони не є взаємовиключними.


## 5.1. Theoretical research (теоретичне дослідження)

Основне джерело evidence -- формальні міркування.

Типові результати:

- теорема;
- доведена властивість алгоритму;
- оцінка складності;
- формальна модель;
- умови збіжності;
- доведені межі методу.

Наприклад, якщо для нового алгоритму доведено:

$$
T(n)=O(n\log n),
$$

це твердження підтримується не benchmark-результатом, а формальним аналізом.

Однак емпіричне дослідження може додатково перевіряти практичну поведінку реалізації.


## 5.2. Empirical research (емпіричне дослідження)

*Empirical research* спирається на систематично отримані спостереження або дані.

Приклади:

- аналіз репозиторіїв програмного забезпечення;
- дослідження поведінки користувачів;
- аналіз соціальних медіа;
- benchmark-дослідження;
- case study;
- mining software repositories;
- аналіз журналів подій інформаційних систем.

Важливо:

$$
\text{Data}
\not\Rightarrow
\text{Knowledge automatically}.
$$

Між даними й висновком завжди є дизайн, припущення, метод аналізу та аргументація.


## 5.3. Experimental research (експериментальне дослідження)

В експерименті дослідник цілеспрямовано змінює умови або фактор і спостерігає наслідок.

Спрощена логіка:

$$
X \longrightarrow Y,
$$

де $X$ -- контрольована зміна, а $Y$ -- вимірюваний результат.

Наприклад:

> Чи спричиняє заміна способу sampling зміну recall для міноритарного класу за інших контрольованих умов?

Сильний експеримент вимагає:

- чіткого фактору;
- контрольних умов;
- протоколу;
- повторюваності;
- аналізу невизначеності;
- контролю альтернативних пояснень.


## 5.4. Computational research (обчислювальне дослідження)

У *computational research* обчислювальна модель або симуляція стає інструментом отримання evidence.

```{mermaid}
    class K result
```

Критичне застереження:

> *Model output* (вихід моделі) ще не є *scientific evidence* (науковим доказовим матеріалом), доки не обґрунтовано модель, умови експерименту та правила інтерпретації.


## 5.5. Design science (наука про проєктування)

У *design science* створення артефакту є частиною дослідницької стратегії.

Але дослідницька цінність виникає не з факту створення артефакту, а з циклу:

$$
\text{Problem}
\rightarrow
\text{Design}
\rightarrow
\text{Artifact}
\rightarrow
\text{Evaluation}
\rightarrow
\text{Knowledge}.
$$

Артефактом можуть бути:

- метод;
- алгоритм;
- модель;
- архітектура;
- програмна система;
- framework;
- мова або нотація;
- процедура.

На рівні PhD потрібно пояснити не лише *що створено*, а й *яке узагальнюване знання отримано через створення та оцінювання*.


# 6. Один дисертаційний задум -- кілька типів досліджень

Розглянемо умовну тему:

> *Методи раннього виявлення ботоподібної активності в дискусіях соціальних медіа.*

У межах однієї дисертації можуть співіснувати:

1. *Empirical study* -- встановлення поведінкових характеристик акаунтів.
2. *Design science study* -- побудова нового методу виявлення.
3. *Experimental study* -- контрольоване порівняння з baseline.
4. *Computational study* -- масштабне моделювання поведінки методу за різних умов.
5. *Theoretical component* -- формалізація ознак, критеріїв або властивостей методу.

Отже, тип дослідження визначається не назвою дисертації, а логікою конкретного *research question* (дослідницького питання) і способом отримання evidence.


# 7. Scientific claim як центральна одиниця доказової архітектури

*Scientific claim* -- це твердження, яке дослідник висуває як результат дослідження і яке має бути підтримане належними evidence.

Наприклад:

> Метод $M$ забезпечує вищу стійкість до class imbalance, ніж baseline $B$, для наборів даних класу $D$ за умов $C$.

У цьому твердженні вже видно:

- об'єкт порівняння;
- властивість;
- baseline;
- область застосовності;
- умови.

Це значно сильніше за фразу:

> "Запропоновано ефективний метод."


## 7.1. Claim повинен мати межі

Наукове твердження можна умовно подати як:

$$
C = (P, D, K, U),
$$

де:

- $P$ -- властивість або відношення, яке стверджується;
- $D$ -- domain (предметна область / клас об'єктів);
- $K$ -- conditions (умови);
- $U$ -- uncertainty (невизначеність твердження).

Наприклад, результат на одному dataset не дозволяє автоматично перейти від:

> "На dataset $D_1$ метод показав вищий $F_1$"

до:

> "Метод є кращим для задачі виявлення ботів взагалі".

Другий claim значно ширший і потребує значно сильнішої доказової бази.


## 7.2. Сила claim не може перевищувати силу evidence

Цей принцип будемо використовувати впродовж усього курсу:

$$
\boxed{
\text{Strength of Claim}
\leq
\text{Strength of Evidence}
}
$$

Типові порушення:

- один dataset $\rightarrow$ універсальний висновок;
- один запуск моделі $\rightarrow$ твердження про стабільність;
- вища середня метрика $\rightarrow$ твердження про статистично значущу перевагу;
- correlation $\rightarrow$ causal claim;
- успішний prototype $\rightarrow$ твердження про масштабованість;
- відсутність статистичної значущості $\rightarrow$ твердження про відсутність ефекту.

На PhD-рівні такі переходи мають бути явно контрольованими.


# 8. Від окремого дослідження до research programme

Дисертація зазвичай містить не один експеримент, а систему взаємопов'язаних досліджень.

Позначимо окремі дослідження як:

$$
S_1,S_2,\ldots,S_k.
$$

Кожне з них може мати власне питання:

$$
RQ_1,RQ_2,\ldots,RQ_k,
$$

і формувати evidence:

$$
E_1,E_2,\ldots,E_k.
$$

Але дисертаційний внесок не повинен бути простим списком:

$$
S_1+S_2+\cdots+S_k.
$$

Потрібна інтеграція результатів у цілісну аргументацію.


## 8.1. Програма дисертаційного дослідження

*Research programme* (програма дослідження) -- це логічно пов'язана система досліджень, у якій результати одного етапу обґрунтовують, уточнюють або перевіряють наступні.

```{mermaid}
%%{init: {'theme': 'base', 'themeVariables': {
  'fontSize': '17px',
  'fontFamily': 'Trebuchet MS, Verdana, Arial, sans-serif',
  'lineColor': '#6366f1',
  'edgeLabelBackground': '#ffffff'
}}}%%
flowchart TD
    P("❓ Research Problem")
    S1("🔎 Study 1<br/>опис / картування")
    S2("🧩 Study 2<br/>метод / модель")
    S3("🧪 Study 3<br/>перевірка")
    S4("🛡️ Study 4<br/>робастність / межі")
    E("📚 Cumulative Evidence")
    C("💬 System of Claims")
    CONTR("🌱 Dissertation Contribution")

    P --> S1 --> S2 --> S3 --> S4
    S1 --> E
    S2 --> E
    S3 --> E
    S4 --> E
    E --> C --> CONTR

    classDef problem fill:#fef3c7,stroke:#d97706,stroke-width:2px,color:#78350f
    classDef study fill:#e0e7ff,stroke:#6366f1,stroke-width:2px,color:#312e81
    classDef evidence fill:#dbeafe,stroke:#2563eb,stroke-width:2px,color:#1e3a8a
    classDef result fill:#dcfce7,stroke:#16a34a,stroke-width:2px,color:#14532d

    class P problem
    class S1,S2,S3,S4 study
    class E,C evidence
    class CONTR result
```

## 8.2. Кумулятивна доказова база

У сильній дисертації різні дослідження можуть підтримувати різні аспекти одного внеску.

Наприклад:

| Study | Що встановлюємо | Роль у загальному доказі |
|---|---|---|
| $S_1$ | Які ознаки стабільно пов'язані з явищем | Формує емпіричну основу |
| $S_2$ | Чи можна на їх основі побудувати новий метод | Формує артефакт і первинний claim |
| $S_3$ | Чи перевершує метод альтернативи | Порівняльна валідація |
| $S_4$ | Де метод перестає працювати | Межі застосовності |
| $S_5$ | Чи відтворюється результат на іншому середовищі | Зовнішня підтримка claim |

Тоді contribution підтримується не одним "кращим числом", а *cumulative evidence* (сукупною доказовою базою).


# 9. Дисертація не дорівнює сумі публікацій

Публікації є важливими носіями результатів, але логіка дисертації має бути ширшою.

Невдала структура:

$$
\text{Paper}_1+\text{Paper}_2+\text{Paper}_3
\Rightarrow
\text{Dissertation}.
$$

Методологічно сильніша структура:

$$
\text{Research Problem}
\rightarrow
\{RQ_i\}
\rightarrow
\{Study_i\}
\rightarrow
\{Evidence_i\}
\rightarrow
\{Claim_i\}
\rightarrow
\text{Integrated Contribution}.
$$

Одна стаття може підтримувати кілька claims. Один claim може потребувати evidence з кількох досліджень. Частина критичних перевірок може взагалі не утворювати окремої статті, але бути необхідною для достовірності дисертації.


# 10. Що означає "нове наукове знання" в Computer Science

Нове знання не обов'язково означає "новий алгоритм".

Науковий внесок може полягати у встановленні:

- нової закономірності;
- нової формальної властивості;
- нового причинного або пояснювального механізму;
- нових меж застосовності відомого методу;
- нової системи умов, за яких метод працює або не працює;
- нового способу формалізації проблеми;
- нового методу або моделі з доведеними властивостями;
- нової архітектури з обґрунтованою перевагою;
- нового dataset або benchmark, якщо його створення саме розв'язує наукову проблему;
- відтворюваного негативного результату, який спростовує поширене припущення.

::: callout-important
"Нове" без доказу -- лише заявка на новизну.  
"Працює" без меж -- лише локальне спостереження.  
Наукове знання потребує *обґрунтованого claim*.
:::

# 13. Стартова самоперевірка аспіранта

Перед ЛР1 корисно спробувати відповісти на сім питань.

1. Яку реальну наукову або науково-прикладну проблему я досліджую?
2. Що в цій проблемі *невідомо*, а не просто "потрібно реалізувати"?
3. Які claims я потенційно хочу мати право сформулювати наприкінці дисертації?
4. Які дослідження могли б дати evidence для цих claims?
5. Що в моєму задумі є artifact, а що -- scientific result?
6. Де я наразі лише *припускаю* новизну, але ще не довів її?
7. Які частини задуму можуть змінитися після систематичного аналізу літератури?

Якщо на частину питань поки немає відповіді -- це нормальна стартова ситуація першого семестру PhD. Завдання курсу саме в тому, щоб перетворити первинний задум на методологічно захищувану дослідницьку програму.


# 14. Перехід до ЛР1

Перша лабораторна робота починається не з написання коду.

Її мета -- провести *audit* (аудит) вихідної дисертаційної проблематики і зафіксувати початкову версію дослідницького каркаса.

На ЛР1 потрібно буде:

1. коротко описати вихідний дисертаційний задум;
2. відокремити практичну / інженерну потребу від проблеми знання;
3. сформулювати *Research Problem v1*;
4. визначити основні сутності, процеси та зв'язки предметної області;
5. побудувати первинний *Conceptual Framework v1*;
6. позначити, які елементи є припущеннями і потребують перевірки літературою;
7. сформувати перелік питань, які мають бути уточнені на наступних етапах.

Перший артефакт курсу:

$$
\boxed{
\text{Problem Audit}
+
\text{Research Problem}_{v1}
+
\text{Conceptual Framework}_{v1}
}
$$
