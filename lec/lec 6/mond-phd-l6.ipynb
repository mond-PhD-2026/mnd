# Вступ

У комп'ютерних науках результат легко сплутати з доказом.

Модель може показати *accuracy = 0.94*. Новий алгоритм може працювати швидше за baseline. Абляційний експеримент може продемонструвати зниження метрики після вилучення компонента. Проте жоден із цих фактів сам по собі ще не визначає, *яке саме наукове твердження ми маємо право зробити*.

На рівні PhD необхідно розрізняти:

- *результат* (result) -- те, що безпосередньо отримано;
- *доказ* (evidence) -- результат, релевантний конкретному Claim;
- *валідацію* (validation) -- систему перевірок, яка визначає надійність і межі цього доказу;
- *висновок* (conclusion) -- твердження, сила якого не повинна перевищувати силу доказової бази.

Тому принцип цієї лекції можна сформулювати так:

> *Сильний числовий результат не компенсує слабку доказову архітектуру.*


# Тема 6. Валідація наукових результатів і система доказів

Відповідно до НРП тема охоплює:

- твердження -- доказ -- валідацію;
- сукупну доказову базу;
- метрики;
- невизначеність;
- розмір ефекту;
- статистичну та практичну значущість;
- перевірки робастності;
- абляційні дослідження;
- аналіз чутливості;
- внутрішню, конструктну, зовнішню валідність і валідність висновків;
- негативні та нульові результати.

Усі ці елементи потрібні не як окремий набір статистичних або експериментальних прийомів, а як *система перевірки наукових тверджень дисертації*.


# 1. Від результату до доказу

Розглянемо твердження:

> «Запропонований метод є стійкішим до зменшення повноти спостережень, ніж базовий метод».

Припустимо, в одному експерименті отримано:

| Метод | Повнота спостережень | F1 |
|---|---:|---:|
| Baseline | 60% | 0.71 |
| Proposed | 60% | 0.78 |

Результат $0.78 > 0.71$ є *спостереженням*. Щоб він став переконливим Evidence для Claim про стійкість, необхідно запитати:

1. Чи повторюється ефект?
2. Яка невизначеність оцінки?
3. Який розмір ефекту?
4. Чи зберігається ефект на інших рівнях повноти?
5. Чи не пояснюється він випадковим seed, конкретною вибіркою або налаштуванням?
6. Чи коректна сама операціоналізація «стійкості»?
7. Наскільки далеко можна узагальнювати висновок?

Отже:

$$
\text{Result} \neq \text{Evidence} \neq \text{Validated Evidence}.
$$

# 4. Метрика як операціоналізація результату

Метрика -- це не нейтральне число. Вона визначає, *яку властивість результату ми фактично вимірюємо*.

Наприклад, для класифікації:

$$
Accuracy=\frac{TP+TN}{TP+TN+FP+FN},
$$

$$
Precision=\frac{TP}{TP+FP},
$$

$$
Recall=\frac{TP}{TP+FN},
$$

$$
F_1=2\cdot\frac{Precision\cdot Recall}{Precision+Recall}.
$$

За незбалансованих класів високе Accuracy може співіснувати з неприйнятним Recall для рідкісного класу.

Тому правильне питання:

> *Чи відповідає метрика тому конструкту, про який сформульовано Claim?*


## 4.1. Метрика не дорівнює науковій меті

Якщо Claim стосується:

- *точності* -- потрібна метрика predictive performance;
- *каліброваності* -- calibration error / Brier score;
- *стійкості* -- зміна якості при контрольованому perturbation;
- *обчислювальної ефективності* -- час, пам'ять, FLOPs, енергоспоживання;
- *справедливості* -- відповідні group fairness metrics;
- *інтерпретованості* -- потрібна окрема операціоналізація, а не автоматично accuracy.

Найкраща метрика -- не та, що дає найбільше число, а та, що найбільш валідно вимірює потрібну властивість.


# 5. Точкова оцінка та невизначеність

Точкова оцінка:

$$
\hat{\theta}=0.84
$$

не повідомляє, наскільки стабільно ми оцінили параметр $\theta$.

Тому наукова інтерпретація має включати *невизначеність* (uncertainty).

Типовий запис:

$$
\hat{\theta} \pm \text{uncertainty}.
$$

Наприклад:

> F1 = 0.84, 95% CI [0.81; 0.87].

Інтервал не перетворює слабкий дизайн на сильний, але дозволяє бачити, наскільки точним є оцінювання в межах прийнятого дизайну.


## 5.1. Звідки виникає невизначеність

У комп'ютерних науках джерелами варіативності можуть бути:

- вибірка об'єктів;
- train/test split;
- random initialization;
- stochastic optimization;
- sampling у генеративних моделях;
- вибір гіперпараметрів;
- шум вимірювання;
- зміна середовища;
- часовий або доменний drift.

Тому один запуск із `random_state=42` забезпечує *відтворюваність конкретного запуску*, але не доводить *стабільність наукового ефекту*.


# 6. Розмір ефекту

Статистичне питання «чи відрізняються результати?» потрібно відділяти від питання:

> *Наскільки велика ця відмінність?*

Нехай:

$$
\Delta = M_{\text{proposed}}-M_{\text{baseline}}.
$$

Якщо:

$$
M_{\text{proposed}}=0.903,\qquad
M_{\text{baseline}}=0.901,
$$

то:

$$
\Delta=0.002.
$$

Навіть якщо за великої кількості спостережень така різниця статистично виявляється, її практична або наукова значущість може бути малою.

Тому *effect size* має інтерпретуватися разом із невизначеністю та предметним контекстом.


# 7. Статистична значущість

У класичній схемі перевірки гіпотез:

$$
H_0:\Delta=0,
$$

$$
H_1:\Delta\neq0.
$$

$p$-value характеризує сумісність спостережуваних даних із моделлю за умови $H_0$. Він *не є*:

- імовірністю істинності $H_0$;
- імовірністю випадковості результату;
- розміром ефекту;
- доказом практичної важливості;
- гарантією відтворюваності.

На PhD-рівні запис «$p<0.05$, отже метод кращий» є методологічно недостатнім.


## 7.1. Множинні перевірки

Якщо дослідник перевіряє багато гіпотез, моделей, датасетів і метрик, імовірність випадково отримати «значущий» результат зростає.

Тому потрібно заздалегідь фіксувати:

- основні та додаткові метрики;
- primary comparison;
- кількість порівнянь;
- правила корекції, якщо вони потрібні;
- exploratory та confirmatory analyses.

Це безпосередньо пов'язує валідацію з експериментальним протоколом лекції 5.


# 8. Практична значущість

*Практична значущість* (practical significance) відповідає на питання:

> *Чи має величина ефекту реальне значення для задачі, системи або користувача?*

Наприклад, приріст F1 на 0.3 процентного пункту може бути:

- несуттєвим, якщо метод у 20 разів повільніший;
- важливим у критичній системі;
- недостатнім для складності впровадження;
- цінним, якщо він стабільний на складних підгрупах.

Отже, науковий висновок має враховувати *trade-offs*, а не одну максимізовану метрику.


# 9. Робастність

*Робастність* (robustness) -- здатність висновку або результату зберігатися за обґрунтованих змін умов.

Перевірки робастності можуть змінювати:

- random seed;
- train/test split;
- підмножину даних;
- рівень шуму;
- ступінь пропусків;
- preprocessing;
- гіперпараметри;
- часовий період;
- домен;
- реалізацію окремого компонента.

Ключове питання:

> *Чи є висновок властивістю досліджуваного явища/методу, а не випадковою властивістю одного налаштування?*


## 9.1. Робастність Claim, а не лише моделі

Робастність моделі і робастність наукового висновку -- різні речі.

Наприклад:

> «Модель зберігає accuracy при додаванні Gaussian noise».

це перевірка моделі.

А твердження:

> «Перевага запропонованого методу над baseline зберігається за різних seed, split, рівнів шуму та двох незалежних наборів даних»

-- уже перевіряє робастність *порівняльного Claim*.


# 10. Абляційні дослідження

*Абляційне дослідження* (ablation study) перевіряє, яку роль відіграють компоненти запропонованого методу.

Нехай система:

$$
S=A+B+C.
$$

Тоді перевіряємо:

$$
S_{-A}=B+C,\qquad
S_{-B}=A+C,\qquad
S_{-C}=A+B.
$$

Якщо повна система має F1 = 0.86, а без компонента $B$ -- 0.79, це створює Evidence щодо внеску $B$.

Але обережно: абляція показує *залежність результату від компонента в заданій системі*, а не автоматично причинний механізм у широкому сенсі.


## 10.1. Факторіальна абляція

Компоненти можуть взаємодіяти. Тому схема «прибрати по одному» не завжди достатня.

Для двох компонентів $A$ і $B$ корисно перевірити:

| A | B | Результат |
|---|---|---:|
| 0 | 0 | $M_{00}$ |
| 1 | 0 | $M_{10}$ |
| 0 | 1 | $M_{01}$ |
| 1 | 1 | $M_{11}$ |

Це дозволяє побачити, чи є ефект комбінації простим додаванням внесків або виникає взаємодія.


# 11. Аналіз чутливості

*Аналіз чутливості* (sensitivity analysis) досліджує, як висновок змінюється при зміні припущень або параметрів.

У найпростішій формі:

$$
S_x=\frac{\Delta y}{\Delta x},
$$

де $x$ -- параметр або припущення, а $y$ -- характеристика результату.

Нас цікавить не лише значення $S_x$, а *область стабільності висновку*.

Наприклад:

> перевага методу зберігається при threshold від 0.35 до 0.70, але зникає після 0.75.

Такий результат значно інформативніший, ніж звіт лише для threshold = 0.5.


# 12. Валідація і валідність

Термін *validation* описує процедури перевірки.

*Validity* -- властивість обґрунтованості інтерпретації та висновку.

Тому питання «ми провели validation?» саме по собі недостатнє. Потрібно запитати:

> *Яку загрозу валідності ця процедура зменшує?*


## 12.1. Чотири виміри валідності

| Вид валідності | Основне питання |
|---|---|
| *Внутрішня* (internal validity) | Чи справді спостережуваний ефект пояснюється досліджуваним фактором, а не конфаундером? |
| *Конструктна* (construct validity) | Чи вимірюємо ми саме те поняття, про яке робимо Claim? |
| *Зовнішня* (external validity) | На які популяції, домени, умови та контексти можна переносити висновок? |
| *Валідність висновків* (conclusion validity) | Чи достатньо коректні дані й аналіз для самого висновку про зв'язок або ефект? |

Ці види валідності не замінюють один одного.


Для кожного ключового Claim у дисертації бажано мати явний список загроз і процедур, що їх зменшують.


# 13. Внутрішня валідність

Типові загрози:

- data leakage;
- неконтрольована різниця між групами;
- confounding;
- зміна preprocessing між умовами;
- різний tuning budget;
- використання test set під час розроблення;
- часовий leakage;
- невраховані залежності між спостереженнями.

Наприклад, якщо Proposed Model отримав більший tuning budget, то різницю в якості не можна безпосередньо приписувати архітектурі.


# 14. Конструктна валідність

Конструктна валідність особливо важлива для абстрактних понять:

- «ботоподібність»;
- «якість пояснення»;
- «стійкість»;
- «складність»;
- «довіра користувача»;
- «семантична подібність»;
- «ситуаційна обізнаність».

Якщо «ботоподібність» операціоналізована лише частотою повідомлень, Claim фактично стосується *інтенсивності активності*, а не повного конструкта bot-like behavior.

Тому:

$$
\text{Construct}
\rightarrow
\text{Operationalization}
\rightarrow
\text{Measurement}
\rightarrow
\text{Claim}.
$$


# 15. Зовнішня валідність

Питання зовнішньої валідності:

- чи працює результат на іншому dataset?
- в іншому домені?
- в іншій мові?
- на іншому часовому періоді?
- на іншій платформі?
- за іншого рівня шуму?
- для іншої популяції користувачів?
- в реальному deployment?

Якщо дослідження виконано лише на одному англомовному наборі даних, твердження «метод ефективний для аналізу соціальних медіа» може бути надто широким.

Коректніше:

> «Метод продемонстрував перевагу в межах перевіреного набору даних і заданого експериментального протоколу».


# 16. Валідність висновків

Навіть за доброго дизайну висновок може бути слабким через:

- малу вибірку;
- високу дисперсію;
- невідповідний статистичний тест;
- порушення припущень тесту;
- cherry-picking метрик;
- множинні непоправлені порівняння;
- нестабільність між повтореннями;
- неправильне трактування $p$-value;
- відсутність effect size та uncertainty.

Тому статистичний аналіз -- частина доказової архітектури, а не формальний додаток наприкінці роботи.


# 17. Негативні та нульові результати

*Нульовий результат* (null result) означає, що запланований аналіз не дав достатніх підстав підтримати очікуваний ефект.

*Негативний результат* може показувати, що:

- гіпотеза не підтримується;
- запропонований метод поступається baseline;
- ефект не переноситься на інший контекст;
- компонент не дає очікуваного внеску;
- результат нестійкий.

Це не автоматично «невдале дослідження».

Якщо дизайн був коректним, нульовий або негативний результат може *звузити простір допустимих пояснень* і стати важливим науковим знанням.


## 17.1. Чого не можна робити з нульовим результатом

Неприпустимо:

- приховувати невигідні запуски;
- замінювати primary metric після перегляду результатів;
- нескінченно перебирати підгрупи;
- змінювати гіпотезу так, ніби вона була сформульована до аналізу;
- подавати «не виявили різниці» як доказ повної еквівалентності без відповідного дизайну.

Наукова цінність дослідження визначається не тим, чи «спрацювала ідея», а тим, чи коректно отримано та інтерпретовано знання.


# 18. Приклад: повна валідація одного Claim

Нехай Claim:

> «Запропонована архітектура забезпечує вищу стійкість класифікації за неповних спостережень порівняно з LSTM».

## Крок 1. Операціоналізація

Стійкість визначаємо як збереження F1 при зменшенні частки доступних спостережень:

$$
r \in \{1.0,0.8,0.6,0.4\}.
$$

## Крок 2. Evidence

Для кожного $r$ порівнюємо Proposed і LSTM.

## Крок 3. Невизначеність

Повторюємо навчання для кількох seed і оцінюємо інтервали.

## Крок 4. Effect size

Обчислюємо:

$$
\Delta(r)=F1_{\text{Proposed}}(r)-F1_{\text{LSTM}}(r).
$$

## Крок 5. Robustness

Перевіряємо кілька split / сценаріїв пропусків.

## Крок 6. External validity

За можливості повторюємо на незалежному наборі даних або сценарії.

## Крок 7. Межа Claim

Якщо перевага стабільна лише при $r\leq0.6$, Claim формулюємо саме для цього діапазону.


## 18.1. Схема прикладу

```{mermaid}
%%{init: {'theme': 'base', 'themeVariables': {
  'fontSize': '17px',
  'fontFamily': 'Trebuchet MS, Verdana, Arial, sans-serif',
  'primaryColor': '#E8F3FF',
  'primaryBorderColor': '#3478A8',
  'secondaryColor': '#FFF0C9',
  'tertiaryColor': '#EAF7EA',
  'lineColor': '#315B7D'
}}}%%
flowchart LR
    C([Claim:<br/>вища стійкість])
    O([Операціоналізація:<br/>F1 vs completeness])
    E([Evidence:<br/>Proposed vs LSTM])
    U([Uncertainty<br/>+ effect size])
    R([Robustness<br/>checks])
    X([External<br/>validation])
    B([Bounded<br/>conclusion])

    C --> O --> E --> U --> R --> X --> B
```

# 19. Evidence hierarchy для власної дисертації

Для кожного ключового Claim корисно визначити рівні підтримки.

### Рівень A -- первинний Evidence

Основний експеримент або теоретичний результат.

### Рівень B -- перевірка стабільності

Повторення, uncertainty, robustness, sensitivity.

### Рівень C -- декомпозиція

Ablation, component analysis, alternative explanations.

### Рівень D -- перенесення

External dataset, інший домен, час або незалежний сценарій.

Не кожний Claim потребує всіх рівнів. Але відсутність певного рівня має визначати *межі формулювання висновку*.


# 20. Сила Claim і сила Evidence

Корисне правило всього курсу:

$$
\boxed{\text{Strength(Claim)} \leq \text{Strength(Evidence)}}
$$

Приклади:

| Evidence | Надто сильний Claim | Коректніший Claim |
|---|---|---|
| один dataset | метод є універсально кращим | метод кращий у перевіреному контексті |
| один seed | метод стабільно кращий | в одному запуску отримано вищий результат |
| кореляція | X спричиняє Y | X асоційований з Y |
| ablation | компонент є причиною явища | компонент робить внесок у результат системи |
| $p>0.05$ | методи однакові | не отримано достатніх підстав виявити відмінність |

Методологічна зрілість проявляється не у максимальному Claim, а у *точному Claim*.


# 21. Коли доказів достатньо?

Універсальної кількісної межі немає.

Достатність Evidence залежить від:

- типу Claim;
- сили Claim;
- альтернативних пояснень;
- ризику помилки;
- зрілості предметної області;
- доступності незалежних даних;
- варіативності результату;
- стандартів відповідної дослідницької спільноти.

Тому питання «скільки експериментів потрібно?» краще замінити на:

> *Які правдоподібні альтернативні пояснення залишаються після запланованої системи перевірок?*


# 22. Evidence debt

За аналогією з technical debt корисно мислити про *evidence debt* -- накопичені твердження, для яких доказова підтримка ще недостатня.

Приклади:

- Claim уже сформульовано, але external validation не проведено;
- у статті заявлена robustness, але перевірено лише один тип шуму;
- novelty claim ширший за evidence base;
- абляція відсутня, хоча внесок компонента є частиною наукової новизни;
- uncertainty не оцінено.

Evidence debt потрібно бачити *до написання фінального тексту дисертації*.


# 23. Аудит доказової архітектури

Для кожного основного Claim поставте сім питань:

1. Який результат безпосередньо підтримує Claim?
2. Чи валідно операціоналізовано ключовий конструкт?
3. Яка невизначеність?
4. Який розмір ефекту?
5. Які альтернативні пояснення?
6. Які robustness / sensitivity / ablation checks потрібні?
7. Яка точна межа узагальнення?

Якщо на одне з питань немає відповіді, це не обов'язково руйнує дослідження -- але створює *явну методологічну задачу*.


# 24. Типові помилки валідації

1. Один числовий результат подається як доказ загальної переваги.
2. Звіт містить лише середнє без оцінки варіативності.
3. $p$-value використовується як міра сили ефекту.
4. Статистична значущість підміняє практичну.
5. Метрика не відповідає конструкту Claim.
6. Test set багаторазово використовується для прийняття рішень.
7. Ablation відсутня, хоча заявлено внесок окремого компонента.
8. Robustness перевіряється лише на одному зручному perturbation.
9. Висновок узагальнюється далеко за межі вибірки.
10. Нульові результати приховуються або переінтерпретуються post hoc.
11. Кількість експериментів зростає, але всі вони перевіряють одну й ту саму властивість.
12. Validation планується після отримання бажаного результату.


# 25. Тест узгодженості доказової системи

Для кожного Claim перевірте ланцюг:

$$
\text{Claim}
\rightarrow
\text{Construct}
\rightarrow
\text{Metric}
\rightarrow
\text{Study}
\rightarrow
\text{Result}
\rightarrow
\text{Uncertainty}
\rightarrow
\text{Validation}
\rightarrow
\text{Conclusion}.
$$

Ланцюг слабшає, якщо хоча б одна стрілка не має методологічного обґрунтування.

Особливо небезпечні розриви:

$$
\text{Metric} \not\Rightarrow \text{Construct},
$$

$$
\text{Result} \not\Rightarrow \text{Claim},
$$

$$
\text{Local Evidence} \not\Rightarrow \text{Universal Conclusion}.
$$


# 26. Що аспірант повинен мати після цієї лекції

До лабораторної роботи 6 необхідно підготувати:

1. актуальний набір основних *Contribution Claims / Novelty Claims*;
2. для кожного Claim -- основний Evidence;
3. метрику або інший критерій перевірки;
4. спосіб оцінювання uncertainty;
5. effect size або інший показник величини ефекту, якщо застосовно;
6. перелік robustness checks;
7. план ablation study, якщо Claim стосується компонентів методу;
8. sensitivity analysis, якщо висновок залежить від параметрів або припущень;
9. основні загрози internal / construct / external / conclusion validity;
10. межі допустимого висновку;
11. перелік можливих null / negative outcomes та правила їх інтерпретації.

У ЛР6 ці елементи будуть інтегровані у робочу *Claim--Evidence--Validation Matrix*.

# 28. Підсумки

1. Науковий результат стає Evidence лише відносно конкретного Claim.
2. Validation визначає, наскільки надійно Evidence підтримує Claim і де проходять межі висновку.
3. Сильна дисертація спирається на *сукупну доказову базу*, а не на один «найкращий» експеримент.
4. Метрика є частиною операціоналізації наукового конструкта.
5. Точкові оцінки потрібно доповнювати оцінюванням uncertainty.
6. Effect size, statistical significance і practical significance відповідають на різні питання.
7. Robustness, ablation і sensitivity analysis перевіряють різні аспекти доказової стійкості.
8. Internal, construct, external і conclusion validity задають різні класи загроз.
9. Null і negative results не слід приховувати: за коректного дизайну вони можуть давати нове знання.
10. Сила Claim не повинна перевищувати силу Evidence.

Після теми 6 магістральна логіка курсу набуває вигляду:

$$
\text{Research Gap}
\rightarrow
\text{RQ}
\rightarrow
\text{Claim}
\rightarrow
\text{Research Design}
\rightarrow
\text{Evidence}
\rightarrow
\text{Validation}
\rightarrow
\text{Bounded Conclusion}.
$$

Наступна тема поставить питання про те, *чи можна незалежно простежити, повторити та перевірити шлях від даних і коду до цього висновку*.
