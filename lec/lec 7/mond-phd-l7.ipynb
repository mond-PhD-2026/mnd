## Місце лекції в логіці курсу

У лекціях 5--6 ми перейшли від *дизайну дослідження* (research design) до системи доказів: сформували зв'язок

*дослідницьке питання → наукове твердження → дизайн → метод → дані → аналіз → доказ → валідація*.

Але навіть сильний результат не стає надійним науковим знанням автоматично. Потрібно, щоб було зрозуміло:

* звідки взялися дані;
* яким кодом і в якому середовищі отримано результат;
* які рішення приймалися в процесі аналізу;
* чи можна повторити обчислення;
* чи може інший дослідник незалежно перевірити висновок;
* чи не було результат «покращено» вибірковим поданням, зміною гіпотез або аналізу після перегляду даних;
* яку роль відігравали інструменти штучного інтелекту.

Тому тема 7 переводить нас від питання *«Чи достатньо сильні докази?»* до питання *«Чи можна простежити, перевірити й відтворити шлях від вихідних даних до наукового твердження?»*.

Саме ця логіка буде практично опрацьована в *лабораторній роботі 7*.
:::


# Вступ

Уявімо типову ситуацію в Computer Science.

Дослідник повідомляє:

> «Запропонований метод забезпечив Accuracy = 0.941 і статистично значущо перевершив baseline».

Саме по собі це формулювання ще не дозволяє перевірити результат. Інший дослідник може поставити десятки цілком методологічних запитань:

- Яка саме версія набору даних використовувалася?
- Як сформовано train/validation/test?
- Який preprocessing було застосовано?
- Якою була версія бібліотеки?
- Чи зафіксовано `random seed`?
- Які гіперпараметри використовувалися?
- Чи всі запуски включено до звіту?
- Чи метрика була визначена *до* перегляду результатів?
- Чи змінювалася гіпотеза після аналізу?
- Чи можна однозначно пов'язати число `0.941` із конкретним кодом, даними та конфігурацією?
- Якщо частину коду, тексту або аналізу створено за допомогою LLM, чи задокументовано це?

Тому сучасне наукове дослідження в комп'ютерних науках має бути не лише *результативним*, а й *аудитованим*.

Ключова ідея лекції:

> *Науковий результат є сильнішим, коли можна простежити його походження, повторити процедуру його отримання, незалежно перевірити ключові твердження та пояснити всі суттєві дослідницькі рішення.*


# Тема 7. Відтворюваність, відтворення незалежною реалізацією, простежуваність і дослідницька доброчесність

## 1. Чотири поняття, які потрібно розрізняти

У літературі терміни *reproducibility* і *replicability* використовуються не цілком однаково в різних дисциплінах. Для цього курсу зафіксуємо операційні визначення, зручні для дисертаційних досліджень у Computer Science.

### 1.1. Відтворюваність (reproducibility)

*Відтворюваність* означає можливість повторно отримати заявлений обчислювальний результат, використовуючи надані дослідником дані, код, параметри та опис середовища.

У спрощеному вигляді:

$$
R = f(D, C, P, E),
$$

де:

- $D$ -- дані (data);
- $C$ -- код (code);
- $P$ -- параметри та конфігурація (parameters/configuration);
- $E$ -- обчислювальне середовище (environment);
- $R$ -- результат (result).

Якщо хоча б один із критичних компонентів невідомий, функція фактично стає неоднозначною.

### 1.2. Відтворення незалежною реалізацією (replication)

*Відтворення незалежною реалізацією* (replication) є сильнішою перевіркою: інша команда або інший дослідник перевіряє наукове твердження, не покладаючись повністю на оригінальний обчислювальний pipeline.

Можуть змінюватися:

- реалізація алгоритму;
- програмна платформа;
- вибірка або набір даних;
- дослідницька група;
- частина експериментальних умов.

Тут перевіряється вже не лише те, чи «запускається код», а чи *зберігається науковий висновок*.

### 1.3. Простежуваність (traceability)

*Простежуваність* -- здатність відновити ланцюг походження результату:

$$
\text{Claim}
\leftarrow
\text{Evidence}
\leftarrow
\text{Analysis}
\leftarrow
\text{Data}
\leftarrow
\text{Source}.
$$

Для дисертації простежуваність повинна працювати і у зворотному напрямі:

$$
\text{Source}
\rightarrow
\text{Data}
\rightarrow
\text{Processing}
\rightarrow
\text{Result}
\rightarrow
\text{Claim}
\rightarrow
\text{Contribution}.
$$

### 1.4. Дослідницька доброчесність (research integrity)

*Дослідницька доброчесність* охоплює чесність, прозорість, відповідальність і належне документування наукової роботи.

Вона не зводиться до перевірки тексту на плагіат.

Для обчислювального дослідження доброчесність стосується також:

- походження даних;
- вибору та зміни аналітичних процедур;
- повноти подання результатів;
- коректності статистичних висновків;
- фіксації негативних і нульових результатів;
- авторства коду та артефактів;
- використання AI-assisted інструментів.

Це не універсальний стандарт і не вимога до назв каталогів. Важлива *функціональна повнота*:

1. зрозуміло, як отримати дані;
2. зрозуміло, як підготувати середовище;
3. зрозуміло, як запустити експеримент;
4. зрозуміло, де параметри;
5. зрозуміло, де первинні та похідні результати;
6. зрозуміло, який результат підтримує конкретне наукове твердження.


### 3.1. Дані: не лише файл `dataset.csv`

Для даних потрібно фіксувати *походження даних* (data provenance).

Корисний мінімум:

| Поле | Що фіксуємо |
|---|---|
| Source | джерело / API / репозиторій / експеримент |
| Retrieval date | дата отримання |
| Version | версія або snapshot |
| License / access | умови доступу |
| Raw checksum | контрольна сума сирого файлу |
| Processing | послідовність перетворень |
| Exclusions | що і чому вилучено |
| Derived version | версія підготовленого набору |
| Responsible script | скрипт, який породив похідні дані |

Важливий принцип:

> *Raw data не повинні непомітно перетворюватися на «ручний Excel, який десь виправили».*

Якщо ручна корекція неминуча, вона теж має стати документованою операцією.


### 3.2. Код і версії

Система контролю версій Git дозволяє пов'язати стан дослідження з конкретним commit.

Наприклад, результат можна описати не так:

> «Експеримент виконано кодом із GitHub».

а так:

> «Результати таблиці 4 отримано з commit `a13f2c7`, конфігурацією `exp_04.yaml`, набором даних `dataset-v2.1`, seed = 42».

Тоді формується адресований стан експерименту:

$$
E_i =
(\text{commit}, \text{data version}, \text{config}, \text{environment}, \text{seed}).
$$

Це суттєво сильніше, ніж посилання на репозиторій, який продовжує змінюватися.


### 3.3. Обчислювальне середовище

Фраза

> «Python 3 та scikit-learn»

майже нічого не гарантує.

Потрібно фіксувати щонайменше:

- версію Python;
- критичні залежності;
- версії бібліотек;
- ОС / контейнер -- якщо це впливає на результат;
- GPU/CUDA/cuDNN -- для відповідних задач;
- зовнішні сервіси та API;
- моделі, checkpoints, tokenizer;
- параметри генеративних моделей, якщо вони є частиною pipeline.

Для складних експериментів середовище можна описувати через `conda`, `pip`, `uv`, `poetry`, `Docker` або інший відтворюваний механізм.

Ключове не назва інструмента, а можливість відновити стан.


## 4. Детермінованість і відтворюваність -- не синоніми

Часто очікують, що відтворюваний експеримент має дати *біт-в-біт* ідентичний результат. Це можливо не завжди.

У машинному навчанні результат може залежати від:

$$
Y = f(D, \theta, s, H, L, E),
$$

де:

- $D$ -- дані;
- $\theta$ -- параметри моделі;
- $s$ -- джерела випадковості;
- $H$ -- hardware;
- $L$ -- реалізація бібліотек;
- $E$ -- середовище виконання.

Тому потрібно розрізняти:

- *точну обчислювальну повторюваність* -- однаковий запуск дає практично ідентичний результат;
- *статистичну відтворюваність* -- повторні запуски дають результати з узгодженого розподілу;
- *відтворюваність висновку* -- ключове наукове твердження зберігається попри допустиму варіативність.

Для стохастичних алгоритмів один `random_state=42` не вирішує проблему повністю. Часто коректніше провести $k$ повторів:

$$
m_1, m_2, \ldots, m_k,
$$

і подати, наприклад,

$$
\bar{m} \pm s_m,
$$

де $\bar{m}$ -- середнє значення метрики, а $s_m$ -- характеристика варіативності.


::: callout-important
## Методологічний акцент

*Seed -- це елемент протоколу, а не спосіб приховати нестабільність.*

Якщо висновок існує лише для одного «вдалого» seed, проблема не у відтворюваності оформлення, а в робастності самого наукового результату.
:::


## 5. Простежуваність як граф походження доказів

Для дисертації зручно трактувати простежуваність як орієнтований граф:

$$
G_T = (V, E),
$$

де вершини $V$ -- дослідницькі сутності, а ребра $E$ -- відношення походження або обґрунтування.

Типові вершини:

- джерело даних;
- версія набору даних;
- preprocessing;
- конфігурація;
- експеримент;
- результат;
- рисунок / таблиця;
- наукове твердження;
- внесок;
- публікація.


Якщо для Claim неможливо пройти цей граф назад до джерела доказу, виникає *розрив простежуваності* (traceability gap).


### 5.1. Матриця простежуваності

Для практичної роботи граф зручно доповнити таблицею.

| Claim ID | Result ID | Experiment ID | Data version | Code commit | Config | Validation |
|---|---|---|---|---|---|---|
| CL1 | R-04 | EXP-04 | D2.1 | `a13f2c7` | `exp04.yaml` | bootstrap + external test |
| CL2 | R-07 | EXP-07 | D2.1 | `b03d81e` | `exp07.yaml` | ablation |

Така матриця виконує одразу три функції:

1. *доказову* -- показує, чим підтримується Claim;
2. *технічну* -- показує, як відновити експеримент;
3. *аудиторську* -- дозволяє знайти слабку або відсутню ланку.


## 6. Журнал дослідження (research log)

Git добре фіксує зміни файлів, але не завжди пояснює *чому* дослідник прийняв рішення.

Тому корисним є *журнал дослідження* (research log).

Мінімальний запис може мати структуру:

| Date | Decision / Event | Rationale | Affected artifact | Consequence |
|---|---|---|---|---|
| 2026-10-05 | замінено macro-F1 на PR-AUC як primary metric | сильний class imbalance | Validation Protocol v2 | повторити EXP-03--EXP-06 |
| 2026-10-08 | вилучено feature X | виявлено leakage | Dataset v2.2 | попередні результати позначено obsolete |

Сильний журнал фіксує не лише успіхи, а й:

- відкинуті гіпотези;
- невдалі експерименти;
- зміни preprocessing;
- зміни метрик;
- зміни критеріїв виключення;
- причини повторного запуску;
- помилки, що вплинули на попередні результати.

Саме такі записи допомагають відрізнити нормальну еволюцію дослідження від прихованої post hoc оптимізації.


## 7. Відкрита наука (open science) -- не тотожна «викласти все у відкритий доступ»

*Відкрита наука* (open science) підтримує прозорість, перевірюваність і повторне використання результатів. Але принцип відкритості не скасовує:

- захист персональних даних;
- ліцензійні обмеження;
- комерційну таємницю;
- кібербезпекові ризики;
- етичні обмеження;
- права третіх сторін.

Тому корисно застосовувати принцип:

> *as open as possible, as closed as necessary*.

Якщо дані не можна оприлюднити, це не означає, що відтворюваність неможлива. Можна надати:

- схему даних;
- data dictionary;
- процедуру отримання;
- synthetic/minimal example;
- preprocessing code;
- checksums;
- контрольовану процедуру доступу;
- агреговані результати;
- достатній опис обмежень.


## 8. FAIR-принципи та відтворюваність

FAIR означає:

- *Findable* -- дані й артефакти можна знайти;
- *Accessible* -- зрозуміло, як отримати доступ;
- *Interoperable* -- використовуються формати та метадані, придатні для взаємодії;
- *Reusable* -- визначені умови та контекст повторного використання.

FAIR і Open -- не синоніми.

Дані можуть бути FAIR, але доступними лише за визначеною процедурою. Для дисертаційного дослідження це важливо, коли повна публікація dataset неможлива.


# 9. Дослідницька доброчесність у computational research

## 9.1. Фабрикація (fabrication)

*Фабрикація* -- створення даних або результатів, яких фактично не було.

Приклади:

- додавання вигаданих спостережень;
- створення «експериментальних» чисел без запуску експерименту;
- вигадані відповіді респондентів;
- посилання на неіснуючі джерела.

У контексті генеративного ШІ особливо небезпечно переносити до наукового тексту неперевірені згенеровані факти або бібліографічні посилання.

## 9.2. Фальсифікація (falsification)

*Фальсифікація* -- маніпулювання матеріалами, даними, процедурами або результатами так, що дослідження подається неточно.

Приклади:

- вилучення «незручних» запусків без заздалегідь визначеного правила;
- зміна графіка так, щоб перебільшити ефект;
- приховування невдалих експериментів, критичних для висновку;
- зміна test set після перегляду результатів;
- неповідомлене ручне редагування даних.


## 9.3. Вибіркове подання результатів (selective reporting)

Припустимо, дослідник перевірив 20 варіантів моделі, але у статті показав лише найкращий.

Якщо читач отримує враження, що цей варіант був єдиним або заздалегідь визначеним, доказова сила перебільшується.

Проблема виникає не через сам пошук кращої моделі. Hyperparameter tuning є нормальною процедурою. Проблема виникає, коли:

- пошук не відокремлений від фінального тестування;
- кількість спроб прихована;
- test set фактично використовується як validation set;
- повідомляється лише найвигідніший результат.


## 9.4. Маніпуляції статистичним аналізом (p-hacking)

*P-hacking* -- множина практик, за яких дослідник варіює аналіз до отримання статистично привабливого результату, а потім подає його як заздалегідь визначений.

Уявімо, що перевіряються $m$ незалежних нульових гіпотез при $\alpha = 0.05$.

Ймовірність отримати хоча б один false positive:

$$
P(\ge 1\ \text{false positive}) = 1 - (1-\alpha)^m.
$$

Для $m=20$:

$$
1 - 0.95^{20} \approx 0.642.
$$

Тобто навіть без реального ефекту ймовірність хоча б одного «значущого» результату вже близько 64%.

Саме тому важливі:

- заздалегідь визначений план аналізу;
- корекція множинних порівнянь;
- подання effect size та uncertainty;
- повний опис проведених аналізів;
- розмежування confirmatory та exploratory analysis.


## 9.5. HARKing

*Формулювання гіпотези після отримання результатів* (Hypothesizing After the Results are Known, HARKing) виникає тоді, коли post hoc пояснення подається так, ніби воно було сформульоване до експерименту.

Розглянемо різницю.

**Коректно:**

> Первинна гіпотеза H1 не підтвердилася. Під час exploratory analysis виявлено можливу взаємодію X×Z. На цій основі сформульовано нову гіпотезу H2, яку потрібно перевірити на незалежних даних.

**Некоректно:**

> Ми висунули H2 і підтвердили її на цих самих даних.

Проблема не в тому, що нова гіпотеза виникла після аналізу. Саме так часто розвивається наука. Проблема -- *приховати її post hoc походження*.


```{mermaid}
%%| fig-cap: "Коректний цикл: exploratory результат породжує нову confirmatory перевірку"
%%{init: {'theme': 'base', 'themeVariables': {
  'fontSize': '17px',
  'fontFamily': 'Trebuchet MS, Verdana, Arial, sans-serif',
  'primaryColor': '#E8F3FF',
  'primaryBorderColor': '#3478A8',
  'secondaryColor': '#FFF0C9',
  'tertiaryColor': '#E7F6EC',
  'lineColor': '#536878'
}}}%%
flowchart LR
    H1("❓ H1<br/>до аналізу") --> D1("📦 Data A")
    D1 --> T1("🧪 Confirmatory Test")
    T1 --> N("🔎 Неочікуваний патерн")
    N --> H2("💡 H2<br/>нова гіпотеза")
    H2 --> D2("📦 Independent Data B")
    D2 --> T2("🧪 New Confirmatory Test")
    T2 --> C("📌 Claim")

    classDef hypothesis fill:#e0e7ff,stroke:#6366f1,stroke-width:2px,color:#111827;
    classDef process fill:#dbeafe,stroke:#2563eb,stroke-width:2px,color:#111827;
    classDef evidence fill:#dcfce7,stroke:#16a34a,stroke-width:2px,color:#111827;
    class H1,H2 hypothesis;
    class T1,T2 process;
    class D1,D2,N,C evidence;
```


## 10. Preregistration як засіб розділення плану і результату

*Попередня реєстрація* (preregistration) -- фіксація ключових елементів дослідницького плану до отримання або аналізу результатів.

Для Computer Science вона доречна не в кожному дослідженні однаковою мірою, але корисна там, де є ризик великої кількості дослідницьких ступенів свободи.

Можна зафіксувати:

- primary research question;
- hypotheses;
- dataset / inclusion criteria;
- primary metric;
- baselines;
- split protocol;
- statistical tests;
- stopping rule;
- planned exclusions;
- критерій, за яким Claim буде вважатися підтриманим.

Preregistration не забороняє exploratory analysis. Вона допомагає *чесно відокремити*:

$$
\text{planned confirmatory analysis}
\neq
\text{post hoc exploratory analysis}.
$$


# 11. AI-assisted Research: новий шар методологічного ризику

Інструменти штучного інтелекту можуть бути корисними для:

- пошуку й первинного структурування літератури;
- пояснення коду;
- генерації чернеток програм;
- рефакторингу;
- створення тестів;
- мовного редагування;
- формування варіантів аналітичних процедур;
- аналізу неструктурованих даних;
- роботи як компонента експериментальної системи.

Але використання AI не переносить відповідальність за результат на модель.

Дослідник залишається відповідальним за:

1. достовірність фактів;
2. коректність посилань;
3. коректність коду;
4. відсутність витоку конфіденційних даних;
5. методологічну обґрунтованість;
6. простежуваність суттєвих AI-assisted кроків;
7. коректне декларування використання AI.


## 11.1. Недетермінованість великих мовних моделей (LLM nondeterminism)

Навіть якщо промпт не змінюється, відповідь LLM може залежати від:

- моделі та її версії;
- серверної реалізації;
- sampling configuration;
- прихованих системних змін;
- контексту;
- зовнішніх інструментів;
- оновлень провайдера.

Тому запис

> «аналіз виконано за допомогою ChatGPT»

є методологічно недостатнім, якщо результат LLM впливає на науковий висновок.

Доцільно фіксувати щонайменше:

| Компонент | Що документувати |
|---|---|
| Model | назва/ідентифікатор моделі |
| Date | дата використання |
| Prompt | суттєвий prompt / шаблон |
| Context | які дані або документи були подані |
| Parameters | доступні параметри генерації |
| Output | первинний output або його hash/version |
| Verification | як перевірено результат |
| Human decision | яке рішення прийняв дослідник |


## 11.2. Дрейф моделей (model drift)

Для зовнішньої AI-моделі важливо враховувати, що сервіс у часі може змінитися.

Позначимо відповідь:

$$
Y_t = f(M_t, P, X),
$$

де $M_t$ -- стан моделі або сервісу в момент часу $t$.

Навіть при незмінних $P$ і $X$:

$$
M_{t_1} \neq M_{t_2}
\quad \Rightarrow \quad
Y_{t_1} \neq Y_{t_2}.
$$

Тому для критичного AI-assisted етапу потрібно зберігати не лише prompt, а й *отриманий output*, дату, модель та спосіб перевірки.


## 11.3. Декларація використання ШІ (AI Usage Statement)

У межах цього курсу доцільно підтримувати окремий `AI Usage Statement`.

Приклад структури:

| Етап | AI tool / model | Мета | Вхід | Що використано | Як перевірено | Вплив на Claim |
|---|---|---|---|---|---|---|
| Literature screening | ... | первинне ранжування | titles + abstracts | список кандидатів | ручна перевірка | непрямий |
| Code assistance | ... | генерація unit tests | function spec | тести | запуск + review | технічний |
| Text editing | ... | мовне редагування | авторський текст | стилістичні правки | авторський review | відсутній |

Критичний принцип:

> *Чим сильніше AI-вихід впливає на Evidence або Claim, тим сильнішими мають бути його документування і незалежна перевірка.*



# 14. Reproducibility debt -- «борг відтворюваності»

У програмній інженерії говорять про *technical debt*. Аналогічно в дослідженні накопичується *борг відтворюваності* (reproducibility debt).

Він виникає, коли дослідник відкладає документування:

- «потім згадаю, який був seed»;
- «потім підпишу версію dataset»;
- «потім розберуся, який notebook дав цю таблицю»;
- «потім відновлю параметри»;
- «потім опишу, що саме робив LLM».

З часом вартість відновлення контексту різко зростає.

Тому відтворюваність потрібно проєктувати *до і під час дослідження*, а не оформлювати наприкінці.


# 15. Мінімальний Reproducibility & Traceability Protocol

Для власної дисертації можна зафіксувати протокол із семи шарів.

| Шар | Контрольне питання |
|---|---|
| 1. Data | Чи можна ідентифікувати точну версію вхідних даних? |
| 2. Code | Чи можна ідентифікувати точний стан коду? |
| 3. Environment | Чи можна відновити залежності й обчислювальне середовище? |
| 4. Experiment | Чи зафіксовано конфігурацію, seed, split, сценарій? |
| 5. Result | Чи кожний ключовий результат має ID та походження? |
| 6. Claim | Чи кожний Claim простежується до Evidence? |
| 7. Integrity | Чи задокументовано зміни плану, exclusions, AI assistance та негативні результати? |

Цей протокол не є окремим бюрократичним додатком. Він має бути вбудований у *Methodological Architecture* дисертації.


# 16. Аудит відтворюваності: Red Flags

Під час самоаудиту варто звертати увагу на такі сигнали:

- результат існує лише як число у тексті;
- неможливо встановити, який запуск породив рисунок;
- dataset перезаписується без версій;
- `final.ipynb`, `final2.ipynb`, `final_really_final.ipynb`;
- параметри змінюються без журналу;
- test set переглядається під час розробки;
- повідомлено лише найкращий seed;
- невдалі експерименти видалено;
- є ручні правки даних без протоколу;
- AI-generated код використано без review і тестування;
- бібліографічні посилання від AI не перевірено за першоджерелом;
- Claim не має однозначного Evidence ID.

Жоден окремий пункт автоматично не доводить недоброчесність. Але кожний із них є приводом перевірити *цілісність дослідницького процесу*.


# 17. Відтворюваність як частина доказової архітектури

Повернімося до результату ЛР6:

$$
\text{Claim}
\rightarrow
\text{Evidence}
\rightarrow
\text{Validation}.
$$

Після цієї лекції конструкція стає повнішою:

$$
\text{Claim}
\rightarrow
\text{Evidence}
\rightarrow
\text{Validation}
\rightarrow
\text{Traceability}
\rightarrow
\text{Reproducibility}.
$$

А для особливо важливих тверджень:

$$
\text{Reproducibility}
\rightarrow
\text{Independent Replication}.
$$

Отже, відтворюваність не є «технічним додатком» до доказу. Вона визначає, *наскільки цей доказ доступний для незалежної перевірки*.


# 18. Що має перейти до лабораторної роботи 7

Після лекції аспірант повинен мати концептуальну основу для створення:

1. `Reproducibility Protocol v1`;
2. `Data & Artifact Provenance Map v1`;
3. `Claim Traceability Matrix v1`;
4. `Research Log Template v1`;
5. `AI Usage Statement v1`;
6. `Research Integrity Risk Register v1`.

У ЛР7 ці артефакти мають бути прив'язані не до абстрактного прикладу, а до *власної дисертаційної програми* та результатів ЛР1--ЛР6.


# 19. Контрольні запитання

1. Чим відтворюваність (reproducibility) відрізняється від відтворення незалежною реалізацією (replication)?
2. Чому наявність відкритого коду сама по собі не гарантує відтворюваність дослідження?
3. Що таке data provenance і які елементи потрібно фіксувати?
4. Яку роль Git відіграє у простежуваності дослідження?
5. Чому `random seed` не є повною гарантією відтворюваності?
6. Що таке traceability gap?
7. Навіщо потрібен research log, якщо використовується Git?
8. Чим FAIR відрізняється від повністю відкритих даних?
9. У чому полягає selective reporting?
10. Як виникає p-hacking і чому множинні перевірки збільшують ризик false positive?
11. Що таке HARKing і як коректно працювати з гіпотезою, сформованою post hoc?
12. Які нові ризики відтворюваності виникають у AI-assisted Research?
13. Що потрібно фіксувати, якщо LLM є частиною дослідницького pipeline?
14. Що таке model drift у контексті зовнішніх AI-сервісів?
15. Як Claim Traceability Matrix пов'язує наукове твердження з даними, кодом і валідацією?


# 20. Підсумки

Після цієї лекції важливо зафіксувати сім положень.

1. *Відтворюваність* стосується не лише коду, а всього ланцюга «дані -- середовище -- конфігурація -- обчислення -- результат».
2. *Відтворення незалежною реалізацією* перевіряє стійкість наукового висновку сильніше, ніж повторний запуск авторського коду.
3. *Простежуваність* зв'язує Claim із конкретними Evidence, Experiment, Data та Code.
4. *Дослідницька доброчесність* включає чесне подання негативних, нульових і exploratory результатів.
5. P-hacking, HARKing і selective reporting небезпечні тим, що створюють сильніше враження доказовості, ніж реально забезпечує дизайн.
6. AI-assisted Research потребує окремого документування через недетермінованість, можливий model drift і ризик неперевірених результатів.
7. Відтворюваність потрібно *проєктувати від початку*, а не реконструювати перед поданням дисертації.

Наступний практичний крок -- перетворити ці принципи на конкретний протокол для власного дисертаційного дослідження в ЛР7.


# Рекомендовані джерела

1. National Academies of Sciences, Engineering, and Medicine. *Reproducibility and Replicability in Science*. Washington, DC: The National Academies Press, 2019. DOI: 10.17226/25303.

2. Peng R. D. Reproducible Research in Computational Science. *Science*. 2011. Vol. 334, No. 6060. P. 1226--1227. DOI: 10.1126/science.1213847.

3. Sandve G. K., Nekrutenko A., Taylor J., Hovig E. Ten Simple Rules for Reproducible Computational Research. *PLoS Computational Biology*. 2013. Vol. 9, No. 10. e1003285. DOI: 10.1371/journal.pcbi.1003285.

4. Wilkinson M. D. et al. The FAIR Guiding Principles for scientific data management and stewardship. *Scientific Data*. 2016. Vol. 3. Article 160018. DOI: 10.1038/sdata.2016.18.

5. Nosek B. A. et al. Promoting an open research culture. *Science*. 2015. Vol. 348, No. 6242. P. 1422--1425. DOI: 10.1126/science.aab2374.

6. Munafò M. R. et al. A manifesto for reproducible science. *Nature Human Behaviour*. 2017. Vol. 1. Article 0021. DOI: 10.1038/s41562-016-0021.

7. ACM. *Artifact Review and Badging*. Association for Computing Machinery.

8. Wilson G. et al. Good enough practices in scientific computing. *PLoS Computational Biology*. 2017. Vol. 13, No. 6. e1005510. DOI: 10.1371/journal.pcbi.1005510.

9. Kerr N. L. HARKing: Hypothesizing After the Results are Known. *Personality and Social Psychology Review*. 1998. Vol. 2, No. 3. P. 196--217. DOI: 10.1207/s15327957pspr0203_4.

10. Simmons J. P., Nelson L. D., Simonsohn U. False-Positive Psychology: Undisclosed Flexibility in Data Collection and Analysis Allows Presenting Anything as Significant. *Psychological Science*. 2011. Vol. 22, No. 11. P. 1359--1366. DOI: 10.1177/0956797611417632.

11. Pineau J. et al. Improving Reproducibility in Machine Learning Research: A Report from the NeurIPS 2019 Reproducibility Program. *Journal of Machine Learning Research*. 2021. Vol. 22, No. 164. P. 1--20.

12. Gundersen O. E., Kjensmo S. State of the Art: Reproducibility in Artificial Intelligence. *Proceedings of the AAAI Conference on Artificial Intelligence*. 2018. Vol. 32, No. 1.
