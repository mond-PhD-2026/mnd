

# Тема 2. Від наукової проблеми до архітектури дисертації

## 1. Архітектура дисертації -- це не зміст розділів

На ранньому етапі легко сплутати дві різні структури.

### Структура тексту

Це майбутні розділи, підрозділи, додатки та послідовність викладу.

### Методологічна архітектура

Це система логічних залежностей між:

- науковою проблемою;
- дослідницькою прогалиною;
- об'єктом і предметом;
- метою;
- дослідницькими питаннями;
- гіпотезами;
- задачами;
- дослідженнями;
- evidence;
- claims;
- очікуваними внесками.

Тому:

$$
\boxed{
\text{Methodological Architecture}
\neq
\text{Table of Contents}
}
$$

Зміст дисертації пізніше має *відображати* дослідницьку логіку, а не підміняти її.


## 2. Проблемна ситуація та наукова проблема

### Проблемна ситуація (*problem situation*)

Проблемна ситуація -- це стан предметної області, у якому існує практична, технологічна або пізнавальна суперечність.

Наприклад:

> У соціальних медіа автоматизовані та координовані облікові записи можуть впливати на дискусії, а наявні системи модерації не завжди дозволяють своєчасно й пояснювано виявляти таку активність.

Це важлива мотивація. Але це ще не наукова проблема.

### Наукова проблема (*research problem*)

Наукова проблема фіксує *дефіцит знання*, який перешкоджає поясненню, прогнозуванню, формалізації або обґрунтованому розв'язанню проблемної ситуації.

У спрощеній формі:

$$
\text{Research Problem}
=
\text{Relevant Unknown}
+
\text{Why it Matters}.
$$

Наукова проблема повинна відповідати не лише на питання *"що погано працює?"*, а й на питання:

> *Що саме ми ще недостатньо знаємо, щоб цю ситуацію науково пояснити або обґрунтовано змінити?*


### 2.1. Типова помилка: практична потреба замість проблеми знання

Розглянемо три формулювання.

| Формулювання | Що це фактично означає |
|---|---|
| Потрібно створити систему виявлення ботів | Інженерна задача |
| Існуючі системи мають недостатню точність | Проблемна ситуація / емпіричне спостереження |
| Недостатньо встановлено, які поведінкові ознаки забезпечують раннє та пояснюване розрізнення ботоподібної активності за неповної історії взаємодій | Кандидат на research problem |

У третьому випадку вже видно:

- *невідоме* (*unknown*) -- які ознаки є достатньо інформативними;
- *контекст* (*context*) -- раннє виявлення за неповної історії;
- *вимір якості* (*quality dimension*) -- пояснюваність;
- потенційну можливість побудувати перевірювані claims.

::: callout-important
Наукова проблема не повинна маскувати технічне завдання словами "метод", "модель", "інформаційна технологія" або "удосконалення".

Спочатку потрібно встановити *дефіцит знання*. Лише потім обирається спосіб його усунення.
:::


## 3. Дослідницька прогалина (*research gap*): що саме відсутнє у наявному знанні

*Research gap* (дослідницька прогалина) -- це обґрунтована характеристика того, чого бракує в наявній сукупності наукових знань щодо сформульованої проблеми.

Важлива різниця:

$$
\text{Problem}
\neq
\text{Gap}.
$$

*Problem* задає науково значуще невідоме.  
*Gap* показує, *що саме не забезпечує наявна наукова література* для усунення цього невідомого.

```{mermaid}
%%{init: {'theme': 'base', 'themeVariables': {
  'fontSize': '17px',
  'fontFamily': 'Trebuchet MS, Verdana, Arial, sans-serif',
  'lineColor': '#6366f1'
}}}%%
flowchart LR
    K("📚 Що вже відомо")
    L("⚠️ Обмеження<br/>наявного знання")
    G("🕳️ Research Gap")
    P("❓ Research Problem")
    N("🔬 Потрібне нове<br/>дослідження")

    K --> L --> G
    P --> G
    G --> N

    classDef knowledge fill:#dbeafe,stroke:#2563eb,stroke-width:2px,color:#1e3a8a
    classDef gap fill:#fef3c7,stroke:#d97706,stroke-width:2px,color:#78350f
    classDef research fill:#e0e7ff,stroke:#6366f1,stroke-width:2px,color:#312e81

    class K knowledge
    class L,G,P gap
    class N research
```

На лекції 3 ми перейдемо до систематичного формування доказової бази, тому зараз *Research Gap v1* є робочою гіпотезою про стан поля, а не остаточно доведеним фактом.


### 3.1. Не кожне "ніхто не робив" є дослідницькою прогалиною

Слабке формулювання:

> "У знайдених роботах не використовувалася модель X для dataset Y."

Це лише констатація відсутності конкретної комбінації.

Сильніше питання:

> *Чому відсутність саме цієї комбінації створює суттєвий дефіцит наукового знання?*

Наприклад, застосування нового Transformer до ще одного dataset може бути технічно новим, але не створювати наукового внеску.

Натомість gap може стосуватися:

- відсутності пояснення певного механізму;
- суперечливості результатів;
- відсутності валідації за критично важливих умов;
- систематичного ігнорування певного класу даних;
- невідомих меж застосовності;
- слабкої відтворюваності;
- відсутності порівняння з релевантними альтернативами;
- невирішеної методологічної проблеми вимірювання.


## 4. Об'єкт і предмет дослідження

У дисертаційній практиці об'єкт і предмет часто формулюють формально. Методологічно корисніше розглядати їх як *межі досліджуваної реальності*.

### Об'єкт дослідження

*Object of research* (об'єкт дослідження) -- процес, явище, система або клас систем, у межах яких виникає наукова проблема.

### Предмет дослідження

*Subject of research* (предмет дослідження) -- ті властивості, закономірності, відношення, механізми, методи або моделі в межах об'єкта, які безпосередньо досліджуються.

Отже:

$$
\text{Subject} \subset \text{Object}.
$$

Але це включення є не стільки "словниковим", скільки змістовим: предмет звужує об'єкт до тієї частини, щодо якої будуть сформульовані й перевірені наукові твердження.


### Приклад

Нехай дисертаційний задум стосується раннього виявлення ботоподібної активності в онлайн-дискусіях.

Можлива робоча конструкція:

**Об'єкт:** процеси формування та прояву поведінкової активності учасників онлайн-дискусій.

**Предмет:** методи, моделі та інформативні поведінкові ознаки раннього виявлення й пояснення ботоподібної активності за неповних спостережень.

Тут предмет одразу задає кілька майбутніх вимірів:

- ознаки;
- методи;
- моделі;
- ранність;
- пояснюваність;
- неповнота спостережень.

Кожен із цих вимірів потенційно потребуватиме evidence.


## 5. Мета дослідження

*Research aim* (мета дослідження) описує, який науково значущий результат має бути досягнутий для розв'язання сформульованої проблеми.

Мета не повинна зводитися до:

> "розробити програму", "створити систему", "реалізувати модель".

Такі дії можуть бути частиною дослідження, але вони не пояснюють, *яке нове знання має бути отримане*.

Корисна логічна конструкція:

$$
\text{Problem}
\rightarrow
\text{Aim}
\rightarrow
\text{Expected Knowledge Change}.
$$

Наприклад, замість:

> "Мета -- розробити систему виявлення ботів."

сильніше:

> "Мета -- розробити та експериментально обґрунтувати методи раннього виявлення й пояснення ботоподібної активності за неповних поведінкових спостережень, що дозволяють встановити інформативність окремих груп ознак і межі застосовності відповідних моделей."

Тут розробка є засобом, а не єдиним змістом мети.


## 6. Від мети до дослідницьких питань (*research questions*)

*Дослідницьке питання* (*research question*, RQ) -- це конкретне питання, відповідь на яке має усунути частину невизначеності, закладеної в research problem.

Сильне RQ повинно бути:

- пов'язаним із проблемою;
- дослідним, а не риторичним;
- достатньо конкретним;
- таким, для якого можна визначити evidence;
- таким, відповідь на яке може бути сформульована як scientific claim.

Тому корисний ланцюг:

$$
RQ_i
\rightarrow
Evidence_i
\rightarrow
Claim_i.
$$


### 6.1. Приклад декомпозиції проблеми на RQ

Припустимо, проблема стосується раннього пояснюваного виявлення ботоподібної активності.

Можлива система питань:

**RQ1.** Які поведінкові та часові ознаки зберігають дискримінативність за скороченої історії активності?

**RQ2.** Як обсяг доступної історії впливає на якість раннього виявлення?

**RQ3.** Які класи моделей забезпечують найкращий компроміс між ранністю, якістю класифікації та пояснюваністю?

**RQ4.** Наскільки стійкими є виявлені закономірності між різними дискусіями, часовими періодами або платформами?

Це вже не перелік робіт. Це система невідомих, які потрібно перетворити на обґрунтовані відповіді.


## 7. Дослідницькі питання та гіпотези

Не кожне дослідницьке питання потребує формальної гіпотези.

### Research question

Доцільне, коли дослідження має:

- описовий;
- розвідувальний;
- порівняльний;
- конструктивний;
- пояснювальний характер без наперед заданого напряму ефекту.

### Hypothesis (гіпотеза)

Доцільна, коли можна сформулювати конкретне перевірюване припущення про:

- різницю;
- зв'язок;
- ефект;
- властивість;
- причинний механізм;
- перевагу за визначених умов.

Наприклад:

$$
H_1:
F1_{\text{behavioral+temporal}}
>
F1_{\text{behavioral}}
$$

за фіксованого раннього вікна спостереження та визначеного протоколу оцінювання.

Тут гіпотеза вже вимагає операціоналізації всіх компонентів.


### 7.1. Гіпотеза -- не бажаний результат

Слабка конструкція:

> "Запропонований метод буде кращим за існуючі."

Невідомо:

- за якою метрикою;
- за яких даних;
- порівняно з якими baseline;
- наскільки кращим;
- чи йдеться про статистичну або практичну значущість;
- за яких умов твердження перестає бути справедливим.

Сильніша гіпотеза повинна бути *спростовуваною* (*falsifiable*).

Тобто має існувати можливий результат дослідження, за якого ми чесно скажемо:

> **Гіпотеза не отримала достатньої емпіричної підтримки.**

::: callout-tip
Якщо гіпотезу неможливо програти, це, швидше за все, не наукова гіпотеза, а декларація наміру.
:::


Одна задача може підтримувати кілька тверджень. Одне твердження також може вимагати кількох задач і кількох окремих досліджень.


### 8.1. Як відрізнити research task від технічної операції

Порівняємо.

| Формулювання | Рівень |
|---|---|
| Зібрати dataset | Технічна операція |
| Реалізувати модель | Технічна операція |
| Побудувати графіки | Технічна операція |
| Визначити групи ознак, що зберігають інформативність за скорочення вікна спостереження | Дослідницька задача |
| Розробити метод раннього виявлення та встановити його межі застосовності | Дослідницька задача |
| Експериментально оцінити стійкість висновків до зміни даних і протоколу | Дослідницька задача |

Технічні операції нікуди не зникають. Вони просто є *засобами виконання дослідницьких задач*.


## 9. Від задач до окремих досліджень (*studies*)

У лекції 1 ми розглядали дисертацію як *research programme*.

Тепер можемо деталізувати:

$$
\text{Research Programme}
=
\{Study_1, Study_2, \ldots, Study_k\}.
$$

Кожне *окреме дослідження* (*study*) повинно мати:

1. конкретне питання або гіпотезу;
2. дизайн;
3. дані або інший матеріал дослідження;
4. процедуру аналізу;
5. критерії оцінювання;
6. evidence;
7. локальне наукове твердження;
8. обмеження.

Дисертаційний внесок формується не тому, що studies багато, а тому, що їхні результати *логічно накопичуються*.


## 10. Очікувані наукові внески

На початку PhD ми ще не можемо чесно записати остаточні внески як встановлені факти.

Але можемо сформувати *очікуваний науковий внесок v1* (*Expected Contribution v1*) -- робочу модель того, який тип нового знання має виникнути.

Наприклад:

- встановлена закономірність;
- новий метод із перевіреними властивостями;
- модель із визначеними межами застосовності;
- нова формалізація;
- емпірично підтверджений механізм;
- новий benchmark із методологічно обґрунтованою процедурою;
- архітектура з підтвердженим компромісом між характеристиками;
- негативний результат, що коригує наявне припущення.

Важливо:

$$
\text{Expected Contribution}_{v1}
\neq
\text{Proven Contribution}.
$$


### 10.1. Внесок потрібно формулювати через claims

Слабко:

> "Розроблено новий метод."

Сильніше:

> "Запропонований метод дозволяє зберігати визначений рівень якості раннього виявлення при скороченні доступної історії спостережень порівняно з базовими підходами за умов $C$."

У другому випадку з'являються:

- об'єкт порівняння;
- вимірювана властивість;
- умови;
- потенційний протокол перевірки;
- можливість спростування.

Тому:

$$
\text{Contribution}
\leftarrow
\text{Supported Claims}.
$$


## 11. Карта майбутніх тверджень (*Claim Map*)

*Карта тверджень* (*Claim Map*) -- це попередня структура тверджень, які дисертація потенційно повинна мати право захищати після завершення досліджень.

На старті це версія `v1`.

Наприклад:

```{mermaid}
%%{init: {'theme': 'base', 'themeVariables': {
  'fontSize': '16px',
  'fontFamily': 'Trebuchet MS, Verdana, Arial, sans-serif',
  'lineColor': '#6366f1'
}}}%%
flowchart TD
    C0("🌱 C0: інтегральний<br/>Contribution Claim")
    C1("💬 C1: інформативність<br/>груп ознак")
    C2("💬 C2: вплив<br/>вікна спостереження")
    C3("💬 C3: властивості<br/>запропонованого методу")
    C4("💬 C4: стійкість<br/>і межі узагальнення")

    C1 --> C0
    C2 --> C0
    C3 --> C0
    C4 --> C0

    classDef local fill:#dbeafe,stroke:#2563eb,stroke-width:2px,color:#1e3a8a
    classDef main fill:#dcfce7,stroke:#16a34a,stroke-width:2px,color:#14532d

    class C1,C2,C3,C4 local
    class C0 main
```

## 12. Логічна узгодженість компонентів

Сильна архітектура дисертації повинна бути *простежуваною* (*traceable*).

Для кожного компонента можна поставити контрольне питання:

| Компонент | Контрольне питання |
|---|---|
| Problem | Яке науково значуще невідоме усуваємо? |
| Gap | Чого саме бракує наявному знанню? |
| Object | У межах якого процесу / явища існує проблема? |
| Subject | Яку частину об'єкта безпосередньо досліджуємо? |
| Aim | Який науковий результат має усунути проблему? |
| RQ | На які конкретні невідомі треба відповісти? |
| Hypothesis | Які перевірювані припущення висуваємо? |
| Task | Яке дослідження необхідно виконати? |
| Claim | Яке твердження матимемо право сформулювати? |
| Contribution | Що це додає до знання? |

Якщо компонент не має логічного зв'язку з попереднім або наступним, виникає *методологічно ізольований елемент* (*methodological orphan*) -- "сирітський" елемент дослідження.


### 12.1. Тест двонапрямної простежуваності

Перевірка має працювати в обидва боки.

### Пряма простежуваність (*forward traceability*)

$$
Problem
\rightarrow
RQ
\rightarrow
Task
\rightarrow
Evidence
\rightarrow
Claim
\rightarrow
Contribution.
$$

Питаємо: *чи кожний наступний елемент випливає з попереднього?*

### Зворотна простежуваність (*backward traceability*)

$$
Contribution
\rightarrow
Claim
\rightarrow
Evidence
\rightarrow
Task
\rightarrow
RQ
\rightarrow
Problem.
$$

Питаємо: *чи кожний заявлений внесок можна простежити назад до конкретних доказів і вихідної проблеми?*

::: callout-important
Якщо внесок не можна простежити до evidence, це декларація.

Якщо задача не простежується до research question, це, ймовірно, зайва або суто технічна робота.
:::


## 13. Матриця узгодженості

На практиці архітектуру зручно перевіряти не лише схемою, а й таблицею.

| RQ | Hypothesis | Research task | Expected evidence | Candidate claim | Contribution |
|---|---|---|---|---|---|
| RQ1 | H1 | Task 1 | Порівняльні результати для груп ознак | C1 | Частина Contribution A |
| RQ2 | H2 | Task 2 | Залежність якості від вікна | C2 | Частина Contribution A |
| RQ3 | -- | Task 3 | Benchmark + аналіз компромісів | C3 | Contribution B |
| RQ4 | H3 | Task 4 | Cross-context validation | C4 | Межі A і B |

Це ще не остаточна *Claim--Evidence--Validation Matrix*, яку ми побудуємо пізніше. На ЛР2 нас цікавить передусім *логічна структура*.


## 14. Типові дефекти методологічної архітектури

### Дефект 1. Невідповідність між проблемою та метою (*Problem--Aim mismatch*)

Проблема стосується невідомої закономірності, а мета -- лише створення програмного продукту.

### Дефект 2. Невідповідність між метою та задачами (*Aim--Task mismatch*)

У меті заявлено пояснюваність, але жодна задача її не досліджує.

### Дефект 3. Невідповідність між задачею та твердженням (*Task--Claim mismatch*)

Виконано порівняльний експеримент, але claim стосується причинного механізму.

### Дефект 4. Невідповідність між доказами та твердженням (*Evidence--Claim mismatch*)

Один dataset використано для твердження про універсальність методу.

### Дефект 5. Необґрунтоване завищення внеску (*Contribution inflation*)

Локальне покращення метрики формулюється як "новий універсальний метод".

### Дефект 6. Ізольована задача (*Orphan task*)

Задача є в плані, але не підтримує жодного RQ або внеску.

### Дефект 7. Ізольований внесок (*Orphan contribution*)

Заявлений внесок не має відповідної задачі та evidence.


## 15. Приклад: від проблеми до архітектури

Розглянемо спрощений приклад.

### Problem v1

Недостатньо встановлено, які поведінкові характеристики дозволяють надійно та пояснювано виявляти ботоподібну активність на ранніх етапах онлайн-дискусії за неповної історії взаємодій.

### Gap v1

Наявні підходи часто оцінюються на повних історіях активності; недостатньо систематично досліджено деградацію якості при скороченні спостереження та стійкість інформативності ознак.

### Aim v1

Розробити й експериментально обґрунтувати методологію раннього виявлення ботоподібної активності за неповних поведінкових спостережень та встановити інформативність ознак, компроміси якості й межі застосовності моделей.

### RQ

- RQ1 -- які ознаки залишаються інформативними?
- RQ2 -- як ранність впливає на якість?
- RQ3 -- який компроміс між ранністю, якістю та пояснюваністю?
- RQ4 -- наскільки висновки стійкі між контекстами?

Тепер архітектура вже може бути перевірена на внутрішню узгодженість.


## 16. Архітектура не повинна передбачати позитивний результат

Припустимо, H1 стверджує, що певна група часових ознак покращує раннє виявлення.

Можливі результати:

1. ефект підтверджено;
2. ефект не підтверджено;
3. ефект існує лише за певних умов;
4. ефект статистично помітний, але практично малий;
5. результат нестійкий між наборами даних (*datasets*).

Усі п'ять результатів можуть бути науково інформативними.

Тому якісна архітектура повинна визначати:

- *що перевіряємо*;
- *як перевіряємо*;
- *який evidence буде достатнім*;

але не вимагати наперед заданого "успішного" результату.

$$
\text{Якісне дослідження}
\neq
\text{Позитивний результат}.
$$


## 17. Архітектура як система версій

На першому семестрі PhD компоненти повинні мати версійність:

- `Research Problem v1`;
- `Research Gap v1`;
- `Object / Subject v1`;
- `Aim v1`;
- `RQ System v1`;
- `Hypotheses v1`;
- `Tasks v1`;
- `Claim Map v1`;
- `Expected Contributions v1`.

Після лекцій 3--4 і ЛР3--ЛР4 значна частина цієї архітектури буде переглянута на основі систематичної доказової бази.

Це принципово:

> *Зміна формулювання після отримання нових evidence є ознакою розвитку дослідницької програми, а не методологічною помилкою.*

Помилкою було б приховати таку зміну або створити враження, що остаточна конструкція існувала від самого початку.

# Висновки

1. *Методологічна архітектура дисертації не є змістом її розділів.* Вона описує логіку переходу від наукової проблеми до доказів, тверджень і внеску.

2. *Problem situation і research problem потрібно розрізняти.* Практична складність або потреба в новій системі ще не визначає дефіцит наукового знання.

3. *Research gap конкретизує обмеження наявного знання*, але на ранньому етапі `Gap v1` є робочою конструкцією, яку потрібно перевірити систематичним аналізом літератури.

4. *Об'єкт і предмет задають межі дослідження.* Предмет визначає ту частину об'єкта, щодо якої формуються і перевіряються наукові твердження (*scientific claims*).

5. *Мета повинна бути спрямована на отримання науково значущого результату*, а не лише на створення технічного артефакту.

6. *Research questions декомпозують проблему на конкретні невідомі.* Гіпотези використовуються там, де можливе чітке перевірюване й спростовуване припущення.

7. *Research tasks -- це кроки отримання знання.* Технічні операції є засобами їх виконання.

8. *Claim Map v1* дозволяє ще до експериментів побачити, які твердження дисертація потенційно повинна мати право захищати.

9. *Логічна узгодженість перевіряється двонапрямно*: від problem до contribution і від contribution назад до конкретних evidence та research questions.

10. Архітектура першого семестру є *версією*, а не остаточною конструкцією. Після систематичного аналізу літератури вона повинна бути критично переглянута.
